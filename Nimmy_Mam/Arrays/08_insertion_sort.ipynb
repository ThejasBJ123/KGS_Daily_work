# **Insertion Sort**


## **Question 1: Insertion Sort in Ascending Order**
Write a Python program to sort a given user-defined integer array using **Insertion Sort** in **ascending order**.


```python
def createIntArray():
    l1 = []
    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)
        except Exception as e:
            return l1

def insertionSortAsc(arr):
    n = len(arr)
    for i in range(0, n - 1):
        for j in range(i + 1, 0, -1):
            if arr[j - 1] > arr[j]:
                arr[j - 1], arr[j] = arr[j], arr[j - 1]
    return arr

# Main program execution
print("Enter the elements into the created array:")
arr = createIntArray()
print("The created array:", arr)

res = insertionSortAsc(arr)
print("Insertion Sort (Ascending):", res)
```

## **Question 2: Insertion Sort in Descending Order**
Write a Python program to sort a given user-defined integer array using **Insertion Sort** in **descending order**.


```python
def insertionSortDesc(arr):
    n = len(arr)
    for i in range(0, n - 1):
        for j in range(i + 1, 0, -1):
            if arr[j - 1] < arr[j]:
                arr[j - 1], arr[j] = arr[j], arr[j - 1]
    return arr

# Main program execution
print("Enter the elements into the created array:")
arr = createIntArray()
print("The created array:", arr)

res = insertionSortDesc(arr)
print("Insertion Sort (Descending):", res)
```
