# **User-Defined Array Creation**

Demonstrates how to dynamically take user inputs to populate integer, string, and character arrays using input loops and exception handling.


## **Question 1: Create a User-Defined Integer Array**
Write a Python program to take integer inputs from the user continuously until a non-integer input (or empty input) is entered, and return the populated integer array.


```python
# WAP to create a user-defined integer array using try-except

def createIntArray():
    l1 = []
    while True:
        try:
            num = int(input("Enter the number: "))
            l1.append(num)
        except Exception as e:
            return l1

print("Enter the elements in the array (enter a non-integer to stop):")
arr = createIntArray()
print("The created integer array is:", arr)
```

## **Question 2: Create a User-Defined String Array**
Write a Python program to take string inputs from the user continuously until an empty string is entered, and return the populated string array.


```python
# WAP to create a user-defined string array

def createStringArray():
    l1 = []
    while True:
        val = input("Enter the string: ")
        if val == "":
            return l1
        l1.append(val)

print("Enter the elements in the array (press Enter on an empty line to stop):")
arr = createStringArray()
print("The created string array is:", arr)
```

## **Question 3: Create a User-Defined Character Array**
Write a Python program to take input strings from the user and store only their first character in an array until an empty input is entered.


```python
# WAP to create a user-defined character array

def createCharArray():
    l1 = []
    while True:
        val = input("Enter the string: ")
        if val == "":
            return l1
        l1.append(val[0])

print("Enter the elements in the array (press Enter on an empty line to stop):")
arr = createCharArray()
print("The created character array is:", arr)
```
