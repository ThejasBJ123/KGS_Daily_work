### Exception: 
- An unexpected even that acocours during exception time. It is very much requried to handle an exception as it " disrupts the exectution flow" of the program.

* exception can be handled by using " try - except" block
* **try:** It includes the lines of code which misht throw an exception, along wiht it the associated lines of code are also included 
* **except:** It includes the handling lines of code that needs to be executed when an Exception is created in the  try block


```python
# WAP to create a user-defined integer array

def createInArray():
    l1 = []

    while True:
        try:
            num = int(input("Enter the number: "))
            l1.append(num)

        except Exception as e:
            return l1


print("Enter the elements in the array:")
arr = createInArray()
print("The created array is:", arr)
```

```python
# WAP to create a user-defined string array

def createInArray():
    l1 = []

    while True:
        val = input("Enter the string: ")
        if val == "":
            return l1
        l1.append(val)

print("Enter the elements in the array:")
arr = createInArray()
print("The created array is:", arr)
```

### linear search

```python
def createInArray():
    l1 = []

    while True:
        try:
            num = int(input("Enter the number: "))
            l1.append(num)

        except Exception as e:
            return l1


def linearSearch(arr, target):
    for i in range(0, len(arr)):
        if target == arr[i]:
            return i, True

    return -1, False


print("Enter the elements in the array:")
arr = createInArray()

print("The created array is:", arr)

target = int(input("Enter the element to search: "))

resInd, flag = linearSearch(arr, target)

if flag:
    print(target, "element is found at index", resInd)
else:
    print(target, "element is not found")
```
