# **Arrays/Lists**

- It is data strufture i.e., helps to store multiple val's in a structured format;  assigns the val's in contiguous memory location
- It is also an iterable object 
- It is a collecions of data
- It can hold both homogeneous or heterogeneous data 
- It is represented by ==> "[]"
- Empty "[]" ==> empty list , each empty list will get allocated by ta separte memory 
- It is an object therfore the memory will get allocated in heap segment 
- IT is a mutable object , any changes applied on a list , will  get implemented in a same memory 
- It can hold both primitives and non- primitive val's 
- It hold the data in insertion order
- Each and every data can be accessed individually by using index

### **index** :
    - It is a n interger that is assigned by PVM  during insertion of data at execution time 
    - Programmer can only access / use an index but can never update or modify on index.
    - The initial index of an array ==>0
    - The last index of an array  ==> len(arr) -1
    - The counting of index starts from 0
    - In Python, both +ve and -ve indexing are supported 
    - If we try to access an index that is not available  in the created array the PVM will immediately throws " IndexError" 

### **len()**
    - It is a pre-defined function that counts to the number of elements presents in a iterable object 
    - Syntax: 
        - len (iterable_obj)
    - It returns whole number
