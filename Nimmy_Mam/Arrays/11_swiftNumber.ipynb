theer are 2 variaties of shifting 

    1. left shift
    2. right shift

    

```python
#left shift  k= only less than n
def createIntArray():
    l = []

    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except:
            return l

def reverseArray(arr, i, j):
    while i < j:
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1


def leftShift(arr, k):
    n = len(arr)

    # Step 1: Reverse first k elements
    reverseArray(arr, 0, k - 1)

    # Step 2: Reverse remaining elements
    reverseArray(arr, k, n - 1)

    # Step 3: Reverse complete array
    reverseArray(arr, 0, n - 1)

    return arr


print("Enter the elements:")
arr = createIntArray()

k = int(input("Enter shift: "))

print("Original array:", arr)
print("Left shifted array:", leftShift(arr, k))
```

```python
# left swifting  k =k% 2
def createIntArray():
    l = []

    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except:
            return l

def reverseArray(arr, i, j):
    while i < j:
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1


def leftShift(arr, k):
    n = len(arr)
    if k>=n:
        k=k%n

    # Step 1: Reverse first k elements
    reverseArray(arr, 0, k - 1)

    # Step 2: Reverse remaining elements
    reverseArray(arr, k, n - 1)

    # Step 3: Reverse complete array
    reverseArray(arr, 0, n - 1)

    return arr


print("Enter the elements:")
arr = createIntArray()

k = int(input("Enter shift: "))

print("Original array:", arr)
print("Left shifted array:", leftShift(arr, k))
```

```python
# right swifting 
def createIntArray():
    l = []

    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except:
            return l


def reverseArray(arr, i, j):
    while i < j:
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1


def rightShift(arr, k):
    n = len(arr)

    k = k % n

    # Step 1: Reverse complete array
    reverseArray(arr, 0, n - 1)

    # Step 2: Reverse first k elements
    reverseArray(arr, 0, k - 1)

    # Step 3: Reverse remaining elements
    reverseArray(arr, k, n - 1)

    return arr


print("Enter the elements:")
arr = createIntArray()

k = int(input("Enter shift: "))

print("Original array:", arr)

rightShift(arr, k)

print("Right shifted array:", arr)
```
