# **Linear Search in an Array**

## **Question:**
Write a Python program to create a user-defined integer array and check whether a given target element is available in the array using **Linear Search**. Return the index of the element and a boolean status indicating whether it was found.


```python
def createIntArray():
    l1 = []
    while True:
        try:
            num = int(input("Enter the number: "))
            l1.append(num)
        except Exception as e:
            return l1

def linearSearch(arr, target):
    for i in range(0, len(arr)):
        if target == arr[i]:
            return i, True
    return -1, False

# Main program execution
print("Enter the elements in the array:")
arr = createIntArray()
print("The created array is:", arr)

target = int(input("Enter the element to search: "))
resInd, flag = linearSearch(arr, target)

if flag:
    print(target, "element is found at index", resInd)
else:
    print(target, "element is not found")
```
