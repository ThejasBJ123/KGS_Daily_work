# **Arrays / Lists in Python**

### **Concept Overview:**
- An **Array/List** is a data structure that helps to store multiple values in a structured format; assigns values in contiguous memory locations.
- It is an **iterable object**.
- It is a collection of data and can hold both **homogeneous** or **heterogeneous** data.
- It is represented by `[]`. An empty `[]` creates an empty list (each empty list gets allocated separate memory).
- It is an object, therefore memory is allocated in the **heap segment**.
- It is a **mutable** object; any changes applied to a list get implemented in the same memory location.
- It can hold both primitives and non-primitives.
- It preserves insertion order.
- Each and every element can be accessed individually using an **index**.

### **Indexing Rules:**
- Index is an integer assigned by PVM during insertion of data at execution time.
- Programmer can access/use an index but cannot directly modify index assignment.
- Initial index: `0`
- Last index: `len(arr) - 1`
- In Python, both positive (`+ve`) and negative (`-ve`) indexing are supported.
- Accessing an index out of bounds raises `IndexError`.

### **`len()` Function:**
- Pre-defined function that counts the number of elements in an iterable object.
- **Syntax**: `len(iterable_obj)`
- Returns a whole number (non-negative integer).


```python
# Basic list creation and append, insert, extend operations
arr = []

print("Initial empty array:", arr)
print("Length:", len(arr))

# Appending elements
arr.append(100)
arr.append(-5)
arr.append("abc")
arr.append({1, 2, 3})
arr.append(100)

print("Array after appends:", arr)
print("Length:", len(arr))
print("Length of set at index 3:", len(arr[3]))
print("Length of string at index 2:", len(arr[2]))

# Insert operations
arr.insert(2, 36.5)
print("After insert(2, 36.5):", arr)

arr.insert(4, 255)
print("After insert(4, 255):", arr)

# Modifying element at index 5
arr[5] = 75
print("After arr[5] = 75:", arr)

# Insert at index beyond current length
arr.insert(15, 85)
print("After insert(15, 85):", arr)

# Extend operation
arr.extend([7, 9, 14])
print("After extend([7, 9, 14]):", arr)
```

```python
# Nested lists demonstration
arr2 = [11, 22]
print("arr2:", arr2)
arr3 = [36, 46]
print("arr3:", arr3)

arr2.append(arr3)
print("arr2 after appending arr3:", arr2)
print("arr3:", arr3)

# Modifying inner list updates reference inside arr2
arr3[1] = 66
print("arr2 after modifying arr3[1]:", arr2)
print("arr3:", arr3)
```

```python
# Demonstration of type() and len() across different data types
l1 = [1, 2, 4, 78]
print(type(l1), len(l1))

s1 = "aslsfdsklds"
print(type(s1), len(s1))

s2 = {1, 3, 4, 4, 5, 4, 23, 4}
print(type(s2), len(s2))

dict1 = {1021: "helld", 1022: "hil", 1021: "helld"}
print(type(dict1), len(dict1))

val1 = 123433
print(type(val1))
# Note: len(val1) will raise TypeError because integer has no length
```
