# **Reverse an Array**


## **Question:**
Write a Python program to reverse a given array using:
1. **Logic 1**: Creating a new reversed array by iterating backwards.
2. **Logic 2**: Reversing the array in-place using two pointers.


```python
# Logic 1: Reversing by creating a new array

def createIntArray():
    l = []
    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except Exception as e:
            return l

def reverseArray(arr):
    res = []
    for i in range(len(arr) - 1, -1, -1):
        res.append(arr[i])
    return res

# Main program execution
print("Enter the elements into the array:")
arr = createIntArray()
print("The original array:", arr)

res = reverseArray(arr)
print("The reversed array (Logic 1):", res)
```

```python
# Logic 2: In-place array reversal using two pointers

def reverseArray2(arr):
    i, j = 0, len(arr) - 1
    while i < j:
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1
    return arr

# Main program execution
print("Enter the elements into the array:")
arr = createIntArray()
print("The original array:", arr)

res = reverseArray2(arr)
print("The reversed array (Logic 2 - In place):", res)
```

```python
# reverse logic 3
def createIntArray():
    l = []

    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except:
            return l


def arraySum(arr):
    total = 0

    for i in arr:
        total = total + i

    return total


def reverseArray3(arr):
    n = len(arr)

    for i in range(0, n // 2):
        arr[i], arr[n - 1 - i] = arr[n - 1 - i], arr[i]

    return arr


print("Enter the elements into the array:")
arr = createIntArray()

print("Created array:", arr)
print("Sum:", arraySum(arr))
print("Reversed array:", reverseArray3(arr))
```
