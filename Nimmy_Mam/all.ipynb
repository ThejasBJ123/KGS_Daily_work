### palindrome

```python
def palindrome(n):
    temp=n
    if n<0:
        n=n*-1

    rev=0
    while n>0:
        base=n%10
        rev=(rev*10)+base
        n=n//10
        
    if temp<0:
        rev=rev*-1

    return rev==temp

num=int(input("enter the num: "))
if palindrome(num):
    print(" palindrome ", num)
else:
    print("not Palindrome", num)
```

### GCD/hcf

```python
def gcd(n1,n2):
    lower=n1
    if n1>n2:
        lower=n2

    hcf=1
    for i in range(2,(lower+1)):
        if n1%i==0 and n2%i==0:
            hcf=i
    return hcf

f_num =int(input("enter the first number : "))
s_num=int(input("Enter the second number : "))
  

res= gcd(f_num, s_num)
print( " the hcf of ",f_num,"and ",s_num," is: ", res)
```

### happy number

```python
def happyNum(n):

    if n == 1:
        return True
    elif n == 4:
        return False

    sum = 0

    while n > 0:
        base = n % 10
        sum = sum + (base * base)
        n = n // 10

    return happyNum(sum)


n = int(input("Enter a num: "))

if happyNum(n):
    print(n, "is a happy num")
else:
    print(n, "is not a happy num")
```
