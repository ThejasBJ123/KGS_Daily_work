# NUMBER PATTERNS 

```python
n=int(input("enter the number : "))
noc=1
for i in range(1,n*2):
    for j in range(1,noc+1):
        print(j,end=" ")
    print()
    if i<n:
        noc+=1
    else:
        noc-=1
```

```python
n = int(input("Enter the number: "))

noc = 1

for i in range(1, n * 2):

    for k in range(n, noc, -1):
        print(" ", end="")
    for j in range(n, n - noc, -1):
        print(j, end="")

    print()

    if i < n:
        noc += 1
    else:
        noc -= 1
```

```python
n=int(input("Enter the number : "))
noc=1
for i in range(1,n*2):
    for k in range(n, noc,-1):
        print(" ",end="")
    for j in range(noc , noc*2):
        print(chr(96+j),end=" ")

    print()
    if i<n:
        noc+=1
    else:
        noc-=1
```
