# Solid Star Patterns

A **Star Pattern** is an arrangement of stars (`*`) or numbers formed by nested loops representing **rows** ($i$) and **columns** ($j$).

### Pattern Building Fundamentals:
1. **Rows ($i$ loop):** Controls height.
2. **Columns ($j$ loop):** Controls number of elements per row.
3. **Spaces ($k$ loop):** Controls explicit leading spaces.
4. **Types of Patterns:** Non-combinational ($n$ rows) and Combinational ($2n-1$ rows).


### Question 1: Vertical Line Star Pattern
**Problem Statement:**
Write a Python program to print $n$ stars vertically.

**Sample Input:** $n = 5$
**Expected Output:**
```text
*
*
*
*
*
```


````python
# Vertical Line Pattern (Stars)
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    print("*")
````

### Question 16: Solid Half-Diamond LHS (Right-Pointing Arrow)
**Problem Statement:**
Write a Python program to print a solid right-pointing arrow / half-diamond pattern on LHS for $n$ rows.

**Sample Input:** $n = 5$
**Expected Output:**
```text
*
**
***
****
*****
****
***
**
*
```


````python
# Solid Half-Diamond LHS (Right-Pointing Arrow)
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for j in range(1, noc + 1):
        print("*", end="")
    print()
    if i < n:
        noc += 1
    else:
        noc -= 1
````

### Question 24: Solid Butterfly Pattern
**Problem Statement:**
Write a Python program to print a solid butterfly pattern for size $n$.

**Sample Input:** $n = 5$
**Expected Output:**
```text
*               * 
* *           * * 
* * *       * * * 
* * * *   * * * * 
* * * * * * * * * 
* * * *   * * * * 
* * *       * * * 
* *           * * 
*               * 
```


````python
# Solid Butterfly Pattern
n = int(input("Enter the number: "))
noc = 1
nor = (n * 2) - 1
for i in range(1, n * 2):
    for j in range(1, n * 2):
        if j <= noc or j >= nor:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
    if i < n:
        noc += 1
        nor -= 1
    else:
        noc -= 1
        nor += 1
````
