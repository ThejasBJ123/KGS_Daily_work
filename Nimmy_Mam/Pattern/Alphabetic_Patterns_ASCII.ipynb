# Alphabetic Patterns

## Python File → PVM → Microprocessor

```text
Python File (HLL)
        ↓
Compiler
        ↓
Bytecode
        ↓
PVM (Python Virtual Machine)
        ↓
Machine-level execution
        ↓
Microprocessor
```

- **Compiler** → Converts Python source code into **Bytecode**.
- **PVM (Python Virtual Machine)** → Executes the Python bytecode.
- The computer ultimately executes instructions at the machine level using binary representations.

> **Note:** It is more accurate to say that the PVM executes bytecode rather than saying that the interpreter directly converts bytecode into binary language.

## ASCII Values

**ASCII** stands for **American Standard Code for Information Interchange**.

ASCII assigns a numerical value to characters. Python can obtain the Unicode code point of a character using the `ord()` function.

### Important ASCII Values

| Character | ASCII Value / Range |
|---|---:|
| Space `' '` | 32 |
| `'0'` to `'9'` | 48 to 57 |
| `'A'` to `'Z'` | 65 to 90 |
| `'a'` to `'z'` | 97 to 122 |

### ASCII Range

Standard ASCII contains **128 characters**, with values from **0 to 127**.

The commonly mentioned **0 to 255** range refers to 8-bit character sets often called extended ASCII, but there is no single universal extended-ASCII standard.

### Unicode

**Unicode** is a much larger character system that supports characters from many languages and writing systems. Python 3 strings use Unicode by default.

````python
# ASCII / Unicode values using ord()

print(ord(' '))
print(ord('0'))
print(ord('9'))
print(ord('A'))
print(ord('Z'))
print(ord('a'))
print(ord('z'))
````

````python
# Display uppercase alphabets using ASCII/Unicode code points

for i in range(65, 91):
    print(chr(i), end=' ')
print()
````

````python
# Display lowercase alphabets using ASCII/Unicode code points

for i in range(97, 123):
    print(chr(i), end=' ')
print()
````

## `ord()` and `chr()` Functions

- `ord(character)` → Returns the numeric Unicode code point of a character.
- `chr(number)` → Returns the character represented by a Unicode code point.

Example:

```python
print(ord('A'))   # 65
print(chr(65))   # A
```

These functions are useful when creating **alphabetic patterns** in Python.

### Logic:

* UpperCase ==> (64 + val)
* LowerCase ==> (96 + val)
