## Section 1: x =int(input("enter the number"))

### ❓ Question / Problem Statement
**Problem Statement (x =int(input("enter the number"))): Implement and trace the Python program for x =int(input("enter the number")).**

--- 
### 💻 My Code

````python
x =int(input("enter the number"))
def printInDecreasing(x):
    # code here
    while (x >= 0):
        print(x,end=" ")
        x -= 1

printInDecreasing(x)
````

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `printInDecreasing` to encapsulate reusable task logic.
2. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.
3. **While Loop Control:** Continues execution while state control condition evaluates to `True`.

### 🔍 Step-by-Step Code Tracing (Dry Run)

#### Sample Execution: `x = 3`

**Initially:**
- `x = 3`

**Loop Execution (`while x >= 0`):**
- **Cycle 1 (`x = 3`):** `3 >= 0` is **True**. Print `3 `. `x -= 1` $\rightarrow$ `x = 2`.
- **Cycle 2 (`x = 2`):** `2 >= 0` is **True**. Print `2 `. `x -= 1` $\rightarrow$ `x = 1`.
- **Cycle 3 (`x = 1`):** `1 >= 0` is **True**. Print `1 `. `x -= 1` $\rightarrow$ `x = 0`.
- **Cycle 4 (`x = 0`):** `0 >= 0` is **True**. Print `0 `. `x -= 1` $\rightarrow$ `x = -1`.

**Termination:**
- `-1 >= 0` is **False**. Loop terminates.

**Output:**
`3 2 1 0 `

---

#### 💡 The Main Idea / Intuition to Remember

```
Start x = 3
Print 3 → x = 2
Print 2 → x = 1
Print 1 → x = 0
Print 0 → x = -1 (STOP)
```

### 💻 My Code

````python
x = int(input("enter the number: "))
def printIncreasingPower(x):
    n=1
    #code here
    # Loop to jump in powers of 2
    while(n**2<=x):
        
        #code here
        
        print (n**2 , end = " ")
        n=n+1

printIncreasingPower(x)
````

### 🧠 Logic & Explanation

### 🔍 Step-by-Step Code Tracing (Dry Run)

**Output:**


## Section 3: def cal_gst(price, gst_p):

### ❓ Question / Problem Statement
**Problem Statement (def cal_gst(price, gst_p):): Implement and trace the Python program for def cal_gst(price, gst_p):.**

--- 
### 💻 My Code

````python
def cal_gst(price, gst_p):
    gst_a=(price*gst_p)/100
    final=gst_a+price

    return final


p=int(input(" enter the price of the product : "))
g=int(input(" enter the gst percentage: "))

amount=cal_gst(p,g)
print("final amount: ", amount)
````

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `cal_gst` to encapsulate reusable task logic.
2. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.
3. **Return Statement:** Passes calculated result back to caller function.

### 🔍 Step-by-Step Code Tracing (Dry Run)

#### Sample Execution: `price = 1000, gst_p = 18`

**Initially:**
- `price = 1000`, `gst_p = 18`

**Calculation (`cal_gst(1000, 18)`):**
- `gst_a = (1000 * 18) / 100` $\rightarrow$ `180.0`
- `final = 180.0 + 1000` $\rightarrow$ `1180.0`
- `return 1180.0`

**Output:**
`final amount:  1180.0`

---

#### 💡 The Main Idea / Intuition to Remember

```
Base Price: 1000
GST 18%:   (1000 * 18) / 100 = 180
               ↓
Total = 1000 + 180 = 1180.0
```

## Section 4: def cal_bmi(weight, height):

### ❓ Question / Problem Statement
**Problem Statement (def cal_bmi(weight, height):): Implement and trace the Python program for def cal_bmi(weight, height):.**

--- 
### 💻 My Code

````python
def cal_bmi(weight, height):
    Bmi=weight/(height*height)
    return Bmi

w=float(input("enter the weight of the body (Kg): "))
h=float(input("enter the height [meter] : "))

result=cal_bmi(w,h)

print("output :", result)


````

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `cal_bmi` to encapsulate reusable task logic.
2. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.
3. **Return Statement:** Passes calculated result back to caller function.

### 🔍 Step-by-Step Code Tracing (Dry Run)

#### Sample Execution: `weight = 70.0, height = 1.75`

**Initially:**
- `w = 70.0`, `h = 1.75`

**Calculation (`cal_bmi(70.0, 1.75)`):**
- Height squared: `1.75 * 1.75 = 3.0625`
- BMI: `70.0 / 3.0625 = 22.857142857142858`
- `return Bmi`

**Output:**
`output : 22.857142857142858`

---

#### 💡 The Main Idea / Intuition to Remember

```
Weight = 70 kg, Height = 1.75 m
Height² = 3.0625
               ↓
BMI = 70 / 3.0625 ≈ 22.86
```
