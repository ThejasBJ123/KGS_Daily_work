## Section 6: Wap to display first n even number

### ❓ Question / Problem Statement
**Problem Statement (Wap to display first n even number): Implement and trace the Python program for wap to display first n even number.**

--- 
### 💻 My Code

````python
# Wap to display first n even number 

def isEvenOdd(num):
    return num % 2 == 0  

count=int(input("enter the numb:"))

series=1
while count>0:
    if isEvenOdd(series):
        print(series , end=" ")
        count-=1
    series+=1
````

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `isEvenOdd` to encapsulate reusable task logic.
2. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.
3. **Modulo Division (`%`):** Uses remainder arithmetic to check divisibility, factors, or extract digits.
4. **Conditional Branching (`if-else`):** Evaluates boolean expressions to branch execution path.
5. **While Loop Control:** Continues execution while state control condition evaluates to `True`.
6. **Return Statement:** Passes calculated result back to caller function.

### 🔍 Step-by-Step Code Tracing (Dry Run)

#### Sample Execution: `count = 3`

**Initially:**
- `count = 3`, `series = 1`

**Loop Execution (`while count > 0`):**
- **Cycle 1 (`series = 1`):** `isEvenOdd(1)` is **False**. `series = 2`, `count = 3`.
- **Cycle 2 (`series = 2`):** `isEvenOdd(2)` is **True**. Print `2 `, `count = 2`, `series = 3`.
- **Cycle 3 (`series = 3`):** `isEvenOdd(3)` is **False**. `series = 4`, `count = 2`.
- **Cycle 4 (`series = 4`):** `isEvenOdd(4)` is **True**. Print `4 `, `count = 1`, `series = 5`.
- **Cycle 5 (`series = 5`):** `isEvenOdd(5)` is **False**. `series = 6`, `count = 1`.
- **Cycle 6 (`series = 6`):** `isEvenOdd(6)` is **True**. Print `6 `, `count = 0`, `series = 7`.

**Termination:**
- `count` is 0. Condition `0 > 0` is **False**. Loop ends.

**Output:**
`2 4 6 `

---

#### 💡 The Main Idea / Intuition to Remember

```
count = 3
  1 → Odd  → Skip
  2 → Even → Print 2 (count = 2)
  3 → Odd  → Skip
  4 → Even → Print 4 (count = 1)
  5 → Odd  → Skip
  6 → Even → Print 6 (count = 0 → STOP)
```

**Problem Statement (Wap to display first n odd number): Implement and trace the Python program for wap to display first n odd number.**

### 💻 My Code

````python
# Wap to display first n odd number 

def isEvenOdd(num):
    return num % 2 == 0


count=int(input("enter the numb:"))

series=1
while count>0:
    if not isEvenOdd(series):
        print(series, end=" ")
        count-=1
    series+=1
````
