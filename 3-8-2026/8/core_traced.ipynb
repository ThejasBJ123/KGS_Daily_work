## Section 3: no parameter and no return value

--- 
### 💻 My Code

```python
# no parameter and no return value
class calculator:
    def __init__(self):
        self.brand="casio"
        self.cost=200

    def add(self):
        a=10
        b=20
        c=a+b
        print(c)
c1=calculator()
print(c1.brand)
print(c1.cost)
c1.add()
```
