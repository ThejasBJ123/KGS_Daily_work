### Question 1
Write a Python program to check whether a given number is **even** or **odd** using simple `if-else` condition.

````python
# 1.	 Write a program to check weather the given number is even or odd 
n=int(input("enter the number:"))
if n%2==0:
    print(n,"even")
else:
    print(n,"odd number")
````

### Simple Code Tracing (Step-by-Step)

**Example 1: Input `n = 7` (Odd Number)**
1. **Input:** You enter `7`, so `n = 7`.
2. **Check Remainder:** `7 % 2` leaves remainder `1`.
3. **Compare:** Is `1 == 0`? **No (False)**.
4. **Action:** Python skips `if` and goes to `else`.
5. **Output:** Prints `7 odd number`.

**Example 2: Input `n = 4` (Even Number)**
1. **Input:** You enter `4`, so `n = 4`.
2. **Check Remainder:** `4 % 2` leaves remainder `0`.
3. **Compare:** Is `0 == 0`? **Yes (True)**.
4. **Action:** Python runs the `if` block.
5. **Output:** Prints `4 even`.

### Question 2
Write a Python program to check whether a given number is even or odd using a custom function `isEvenOdd(n)` and a `flag` variable.

````python
# 2. write a program to check wheather the given number is even or odd using a customized function  using flag
def isEvenOdd(n):
    if n%2==0:
        return True
    else:
        return False
num=int(input("enter the number "))
flag=isEvenOdd(num)

if flag:
    print(num,"is Even")
else:
    print(num," is odd")
````

6. **Output:** Prints `10 is Even`.

### Question 6
Write a Python program to display the first `n` even numbers using a `while` loop.

````python
# Wap to display first n even number 

def isEvenOdd(num):
    return num % 2 == 0  

count=int(input("enter the numb:"))

series=1
while count>0:
    if isEvenOdd(series):
        print(series , end=" ")
        count-=1
    series+=1
````

### Simple Code Tracing (Step-by-Step)

**Example: Input `count = 3` (We want first 3 even numbers)**
1. **Start:** `count = 3`, starting number `series = 1`.
2. **Check `series = 1`:** Not even. Increment `series` to `2`.
3. **Check `series = 2`:** Even! Print `2`. Decrement `count` to `2`. Increment `series` to `3`.
4. **Check `series = 3`:** Not even. Increment `series` to `4`.
5. **Check `series = 4`:** Even! Print `4`. Decrement `count` to `1`. Increment `series` to `5`.
6. **Check `series = 5`:** Not even. Increment `series` to `6`.
7. **Check `series = 6`:** Even! Print `6`. Decrement `count` to `0`.
8. **Stop:** `count` reached `0`, loop ends!
9. **Final Output:** `2 4 6`

### Question 7
Write a Python program to display the first `n` odd numbers using a `while` loop.

````python
# Wap to display first n odd number 

def isEvenOdd(num):
    return num % 2 == 0


count=int(input("enter the numb:"))

series=1
while count>0:
    if isEvenOdd(series):
        print(series , end=" ")
        count-=1
    series+=1

    if not isEvenOdd(series):
        print(series,end="")
        count-=1
    series+=1



````

### Simple Code Tracing (Step-by-Step)

**Example: Input `count = 2`**
1. **Start:** `count = 2`, `series = 1`.
2. **Iteration 1 (`series = 1`):**
   - First `if isEvenOdd(1)`: `1` is odd (False) $\rightarrow$ Skips. `series` becomes `2`.
   - Second `if not isEvenOdd(2)`: `2` is even (False) $\rightarrow$ Skips. `series` becomes `3`.
3. **Iteration 2 (`series = 3`):**
   - First `if isEvenOdd(3)`: `3` is odd (False) $\rightarrow$ Skips. `series` becomes `4`.
   - Second `if not isEvenOdd(4)`: `4` is even (False) $\rightarrow$ Skips. `series` becomes `5`.
4. **Note:** In this loop structure, `series` increments twice per loop iteration.

## Theory & Key Concepts Used in the Codes

1. **`input()` Function**: Reads user input as a string.
2. **`int()` Function**: Converts string input into an integer for calculation.
3. **Modulus Operator (`%`)**: `n % 2 == 0` checks if a number is divisible by 2 (Even).
4. **`if-else` Conditional Statements**: Executes code block based on True/False evaluation.
5. **Custom Functions (`def`)**: Reusable block of code performing a specific task.
6. **Boolean Expression (`return n % 2 == 0`)**: Directly returns `True` or `False`.
7. **Flag Variable**: Variable storing boolean state (`flag = isEvenOdd(num)`).
8. **`range(start, end)`**: Generates a sequence of numbers from `start` to `end - 1`.
9. **`for` & `while` Loops**: Iteration controls for repeating execution.
10. **Logical Operator (`not`)**: Reverses boolean state (`not isEvenOdd(i)` for odd numbers).
