### Question 7

```python
a=10  # global variable
def fun1():
    global a
    a=100
    print(a)
    b=20
    print(b)
def fun2():
    global a
    a=1000
    print(a)
    c=30
    print(c)

print(a)
fun1()
fun2()
print(a)
```

### Question 8
Write a Python program for: **Python Program Implementation 8**

```python
def fun1():
    print("inside a KGS")


def add(x, y):
    z = x + y
    print(z)


print(fun1)
print(add)

fun1()
add(10, 20)


ptr1 = fun1
print(ptr1)
ptr1()


ptr2 = add
print(ptr2)
ptr2(10, 20)
```

### Question 9
Write a Python program for: **Python Program Implementation 9**

```python
def fun1():
    print("inside the kgs ")


def add(x,y):
    z=x+y
    print(z)

def display(ptr1,ptr2):
    ptr1()
    ptr2(10,20)

ptr3=fun1
ptr4=add

print(ptr3)
print(ptr4)


ptr3()
ptr4(100,10)

print("====")
display(ptr3,ptr4)  # higher order function
print("=====")
fun1()
add(10,20)
```
