**1.Database**: The Systematic and organized collection of data stored in an electronic device.

**4.Views**: It is a virtual table that includes the rows and cols of the original table which is create during SELECT query.

**6.STORED PROCEDURE:** pre-difined collection of queries that help to perfrom operation.

### The 5 important commands is SQL

1. DDL (Data dinitionlanguage): used to define the DB objects.
    - **CREATE, DROP, TRUNCATE, ALTER**

2. DML (Data Manipulation Language): Used to perform operations on a table 
    - **INSERT, UPDATE, DELETE**

3. DQL( Data Query Language): Used to select the display the data form the table 
    - **SELECT**

4. TCL(Transaction Control Language): Used to control and save the transactions
    - **COMMIT, ROLLBACK, SAVEPOINT**

5. DCL(Data Control Language): Used to control the user accessibilit towards the DB.
    - **GRANT, REVOKE**
