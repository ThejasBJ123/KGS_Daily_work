## Section 1: class Employee:

### ❓ Question / Problem Statement
**Problem Statement (class Employee:): Implement and trace the Python program for class employee:.**

--- 
### 💻 My Code

```python
class Employee:
    company_name="Kaizentrix Global solution"
    employee_count=1000
    def __init__(self, employee_name, department):
        self.employee_name=employee_name
        self.department=department
    
    def employee_generated_id(self):
        Employee.employee_count+=1
        employee_id="EMP"+"KGS"+str(Employee.employee_count)
        print("\n =====================================")
        print("CompanyName: ", Employee.company_name)
        print("Name: ", self.employee_name)
        print("Empoyee_id", employee_id)
        print("Department: ", self.department)
        print("/n =================================")

ename=input("Enter the name :  ")
Dept=input("Enter the Department : ")

 
e1=Employee(ename,Dept)
e1.employee_generated_id()
```

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `__init__, employee_generated_id` to encapsulate reusable task logic.
2. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.
3. **Object-Oriented Programming:** Defines class structure with attributes and methods (`__init__`, static, or instance methods).
