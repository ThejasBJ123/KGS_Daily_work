## Section 4: factorial

### ❓ Question / Problem Statement
**Problem Statement (factorial): Implement and trace the Python program for factorial.**

--- 
### 💻 My Code

````python
# factorial

def factorial(n):
    if n <= 1:
        return 1

    return n * factorial(n - 1)


num = int(input("Enter the number: "))
res = factorial(num)

print("The factorial of", num, "is:", res)
````

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `factorial` to encapsulate reusable task logic.
2. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.

## Section 5: def displayfactors(n,i):

### ❓ Question / Problem Statement
**Problem Statement (def displayfactors(n,i):): Implement and trace the Python program for def displayfactors(n,i):.**

--- 
### 💻 My Code

````python
def displayfactors(n,i):
    if n>i:
        return
    
    if n%i==0:
        print(i,end =" ")
    displayfactors(n,(i+1))
    
num=int(input("enter the num:"))
res=displayfactors(num,1 )
print(res)
````
