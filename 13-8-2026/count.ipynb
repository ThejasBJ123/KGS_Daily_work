### Question 1
Write a Python program for: **count digits**

```python
 # count digits 

def isCount(n,count):

    if n<=0:
        return count

    n=n//10
    count+=1
    return isCount(n, count)


num=int(input("enter the number : "))
c = isCount(num,0)
print("Number of digits:",num,"is: ", c)
```

### Question 4
Write a Python program for: **factorial**

```python
# factorial

def factorial(n):
    if n <= 1:
        return 1

    return n * factorial(n - 1)


num = int(input("Enter the number: "))
res = factorial(num)

print("The factorial of", num, "is:", res)
```

-	WAP to sum all the digits present in a given number. 
	N = 5481
	N = 5481 	0+1
	N = 548 	1+8
	N = 54  	9+4
	N = 5 		13+5
	N = 0		ret 18

```python
def sum_num(n, sum):

    if n <= 0:
        return sum

    digit = n % 10
    sum = sum + digit

    n = n // 10

    return sum_num(n, sum)


num = int(input("Enter the number: "))

result = sum_num(num, 0)

print("Sum is:", result)
```
