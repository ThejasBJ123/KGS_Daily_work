```python
#no parameter no return value
def add():
    a=10
    b=20
    c=a+b
    print(c)
add()

```

```python
# no parameter wiht return value
def add(): 
    a=10
    b=20
    c=a+b
    return c
r1=add()
print(r1)
```

```python
# with parameter and no return value
def add(a,b):
    c=a+b
    print(c)

x=10
y=20
add(x,y)
```

```python
# wiht parameter and wiht return value
def add(a,b):
    c=a+b
    return c
x=10
y=20
r1=add(x,y)
print(r1)
```
