### Question 1
Write a Python program for: **no parameter no return value**

```python
#no parameter no return value
def add():
    a=10
    b=20
    c=a+b
    print(c)
add()

```

### Question 2
Write a Python program for: **no parameter wiht return value**

```python
# no parameter wiht return value
def add(): 
    a=10
    b=20
    c=a+b
    return c
r1=add()
print(r1)
```

### Question 3
Write a Python program for: **with parameter and no return value**

```python
# with parameter and no return value
def add(a,b):
    c=a+b
    print(c)

x=10
y=20
add(x,y)
```

### Question 4
Write a Python program for: **wiht parameter and wiht return value**

```python
# wiht parameter and wiht return value
def add(a,b):
    c=a+b
    return c
x=10
y=20
r1=add(x,y)
print(r1)
```
