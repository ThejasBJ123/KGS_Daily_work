### Question 3: Oops copy Exercise 3

```python
class Farmer:
    r=2.5
    def __init__(self, p, t):
        self.p=p
        self.t=t
    
    def Simple_calculatore(self):
        si=(self.p*self.t*self.r)/100
        print(si)

s1=Farmer(1000,2)
s1.Simple_calculatore()
```
