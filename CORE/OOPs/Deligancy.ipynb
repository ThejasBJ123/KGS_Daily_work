```python
class Radio:
    def turnon(self,x):
        if x==1:
            print("Radio is on")
        else:
            print("Radio is off")

class car:
     def __init__(self, min, max):
        self.cmin=min
        self.cmax=max
        self.r=Radio()

c=car(60,180)
print(c.cmin)
print(c.cmax)
c.r.turnon(1)
c.r.turnon(2)
```

```python
class os:
    def __init__(self):
        self.status=True 
        print("OS is created")
    def getos(self):
        print("os is ready")

class Mobile:
    def __init__(self,name):
        self.mname=name
        self.o=os()
        print("Mobile is ready")

    def getmobile(self):
        print("Mobile is installed wiht os")

m=Mobile("iphone")
print(m.mname)
m.getmobile()
print(m.o.status)
m.o.getos()
del m
# print(m.o.status)

```
