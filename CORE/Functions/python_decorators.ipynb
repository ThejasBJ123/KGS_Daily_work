```python
# without decorator 
def greet():
    print("Hi, Everyone!")

def outer(print1):
    print("entering outer")
    def inner():
        print("Entering inner")
        ref=print1
        ref()
        print("leaving inner")
    return inner 
ptr1=greet
ptr2=outer(ptr1)
ptr2()
```

```python
#decorator
def outer(fun):
    print("entering outer")
    def inner():
        print("Entering inner")
        fun()
        print("leaving inner")
    return inner

@outer
def greet():
    print("hi hello")
greet()


```

```python
def outer(fun):
    print("entering outer")
    def inner():
        print("Entering inner")
        fun()
        print("leaving inner")
    return inner

@outer
def greet():
    name = input("Enter the name: ")
    print(name.upper())
greet()

```
