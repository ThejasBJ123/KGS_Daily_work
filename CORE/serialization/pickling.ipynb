# **Serializaiton/pickling** ==> dump()

```python
import pickle
class Employee:
    def __init__(self,name, age, sal,mob):
        self.name=name
        self.age=age
        self.sal=sal
        self.mob=mob
    
    def disp(self):
        print(self.name)
        print(self.age)
        print(self.sal)
        print(self.mob)
        

e1=Employee("ramu",22, 2000044, 95365651)
f=open("emp.txt", "wb")
pickle.dump(e1,f)
e1.disp()

f.close()
print("data are stored successfully into a file : ")
```

```python
import pickle
class Employee:
    def __init__(self,name, age, sal,mob):
        self.name=name
        self.age=age
        self.sal=sal
        self.mob=mob
    
    def disp(self):
        print(self.name)
        print(self.age)
        print(self.sal)
        print(self.mob)
        

f=open("emp.txt", "rb")
e1=pickle.load(f)
e1.disp()

f.close()
print("data are retived successfully into a file : ")
```
