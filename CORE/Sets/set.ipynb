# traversing the set

# note :
 in set we ca use a for loop for the travsering  but while loop we cant able to use in set because while loop should need a specific postion[index of a element in set indexing and slicing is not allowed ]

```python
s={10,20,30,40,50}

for i in s:
    print(i)
```

```python
s=set()
for i in range(5):
    data=int(input("Enter a value: "))
    s.add(data)
    
print(s)
```

```python
s={10,20,30,40,50}
s.remove(40)  # by using a remove if element is not in the set get error 
print(s)

s.discard(200)  # if by using the discard element is not they it show none or -1
print(s)
```
