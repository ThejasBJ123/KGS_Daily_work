# **Files**


```python
fname=input("Enter a file name: ")
fptr = open(fname, "w")
for i in range(5):
    data=input("Enter the name : ")
    fptr.write(data+"\n")
fptr.close()
print("Data are stored in a file ")
```

```python
fname=input("Enter a file name: ")
fptr = open(fname, "a")
for i in range(5):
    data=input("Enter the name : ")
    fptr.write(data+"\n")
fptr.close()
print("Data are stored in a file ") 
```

```python
import os

oldname = input("Enter existing file name: ")
newname = input("Enter new file name: ")

os.rename(oldname, newname)

print("File renamed successfully")
```

# write a program to display the empoolyee details in a single details 

```python
fname = input("Enter the file name: ")
fptr=open(fname, "w")
for i in range(3):
    eid=input("Enter a empid: ")
    ename=input("Enter a ename: ")
    emailid=input("Enter a mailid ")
    esal=input("Enter a salary: ")
    econtact=input("Enter a contact: ")
    ecompany=input("Enter a companyname: ")
    data = eid + " " +ename+" "+emailid+ " "+esal+" "+econtact+ " "+ecompany
    
    fptr.write(data +"\n")
fptr.close()
print("Successfully data stored in a file ")
```

### reading a data fom the file: 
1. tell()
2. Seek()
3. read()
4. readline()
5. readlines()


```python
fptr=open("emp.txt", "r")

data = fptr.tell()
print(data)

data1=fptr.seek(10)
print(data1)
print(fptr.tell())


data2=fptr.read()
print(data2)

data3=fptr.readline()
print(data3)

data4=fptr.readlines()
print(data4) # it prints all the dat inside list forma which is present in a given a file 
```
