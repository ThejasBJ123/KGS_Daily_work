# **Files**


```python
fname=input("Enter a file name: ")
fptr = open(fname, "w")
for i in range(5):
    data=input("Enter the name : ")
    fptr.write(data+"\n")
fptr.close()
print("Data are stored in a file ")
```

```python
fname=input("Enter a file name: ")
fptr = open(fname, "a")
for i in range(5):
    data=input("Enter the name : ")
    fptr.write(data+"\n")
fptr.close()
print("Data are stored in a file ") 
```

```python
import os

oldname = input("Enter existing file name: ")
newname = input("Enter new file name: ")

os.rename(oldname, newname)

print("File renamed successfully")
```
