```python
d = {
    'name': 'rama',
    'age': 23,
    'place': 'Bangalore'
}

print(type(d))
print(len(d))
print(d['age'])

d1={}
print(type(d1))
```

```python
d = {
    'name': 'rama',
    'age': 23,
    'place': 'Bangalore'
}

for i in d: 
    print(i)
```

```python
d = {
    'name': 'rama',
    'age': 23,
    'place': 'Bangalore'
}

for i in d: 
    print(i," ",d[i])
```

```python
d = {
    'name': 'rama',
    'age': 23,
    'place': 'Bangalore'
}

for i,j in d.items():
    print(i," ",j)
```

```python
d = {
    'name': 'rama',
    'age': 23,
    'place': 'Bangalore'
}

for i in d.values():
    print(i) 
    
```

```python
d = {
    'name': 'rama',
    'age': 23,
    'place': 'Bangalore'
}
print(d)
    
```

### nested Dictionary

```python
d = {
    'name': 'rama',
    'age': 23,
    'place': {'curr':'Bangalore',
              'per': 'Andhra'},
    'mob': {'mob1':9988,
            'mob2':6677}
}    
print(d)
print(d['name'])
print(d['place']['curr'])
print(d['mob']['mob2'])
```

```python
# insert to dictionary
d = {
    'name': 'rama',
    'age': 23,
    'place': {'curr':'Bangalore',
              'per': 'Andhra'},
    'mob': {'mob1':9988,
            'mob2':6677}
}   

d['place']['per']='america'
print(d)
```

### *  Deepcopy
### *  shallowcopy

```python
import copy
d = {
    'name': 'rama',
    'age': 23,
    'place': {'curr':'Bangalore',
              'per': 'Andhra'},
    'mob': {'mob1':9988,
            'mob2':6677}
}   

d1=d   # shallow copy method
print(d)
print(d1)

d2=d.copy()
print(d2)


d3=copy.deepcopy(d)
print(d3)

```

# Zip funcion in dictinary

```python
emp_id =[101,102,103,104]
emp_name=['rama','sita','ravana','abhi']

info=dict(zip(emp_id, emp_name))
print(info)
```

```python
emp_id =[101,102,103,104]
emp_name=['rama','sita','ravana','abhi']
emp_salary=[2000, 25000, 30000, 35000]

info=list(zip(emp_name, emp_salary))
print(info)


final_info=dict(zip(emp_id, info))
print(final_info)
```
