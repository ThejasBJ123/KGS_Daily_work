# **Iterators**

```python
s=['pavi', 'chandra', 'mouli']
it=iter(s)

print(next(it))
print(next(it))
print(next(it))
itt=iter(s)
print(next(itt))
# print(next(it))
```

```python
def number ():
    yield 10
    yield 20
    yield 30
g=number()
print(next(g))
print(next(g))
print(next(g))
```

```python
def number ():
    yield 10
    yield 20
    yield 30
for i in number():
    print(i)
```
