# Inheritance in Python (OOP)
Inheritance allows a class (child class) to inherit attributes and methods from another class (parent class). Includes Single, Multilevel, Hierarchical, Multiple, Hybrid inheritance, `super()`, and Method Resolution Order (MRO).

### Question 5: Single-Level Inheritance - Parent and Child Classes
Write a Python program demonstrating single-level inheritance where child inherits parent attributes.

```python
class Parent:
    def __init__(self):
        self.a = 10


class Child(Parent):
    def __init__(self):
        super().__init__()
        self.b = 20


c1 = Child()
print("Child Variable b:", c1.b)
print("Inherited Parent Variable a:", c1.a)
```

### Question 6: Multilevel Inheritance - Multi-tier Class Hierarchy
Write a Python program demonstrating multilevel inheritance across classes A, B, and C.

```python
class A:
    def __init__(self):
        self.a = 10


class B(A):
    def __init__(self):
        super().__init__()
        self.b = 20


class C(B):
    def __init__(self):
        super().__init__()
        self.c = 30


c1 = C()
print("Level C Attribute c:", c1.c)
print("Level B Attribute b:", c1.b)
print("Level A Attribute a:", c1.a)
```

### Question 7: Object-Oriented Modeling - Airplane Specialized Classes
Write a Python program defining specialized Airplane classes (CargoPlane, PassengerPlane, FighterPlane) with specific behaviors.

```python
class CargoPlane:
    def takeoff(self):
        print("Cargo plane is taking off")

    def land(self):
        print("Cargo plane is landing")

    def fly(self):
        print("Cargo plane is flying")

    def carry(self):
        print("Cargo plane carries cargo")


class PassengerPlane:
    def takeoff(self):
        print("Passenger plane is taking off")

    def land(self):
        print("Passenger plane is landing")

    def fly(self):
        print("Passenger plane is flying")

    def carryp(self):
        print("Passenger plane carries passengers")


class FighterPlane:
    def takeoff(self):
        print("Fighter plane is taking off")

    def land(self):
        print("Fighter plane is landing")

    def fly(self):
        print("Fighter plane is flying")

    def carryw(self):
        print("Fighter plane carries weapons")


c = CargoPlane()
p = PassengerPlane()
f = FighterPlane()

c.takeoff()
c.land()
c.fly()
c.carry()

p.takeoff()
p.land()
p.fly()
p.carryp()

f.takeoff()
f.land()
f.fly()
f.carryw()
```

### Types of Inhertance

* single inheritance
* multilevel inheritance
* hierarchial inheritance
* multiple inheritance
* hybrid inheritance


```python
# single inheritance

class A:
    def dispA(self):
        print("A")


class B(A):
    def dispB(self):
        print("B")
        A.dispA(self)


b1 = B()
b1.dispB()
b1.dispA()
```

```python
#  multilevel inheritance

class A:
    def dispA(self):
        print("A")
    
class B(A):
    def dispB(self):
        print("B")

class C(B):
    def dispC(self):
        print("c")

c1=C()
c1.dispC()
c1.dispB()
c1.dispA()

    
```

```python
# hierarchial inheritance
class A:
    def dispA(self):
        print("A")

class B(A):
    def dispB(self):
        print("B")

class C(A):
    def dispC(self):
        print("C")

b1 = B()
b1.dispB()
b1.dispA()

c1 = C()
c1.dispC()
c1.dispA()
```

```python
# multiple inheritance
class A:
    def dispA(self):
        print("A")

class B:
    def dispB(self):
        print("B")

class C(A, B):
    def dispC(self):
        print("C")

c1 = C()
c1.dispC()
c1.dispA()
c1.dispB()
```

```python
# hybrid inheritance
class A:
    def dispA(self):
        print("A")

class B(A):
    def dispB(self):
        print("B")

class C(A):
    def dispC(self):
        print("C")

class D(B, C):
    def dispD(self):
        print("D")

d1 = D()
d1.dispD()
d1.dispB()
d1.dispC()
d1.dispA()
```

### Super Key

```python
# invoking a parent class construction manually
# super().__init__()

class A:
    def __init__(self):
        self.a=10

class B(A):
    def __init__(self): 
        super().__init__()
        self.b=20

class C(B):
    def __init__(self):
        super().__init__()
        self.c1=30

cf=C()
print(cf.c1)
print(cf.b)
print(cf.a)
```
