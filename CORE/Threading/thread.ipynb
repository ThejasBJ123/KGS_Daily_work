# **Threading**

```python
from threading import Thread
import time

class Task1(Thread):
    def run(self):
        for i in range(1,5):
            print(i)
            time.sleep(2)

class Task2(Thread):
    def run(self):
        for i in range(5, 1, -1):
            print(i)
            time.sleep(1)

t1 = Task1()
t2 = Task2()

t1.run()
t2.run()


print("Thread Completed")
```

```python
from threading import Thread
import time

class Task1(Thread):
    def run(self):
        for i in range(1, 20):
            print(i)
            time.sleep(2)

class Task2(Thread):
    def run(self):
        for i in range(5, 1, -1):
            print(i)
            time.sleep(1)

t1 = Task1()
t2 = Task2()

t1.start()
t2.start()
t1.join()
t2.join()

print("Thread Completed")
```

```python
from threading import Thread, Lock
import time

lock = Lock()


class Task1(Thread):

    def run(self):
        for i in range(1, 20, 2):
            with lock:
                print("odd",i)
                time.sleep(1)


class Task2(Thread):

    def run(self):
        for i in range(2, 20, 2):
            with lock:
                print("even",i)
                time.sleep()


t1 = Task1()
t2 = Task2()

t1.start()
t2.start()

t1.join()
t2.join()

print("Thread Completed")
```
