# **TUPLE**

* Heterogenous
* Duplicates are allowed
* Immutable
* insertion order

```python
t=(10,20,30,"Apple",True, 20.5)
print(t)
print(type(t))

print(len(t))
print(t[2])
print(t[4])
print("="*30)

t2=(10)  # without using the , this is display as int 
print(t2)
print(type(t2))
print("="*30)

t3=(10,)   # wiht using the , this is diplay as tuple
print(t3)
print(type(t3))

```
