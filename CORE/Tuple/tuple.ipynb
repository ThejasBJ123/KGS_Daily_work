# **TUPLE**

* Heterogenous
* Duplicates are allowed
* Immutable
* insertion order

```python
t=(10,20,30,"Apple",True, 20.5)
print(t)
print(type(t))

print(len(t))
print(t[2])
print(t[4])
print("="*30)

t2=(10)  # without using the , this is display as int 
print(t2)
print(type(t2))
print("="*30)

t3=(10,)   # wiht using the , this is diplay as tuple
print(t3)
print(type(t3))

```

```python
t=(10,20,30,40,50)
print(t)
print(t[::1])
print(t[:-2:1])
print(t[2::-1])
print(t[-4::1])
print(t[-4:-1:-1])
# print(t[-4:-2:0])  0 not comes

```

```python
t=(10,20,30,40,50)

print(t[3:1:1])
print(t[2:-5:-1])
```

```python
t=(10,20,30,(1.1, 1.2,('a','b','c'),1.3),40)

print(t)
print(type(t))
print(len(t))
print(t[4])
print(t[3][0])
print(t[3][2][1])
```

```python
# delting the delete and insert the element 

t=(10,20,30,40,50)

t=t[ :2] +(25,) + t[2: ]
print(t)

t=t[:3] +t[4:]
print(t)
```

```python
# zip funcion in the tuple 


player=["Dhoni","virat", "K.L.Rahul","Rohith Sharma", "Abhi"]
age=[45,34,33,45,23]
ipl_runs=[5439, 9339,2343, 9833]
ipl_team=['csk','RCB','Lucknow','MI','rcb']
info=list(zip(player, age,ipl_runs, ipl_team))
print(info)
```
