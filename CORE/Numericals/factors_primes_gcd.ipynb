### Question 3
Write a Python program for: **3 - wap to count the num of cycles taken to get the all the factor of a given number**

```python
# 3 - wap to count the num of cycles taken to get the all the factor of a given number 
def factor_display(n):
    for i in range(1, n + 1):
        if n % i == 0:
            print(i, end=" ")


def countFactors(n):
    countFact = 0
    for i in range(1, n + 1):
        if n % i == 0:
            countFact += 1
    return countFact


def countFactCycles(n):
    countCycles = 0
    for i in range(1, n + 1):
        countCycles += 1
    return countCycles

def isPrimeNumber(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

# Main Program
num = int(input("Enter the number: "))

print("The factors of", num, "are:")
factor_display(num)

res = countFactors(num)
print("\nThe count of factors of", num, "is:", res)

resCycles = countFactCycles(num)
print("The count of cycles taken to get the factors of", num, "is:", resCycles)


if isPrimeNumber(num):
    print(num, "is a Prime Number")
else:
    print(num, "is Not a Prime Number")
```

## Theory & Key Concepts Used in the Codes

1. **Factors**: Numbers `i` where `n % i == 0`.
2. **Prime Number**: Number greater than 1 having exactly 2 factors (1 and itself).
3. **Optimized Loop (Logic 2)**: Checking factors up to $n/2$ or $\sqrt{n}$ drastically reduces iterations.
4. **Factor Counting**: Counting factors to determine primality (`count == 2`).
5. **Range Filtration**: Printing prime and non-prime numbers separately in a user-defined range.

GCD /HCF
-------
--> Greates common divisor/highest common factor
--> given two numbers find the common factors that divides both the I/p num's down to 0 and return the hightest factor amongst them 
--> the lower will never hava a factor in the higher range of the larger num

* wap to find the GCD/ HCF  of the given 2 num's 


```python
# wap to check the number 
def Hcf_num(n1,n2):
    lower = n1
    if n2<n1:
        lower=n2

    hcf=1
    for i in range(2,(lower+1)):
        if n1%i==0 and n2%i==0:
            hcf=i

    return hcf
   

f_num =int(input("enter the first number : "))
s_num=int(input("Enter the second number : "))
  

res= Hcf_num(f_num, s_num)
print( " the hcf of ",f_num,"and ",s_num," is: ", res)
```

Co- Prime's
----------

* if the HCF/ GCD of the given 2 num's is exactly 1, then they are said to be the co-Prime's of the each other

eg: 
   n1 =18     n2= 19    hcf=1 ==> co - prime's 
   n1=21      n2 = 18  hcf = 3 ===> not co primes's 

* wap to check weather the given 2 num's are coprime or each other or not 


```python
# wap to check weather the given 2 num's are coprime or each other or not 
def isCoprime(n1,n2):
    lower=n1
    if n2 < n1 :  
        lower=n2
    hcf =1
    for i in range(2, (lower+1)):
        if n1%i==0 and n2%i==0:
            hcf=i
    return hcf==1


num1=int(input("enter the first number: "))
num2=int(input("Enter the secont number : "))


if isCoprime(num1, num2):
     print(" co prime number ")
    
else: 
    print("not a coprime number ")
```

```python
# increment number 
def Fibonacci1(pos):
    n1,n2=0,1
    while pos>0:
        print(n1, end=" ")
        temp=n1+n2
        n1=n2
        n2=temp
        pos-=1


def Fibonacci3(pos):
    n1, n2=0,1
    for i in range(pos, 1,-1,-1):
        print(n1, end=" ")
        temp=n1 +n2
        n1=n2
        n2=temp


def Fibonacci4(pos):
    n1,n2= 0,1
    for i in range(1, pos+1):
        print(n1, end=" ")
        temp =n1+n2
        n1=n2
        n2=temp
    
pos=int(input("Enter the number: "))
Fibonacci(pos)


print("========================")
Fibonacci3(pos)

```
