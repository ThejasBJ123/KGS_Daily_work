## 1. Extract ASCII Value (`ord()` & `chr()`)

### 📘 Definition:
- **`ord(char)`**: Takes a single character and returns its ASCII number value.
- **`chr(code)`**: Takes an ASCII number value and returns its character.

### ⚙️ Syntax:
```python
ascii_num = ord('A')  # Returns 65
character = chr(65)   # Returns 'A'
```



````python
# Simple Example: ASCII Value Extraction

char = 'A'
ascii_val = ord(char)
print("ASCII value of 'A' :", ascii_val)

num = 65
char_val = chr(num)
print("Character for 65    :", char_val)
````

````python
print(chr(64+4))
````

## 2. Programs on Strings

### 📘 Definition:
Simple Python programs demonstrating basic string manipulation like character iteration, string reversing, and counting.

---


````python
# Program 1: Print ASCII value of each character in a string

text = "Hi"

for char in text:
    print(char, "->", ord(char))
````

````python
# Program 2: Reverse a string

text = "python"
reversed_text = text[::-1]

print("Original string :", text)
print("Reversed string :", reversed_text)
````

````python
str1 = "programming"
print("".join(set(str1)))
````

````python
str1 = "programming"
print("".join(dict.fromkeys(str1)))
````

````python
str1 = "programming"

for ch in str1:
    if str1.count(ch) > 1:
        str1 = str1.replace(ch, "", 1)

print(str1)
````

````python
# Program 3: Count vowels in a string

text = "hello world"
vowel_count = 0

for char in text:
    if char in "aeiou":
        vowel_count += 1

print("Text        :", text)
print("Vowels count:", vowel_count)
````

## 3. Inbuilt String Methods (Simple Examples)

---


### 1️⃣ `upper()`

#### 📘 Definition:
Converts all letters in the string to uppercase.

#### ⚙️ Syntax:
```python
string.upper()
```



````python
# Simple Example: upper()
s = "hello"
print(s.upper())  # Output: HELLO
````

### 2️⃣ `lower()`

#### 📘 Definition:
Converts all letters in the string to lowercase.

#### ⚙️ Syntax:
```python
string.lower()
```



````python
# Simple Example: lower()
s = "HELLO"
print(s.lower())  # Output: hello
````

### 3️⃣ `strip()`

#### 📘 Definition:
Removes spaces from both the beginning and end of the string.

#### ⚙️ Syntax:
```python
string.strip()
```



````python
# Simple Example: strip()
s = "  hello  "
print(s.strip())  # Output: 'hello'
````

### 4️⃣ `split()`

#### 📘 Definition:
Splits a string into a list of words.

#### ⚙️ Syntax:
```python
string.split()
```



````python
# Simple Example: split()
s = "python is easy"
print(s.split())  # Output: ['python', 'is', 'easy']
````

### 5️⃣ `find()`

#### 📘 Definition:
Returns index of first match. Returns `-1` if not found.

#### ⚙️ Syntax:
```python
string.find("search_text")
```



````python
# Simple Example: find()
s = "hello world"
print(s.find("world"))  # Output: 6
print(s.find("python")) # Output: -1
````

### 6️⃣ `index()`

#### 📘 Definition:
Returns index of first match. Gives error if not found.

#### ⚙️ Syntax:
```python
string.index("search_text")
```



````python
# Simple Example: index()
s = "hello world"
print(s.index("world"))  # Output: 6
````

### 7️⃣ `rfind()`

#### 📘 Definition:
Returns index of last (rightmost) match. Returns `-1` if not found.

#### ⚙️ Syntax:
```python
string.rfind("search_text")
```



````python
# Simple Example: rfind()
s = "hello hello"
print(s.rfind("hello"))  # Output: 6
````

### 8️⃣ `rindex()`

#### 📘 Definition:
Returns index of last (rightmost) match. Gives error if not found.

#### ⚙️ Syntax:
```python
string.rindex("search_text")
```



````python
# Simple Example: rindex()
s = "hello hello"
print(s.rindex("hello"))  # Output: 6
````

### 9️⃣ `lstrip()`

#### 📘 Definition:
Removes spaces from the left side only.

#### ⚙️ Syntax:
```python
string.lstrip()
```



````python
# Simple Example: lstrip()
s = "  hello  "
print(repr(s.lstrip()))  # Output: 'hello  '
````

### 🔟 `rstrip()`

#### 📘 Definition:
Removes spaces from the right side only.

#### ⚙️ Syntax:
```python
string.rstrip()
```



````python
# Simple Example: rstrip()
s = "  hello  "
print(repr(s.rstrip()))  # Output: '  hello'
````

### 1️⃣1️⃣ `count()`

#### 📘 Definition:
Counts how many times a character or word appears in the string.

#### ⚙️ Syntax:
```python
string.count("target")
```



````python
# Simple Example: count()
s = "banana"
print(s.count("a"))  # Output: 3
````
