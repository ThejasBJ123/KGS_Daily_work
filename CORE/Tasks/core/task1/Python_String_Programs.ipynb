# Python String Programs

Five string-manipulation programs with the given input/output examples.

## 1) Reverse a String
**Input:** `Rama`  
**Output:** `amaR`

```python
str1 = "lewis"
result1 = " "
for i in str1:
    result1 = i + result1
print("Original string is: ", str1)
print("Reversed string is: ", result1)
```

## 2) Reverse the Order of Words
**Input:** `Kaizentrix globaal solutions`  
**Output:** `solutions globaal Kaizentrix`

```python
s1 = "Kaizentrix global solutions"
s2 = s1.split()
print(s2)
print(len(s2))
new = ""
for i in s2:
    new = i + " " + new
print(new)
```

## 3) Remove Duplicate Letters
**Input:** `Hello`  
**Output:** `Helo`

```python
input_str = "Hello"
res = ""

for char in input_str:
    if char not in res:
        res += char
print(res)
```

## 4) Remove Duplicate Words
**Input:** `Rama is sleeping and Rama is walking`  
**Output:** `Rama is sleeping and walking`

```python
string1 = "Rama is sleeping and Rama is walking"
words = string1.split()
print(words)
print(len(words))
seen_words = " "

for word in words:
    seen_words = seen_words + " " + word

print("Original string:", string1)
print("Output string:  ", seen_words)
```
