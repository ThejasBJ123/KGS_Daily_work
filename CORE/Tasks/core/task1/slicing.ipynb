Email Validator
Scenario
The user enters an email address.
Example:
### rahul@gmail.com

*Requirements*
- Print the username.
- Print the domain name.
- Print the extension.
- Reverse the email

```python
str= input("Enter the email id: ")
print(str)
print(str[-4:])
print(str[-10:-4])
print(str[::-1])
```

Aadhaar Number Formatter
Scenario
Input

### 123456789012
Requirements
- Print the first 4 digits.
- Print the last 4 digits.
- Print the middle 4 digits.
- Display: 1234-5678-9012

```python
a_num= input("enter the aadhaar number : ")
print(a_num[:4])
print(a_num[-4:])
print(a_num[4:8])
print(a_num[:4],a_num[4:8], a_num[-4:], sep="-")
```

Company Name Analyzer
Scenario

### Kaizentrix Global Solutions
Requirements
- Print the first character.
- print the middle character
- Print the last character.
- Print every second character.
- Print the name in reverse.
- Count the total characters.


```python
com="Kaizentrix Global Solutions"

print(com[:10])
print(com[11:17])
print(com[-9:])
print(com[::2])
print(com[::-1])
print(len(com))
```
