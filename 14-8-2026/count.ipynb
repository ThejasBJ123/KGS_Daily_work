Wap to display the count of number of factors in a given number  

```python
def displayfactors(n, i):
    if i > n:
        return

    if n % i == 0:
        print(i, end=" ")

    displayfactors(n, i + 1)


def countFactors(n, i, countfact):
    if i > n:
        return countfact

    if n % i == 0:
        countfact += 1

    return countFactors(n, i + 1, countfact)


# Prime number
def prime(countfact):
    if countfact == 2:
        return True
    else:
        return False


num = int(input("Enter the num: "))

print("Factors of", num, "are:")
displayfactors(num, 1)

res = countFactors(num, 1, 0)

print("\nThe count of factors of", num, "is", res)

if prime(res):
    print(num, "is a Prime Number")
else:
    print(num, "is not a Prime Number")
```
