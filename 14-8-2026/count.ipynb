### Question 1
Write a Python program to: **Program 1**

```python
def displayfactors(n, i):
    if i > n:
        return

    if n % i == 0:
        print(i, end=" ")

    displayfactors(n, i + 1)


def countFactors(n, i, countfact):
    if i > n:
        return countfact

    if n % i == 0:
        countfact += 1

    return countFactors(n, i + 1, countfact)


# Prime number
def isprime(n,i=1,count=0):
    if(i>n):
        return count==2
    if n%i==0:
        count+=1
    return isprime(n,i+1,count)



num = int(input("Enter the num: "))

print("Factors of", num, "are:")
displayfactors(num, 1)

res = countFactors(num, 1, 0)

print("\nThe count of factors of", num, "is", res)

if isprime(res):
    print(num, "is a Prime Number")
else:
    print(num, "is not a Prime Number")
```

### Question 3
Write a Python program for: **Python Program Implementation 3**

```python
def Gcd(n1, n2, i, hcf, lower):
    if i > lower:
        return hcf

    if n1 % i == 0 and n2 % i == 0:
        hcf = i

    return Gcd(n1, n2, i + 1, hcf, lower)


num1 = int(input("Enter the number: "))
num2 = int(input("Enter the number: "))

lower = num1
if num2 < num1:
    lower = num2

result = Gcd(num1, num2, 1, 0, lower)
print("HCF =", result)
```

### Question 5
Write a Python program for: **CO PRIME**

```python
# CO PRIME

def COPRIME(n1, n2, i, hcf, lower):

    if i > lower:
        return hcf==1

    if n1 % i == 0 and n2 % i == 0:
        hcf = i

    return COPRIME(n1, n2, i + 1, hcf, lower)


num1 = int(input("Enter the number: "))
num2 = int(input("Enter the number: "))

lower = num1

if num2 < num1:
    lower = num2

result = COPRIME(num1, num2, 1, 1, lower)

if result :
    print("Coprime")
else:
    print("Not Coprime")
```

### Question 7
Write a Python program for: **Python Program Implementation 7**

```python
def Gcd(n1, n2):

    if n1 == 0:
        return n2==1

    if n1 < n2:
        n1, n2 = n2, n1

    return Gcd(n1 % n2, n2)


num1 = int(input("Enter the number: "))
num2 = int(input("Enter the number: "))

result = Gcd(num1, num2)

if result:
    print("Coprime")
else:
    print("Not Coprime")
```
