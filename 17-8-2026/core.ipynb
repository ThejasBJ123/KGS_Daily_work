```python
def even(num):
    return num%2==0


L=[ ]
i=0
while(i<=4):
    n=int(input("Enter the number: "))
    L.insert(i,n)
    i=i+1
print(L)

i=0
while(i<=4):
    data=L[i]
    choice=even(data)
    if choice==True:
        print(L[i])

    i=i+1
```

syntax: filter(fun_name, list of element)


```python
def even(num):
    return num%2==0


L=[ ]
i=0
while(i<=4):
    n=int(input("Enter the number: "))
    L.insert(i,n)
    i=i+1
print(L)


k=list(filter(even,L))
print(k)
```

```python
def even(num):
    return num%2==0


L = []
for i in range(5):
    n = int(input("Enter the number: "))
    L.append(n)
print(L)


k=list(filter(even,L))
print(k)
```

```python
add = lambda a, b: a + b
sub = lambda a, b: a - b
multi = lambda a, b: a * b
div=lambda a , b: a / b

a = int(input("Enter first number"))
b = int(input("Enter second number : "))

print("Addition:", add(a, b))
print("Subtraction:", sub(a, b))
print("Multiplication:", multi(a, b))
print("division ", div(a,b))
```
