Write a program to display the LCM of given two numbers using Recursion 
 n1=4 n2 =6

logic : if lcm %n1==0 and lcm %n2==0:
            return lcm

base cond: if n1*n2 ==lcm:
            ret lcm

para  : n1, n2, lcm

```python
def Lcm(n1,n2,lcm):
    if (n1*n2==lcm):
        return lcm


    if lcm%n1==0 and lcm%n2==0:
        return lcm 
    return Lcm(n1, n2, (lcm+1))

n1=int(input("Enter the first number : "))
n2=int(input("Enter the second number : "))

lcm=n1
if n2>n2:
    lcm=n2 

res=Lcm(n1,n2, lcm)
print("The LCM : ", res)

        
```
