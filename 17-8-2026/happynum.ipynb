```python
def happyNum(n):

    if n == 1:
        return True
    elif n == 4:
        return False

    sum = 0

    while n > 0:
        base = n % 10
        sum = sum + (base * base)
        n = n // 10

    return happyNum(sum)


n = int(input("Enter a num: "))

if happyNum(n):
    print(n, "is a happy num")
else:
    print(n, "is not a happy num")
```
