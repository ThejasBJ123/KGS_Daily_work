## Theory & Key Concepts Used in the Codes

1. **String Slicing**: `str[start:end:step]` for sub-string extraction.
2. **Built-in String Methods**: `upper()`, `lower()`, `startswith()`, `endswith()`, `split()`, `replace()`.
3. **String Masking**: Obfuscating sensitive data (Aadhaar, Bank Account) using wildcard characters (`*`).
4. **Validation Logic**: Conditional checks for data formats (Email, Password, OTP, License Plate).
