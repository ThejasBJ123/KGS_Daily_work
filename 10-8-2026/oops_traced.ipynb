## Section 3: convert lower case to upper case using while loop

### ❓ Question / Problem Statement
**Problem Statement (convert lower case to upper case using while loop): Implement and trace the Python program for convert lower case to upper case using while loop.**

--- 
### 💻 My Code

```python
# convert lower case to upper case using while loop
str=input("Enter the string : ")
newstr=" "
i=0
while(i<=len(str)-1):
    data =str[i]
    ASCII=ord(data)
    if (ASCII >= 97 and ASCII<=122 ):
        Newascii=ASCII-32

        convchar=chr(Newascii)
        newstr=newstr+convchar

    else:
        newstr=newstr+data

    i=i+1
print(newstr)
```

## Section 7: str = input("Enter the string: ")

### ❓ Question / Problem Statement
**Problem Statement (str = input("Enter the string: ")): Implement and trace the Python program for str = input("enter the string: ").**

--- 
### 💻 My Code

```python
str = input("Enter the string: ")
newstr = ""
i = 0

while i <= len(str) - 1:
    data = str[i]
    ASCII = ord(data)

    # lowercase → uppercase
    if ASCII >= 97 and ASCII <= 122:
        Newascii = ASCII - 32
        convchar = chr(Newascii)
        newstr = newstr + convchar

    # uppercase → lowercase
    elif ASCII >= 65 and ASCII <= 90:
        Newascii = ASCII + 32
        convchar = chr(Newascii)
        newstr = newstr + convchar

    # spaces, numbers, special characters
    else:
        newstr = newstr + data

    i = i + 1

print(newstr)
```

### 🧠 Logic & Explanation
1. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.
2. **Conditional Branching (`if-else`):** Evaluates boolean expressions to branch execution path.
