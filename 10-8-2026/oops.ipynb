```python
# convert lower case to upper case using while loop
str=input("Enter the string : ")
newstr=" "
i=0
while(i<=len(str)-1):
    data =str[i]
    ASCII=ord(data)
    if (ASCII >= 97 and ASCII<=122 ):
        Newascii=ASCII-32

        convchar=chr(Newascii)
        newstr=newstr+convchar

    else:
        newstr=newstr+data

    i=i+1
print(newstr)
```

```python
# convert lower case to upper case using for loop
str=input("Enter the string : ")
newstr=" "
for i in str:
    ASCII=ord(i)
    if (ASCII >= 97 and ASCII<=122 ):
        Newascii=ASCII-32

        convchar=chr(Newascii)
        newstr=newstr+convchar

    else:
        newstr=newstr+i


print(newstr)
```

```python
str = input("Enter the string: ")
newstr = ""
i = 0

while i <= len(str) - 1:
    data = str[i]
    ASCII = ord(data)

    # lowercase → uppercase
    if ASCII >= 97 and ASCII <= 122:
        Newascii = ASCII - 32
        convchar = chr(Newascii)
        newstr = newstr + convchar

    # uppercase → lowercase
    elif ASCII >= 65 and ASCII <= 90:
        Newascii = ASCII + 32
        convchar = chr(Newascii)
        newstr = newstr + convchar

    # spaces, numbers, special characters
    else:
        newstr = newstr + data

    i = i + 1

print(newstr)
```
