a number is set to be arm strong if and only if the original number is same as the sum of exponent value (where the base is the individual digit and power is count of digits).

```python
# wap in dispaly the asn number with customized function
def CountDigit(n):
    count=0
    while n>0:
        n=n//10
        count+=1
    return count

def isASN(n):
    temp=n
    if n<0:
        n=n*-1

    asn, pow=0, CountDigit(n)

    while n>0:
        base =n%10
        asn=asn+(base**pow)
        n=n//10

    if temp<0:
        asn=asn*-1

    return temp==asn


num=int(input("Enter the number :"))
if isASN(num):
    print(num,"is a ASN")
else:
    print(num,"is not a ASN")
```

## Theory & Key Concepts Used in the Codes

1. **Armstrong Number**: A number equal to the sum of its digits raised to the power of total digits (e.g. $153 = 1^3 + 5^3 + 3^3$).
2. **Power Operator (`**`)**: `digit ** count` computes power of digit.
3. **Digit Extraction & Reduction**: `digit = n % 10` gets digit, `n = n // 10` drops digit.
4. **Number Reversal**: `rev = rev * 10 + digit` constructs reverse integer.
5. **Palindrome Number**: Number equal to its reverse (`num == rev`).
