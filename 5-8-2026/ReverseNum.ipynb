### Question 2
Write a Python program for: **wap to reverse a number present in a user defined range**

```python
# wap to reverse a number present in a user defined range
def integerReversal(n):
    temp=n
    if n<0:
        n=n*-1
    rev=0
    
    while n>0:
        rem=n%10
        rev=(rev*10)+rem
        n=n//10

    if temp<0:
        rev=rev*-1
    
    return rev


sr=int(input("enter the  starting range : "))
er= int(input("enter the ending range: "))

if sr>er:
    print("invalid numebr")

else:
    print(" the reverse number : ")
    for i in range (sr, er+1):
        res = integerReversal(i)
        print(res ,end=" ")
```

### Question 3
Write a Python program for: **wap to display the first n natural num in reve order**

```python
# wap to display the first n natural num in reve order 

# eg: n=15
# 1 2 3 4 5 6 7 7 9 1 11 21 31 41 51 

def reverNum(n):
    temp=n
    if n<0:
        n=n*-1
    rev=0

    while n>0:
        rem=n%10
        rev=(rev*10)+rem
        n=n//10

    if temp <0:
        rev=rev*-1

    return rev

count=int(input("enter the number : "))
series=1

while count>0:
    print(reverNum(series), end=" ")
    count -= 1
    series += 1
```

## Theory & Key Concepts Used in the Codes

1. **Reverse Number Logic**: Extract last digit with `% 10` and append using `rev * 10 + digit`.
2. **Floor Division (`//`)**: Truncates rightmost digit after extraction.
3. **Range Reversal**: Iterating range in descending order or displaying series in reverse.
