```python
# 1. cheak weather the number is palindrome or not 
def num_Palindrome(n):
    temp=n
    if n<0:
        n=n*-1
    rev=0
    
    while n>0:
        rem=n%10
        rev=(rev*10)+rem
        n=n//10

    if temp<0:
        rev=rev*-1

    return rev==temp


num=int(input("enter the numbr : "))
if num_Palindrome(num):
    print("the palindrome number is",num)

else:
    print("not a palindrome number ", num )
  
```
