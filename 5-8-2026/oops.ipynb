```python
class Inters:
    def __init__(self, intern_name, intern_id, intern_age, intern_mob):
        self.name=intern_name
        self.Iid=intern_id
        self.age=intern_age
        self.mob=intern_mob

    def Intern_details(self):
        print(self.name)
        print(self.Iid)
        print(self.age)
        print(self.mob)

name=input("Enter the name: ")
Iid=input("enter the id: ")
age=int(input("Enter the age : "))
mob=int(input("enter the mobile : "))


i1=Inters(name,Iid,age,mob)
i1.Intern_details()
```

```python
class Farmer:
    r=2.5
    def __init__(self, p, t):
        self.p=p
        self.t=t
    
    def Simple_calculatore(self):
        si=(self.p*self.t*self.r)/100
        print(si)

s1=Farmer(1000,2)
s1.Simple_calculatore()
```
