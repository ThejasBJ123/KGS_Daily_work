```python
def factor(n):
    for i in range(1,n+1):
        if n%i==0:
            print(i,end=" ")

def countfactor(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count+=1
    return count

def countcycle(n):
    countc=0
    for i in range(1, n+1):
        countc+=1
    return countc

def prime(n):
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count+=1
    return count==2

num=int(input("Enter the number : "))

print("countfactor: ")
factor(num)

r1= countfactor(num)
print("countdigit:",r1)

c1=countcycle(num)
print("countcycle: ",c1)


if prime(num):
    print("prime")
else:
    print("not ")
```

```python
def factor(n,i):
    if i>n:
        return
    if n%i==0:
        print(i,end=" ")
    factor(n,i+1)

def countfactor(n,i,count):
    if i>n:
        return count 
    if n%i==0:
        count+=1
    return countfactor(n,i+1,count)

def countcycle(n,i,countc):
    if i>n:
        return countc
    if n%i==0:
        countc+=1
    return countcycle(n,i+1,countc)

def prime(n,i,count):
    if i>n:
        return count==2
    if n%i==0:
        count+=1
    return prime(n, i+1,count)

num=int(input("Enter the number : "))

print("countfactor: ")
factor(num,1)

r1= countfactor(num, 1,0)
print("countdigit:",r1)

c1=countcycle(num,1,0)
print("countcycle: ",c1)


if prime(num,1,0):
    print("prime")
else:
    print("not ")
    
```
