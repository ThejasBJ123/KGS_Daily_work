```python
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    for j in range(n,i-1,-1):
        if i==1 or j==n or i==j:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
# Hollow Half-Diamond LHS (Right-Pointing Arrow)
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for j in range(1, noc+1):
        if j==1 or j==noc:
            
            print("*",end="")
        else:
            print(" ",end="")
    print()
    if i<n:
        noc+=1
    else:
        noc-=1
```

```python
# Hollow Half-Diamond LHS (Right-Pointing Arrow)
n = int(input("Enter the number: "))
noc = n
for i in range(1, n * 2):
    for k in range(n,noc,-1):
        print(" ",end="")
    for j in range(1,noc+1):
        if i==1 or i==(n*2)-1 or j==noc or j==1:
            print("*",end="")
        else:
            print(" ",end="")
    print()
    if i<n:
        noc-=1
    else:
        noc+=1
```
