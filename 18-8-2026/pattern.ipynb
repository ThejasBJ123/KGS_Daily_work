```python
# LHS Right angle triangle
n=int(input("enter the number :" ))
for i in range(1, (n+1)):
    for j in range(1, (i+1)):
        print("*", end="")

    print() 


```

```python
# LHS inverted right angle triangle
n=int(input("Enter the num : "))
for i in range(n, (1-1), -1):
    for j in range(1,(i+1)):
        print("*", end=" ")

    print()
```
