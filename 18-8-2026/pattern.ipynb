```python
# LHS inverted right angle triangle
n=int(input("Enter the num : "))
for i in range(n, (1-1), -1):  # (n, 0, -1)
    for j in range(1,(i+1)):
        print("*", end="")

    print()
```

```python

# right angle as triangle
n=int(input("enter the num: "))

for i in range(1,(n+1)):
    for k in range(n,i,-1):
        print(" ",end="")
    for j in range(1,(i+1)):
        print("*",end="")
    print()
```

```python
# RHS inverted right angle

for i in range(1, n+1):
    # spaces
    for k in range(1, i):
        print(" ", end="")
    # stars
    for j in range(n, i-1, -1):
        print("*", end="")
    print()
```

```python
# Pyramid    RHS right angle triangle

n = int(input("Enter the number: "))

for i in range(1, n + 1):
    for k in range(n, i, -1):
        print(" ", end="")
    for j in range(1, i + 1):
        print("* ", end="")
    print()
```

```python
# Pyramid  inverted    
# RHS inverted right angle
n = int(input("Enter the number: "))
for i in range(1, n+1):
    # spaces
    for k in range(1, i):
        print(" ", end="")
    # stars
    for j in range(n, i-1, -1):
        print("* ", end="")
    print()
```

```python
# special pyramid 
# Pyramid    RHS right angle triangle  *2

n=int(input("enter the num : "))
for i in range(1,(n+1)):
    for k in range(n, i, -1):
        print(" ",end="")
    for j in range(1,(i*2)):
        print("*",end="")

    print()
 
```

```python
# special inverted pyramid

n = int(input("Enter the number: "))

for i in range(1, n + 1):

    for k in range(n, i, -1):
        print(" ", end="")

    for j in range(1, 2 * i):
        print("*", end="")
    print()

```

```python
# LHS  parallelogram

n = int(input("Enter the number: "))

for i in range(1,n+1):       # i decreases
    for k in range(1, i+1):   # spaces increase
        print(" ", end="")

    for j in range(1, n+1):     # j increases
        print("*", end="")

    print()
```

```python
# LHS parallelogram

n = int(input("Enter the number: "))

for i in range(n, 0, -1):       # i decreases
    for k in range(1, n-i+1):   # spaces increase
        print(" ", end="")

    for j in range(1, n+1):     # j increases
        print("*", end="")

    print()
```

```python
# RHS Parallelgram

n = int(input("Enter the number: "))

for i in range(1,n+1):       
    for k in range(n, i-1,-1):   
        print(" ", end="")

    for j in range(1, n+1):     
        print("*", end="")

    print()
```

```python
# combinational pattern  LHS
n=int(input("Enter the num: "))
noc=1
for i in range(1,(n*2)):
    for j in range(1, (noc+1)):
        print("*",end="")
    print()
    if i<n:
        noc+= 1
    else:
        noc-=1
```

```python
# combinational pattern RHS
n=int(input("Enter the num: "))
noc=1
for i in range(1,(n*2)):
    for k in range(n,noc,-1):
        print(" ",end="")
    for j in range(1, (noc+1)):
        print("*",end="")
    print()
    if i<n:
        noc+= 1
    else:
        noc-=1

```
