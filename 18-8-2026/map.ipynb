```python
def operation(data):
    val=data+10
    return val

L=[]
i=0
while (i<=4):
    n=int(input("Enter the num: "))
    L.insert(i,n)
    i=i+1

print(L)
k=list(map(operation, L))
print(k)
```

```python
L=[]
i=0
while (i<=4):
    n=int(input("Enter the num: "))
    L.insert(i,n)
    i=i+1
print(L)

k=list(map(lambda data : data+10, L))
print(k)
```
