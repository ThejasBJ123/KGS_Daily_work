# Happy numbers

- Recursion Logic 
------------------

logic:  sum=0
        while n>0:
            base=n%10
            sum=sum+(base*base)
            n=n//10

para:   n

base:   if n==1:
            return True
        elif n==4:
            return False
        

    

```python
def happyNum(n):

    if n == 1:
        return True
    elif n == 4:
        return False

    sum = 0

    while n > 0:
        base = n % 10
        sum = sum + (base * base)
        n = n // 10

    return happyNum(sum)


n = int(input("Enter a num: "))

if happyNum(n):
    print(n, "is a happy num")
else:
    print(n, "is not a happy num")
```

Write a program to display the LCM of given two numbers using Recursion 
 n1=4 n2 =6

logic : if lcm %n1==0 and lcm %n2==0:
            return lcm

base cond: if n1*n2 ==lcm:
            ret lcm

para  : n1, n2, lcm

```python
def Lcm(n1,n2,lcm):
    if (n1*n2==lcm):
        return lcm


    if lcm%n1==0 and lcm%n2==0:
        return lcm 
    return Lcm(n1, n2, (lcm+1))

n1=int(input("Enter the first number : "))
n2=int(input("Enter the second number : "))

lcm=n1
if n2>n2:
    lcm=n2 

res=Lcm(n1,n2, lcm)
print("The LCM : ", res)

        
```
