# NUMBER PATTERN

### Number patterns :
--------------------
* for every updated row, if the complete row is filled wiht the same value then it is displaying the value present in the variable "i" 
* for every updated row, if the displaying value is starting from same point and moving to certain point or starting at certain point and moving to the same point then it is displaying the value present in the variable "j".
* If the complete pattern is diplaying the val's in continous series then it is using 3rd variable.


```python
# LHS right angle triangle 
n=int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(1, (i+1)):
        print(i, end=" ")
    print()
```

```python
# LHS right angle triangle

for i in range(1, n+1):
    for j in range(1, (i+1)):
        print(j, end=" ")
    print()
```

```python
# LHS right angle triangle

n=int(input("Enter the number: "))
for i in range(n,0,-1):
    for j in range(n, i-1,-1):
        print(i, end=" ")
    print()
```

```python
# combinational patter

n = int(input("Enter the number: "))

noc=1
for i in range(1, n *2 ):
    for j in range(1,noc+1):
        print(j,end=" ")

    print()

    if i<n:
        noc+=1
    else:
        noc-=1
```

```python
# combinational pattern

n = int(input("Enter the number: "))

noc = n

for i in range(1, n * 2):

    for j in range(noc, n + 1):
        print(j, end=" ")

    print()

    if i < n:
        noc -= 1
    else:
        noc += 1
```
