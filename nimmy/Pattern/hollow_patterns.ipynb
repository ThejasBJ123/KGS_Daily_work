# Hollow Patterns (Allow Patterns)

A **Hollow Pattern** displays only the outer border/edges of geometric shapes (triangles, pyramids, diamonds, butterfly, parallelograms) while leaving internal spaces empty.

### Key Logic for Hollow Patterns:
- Use boundary conditions in an `if` statement to print `*`:
  - First / Last Row (`i == 1` or `i == n` or `i == (n*2)-1`)
  - First / Last Column (`j == 1` or `j == n` or `j == noc`)
  - Diagonals (`i == j` or `noc == j` or `nor == j`)
- Print space `" "` inside empty areas.


### Question 1: Hollow LHS Right-Angled Triangle
**Problem Statement:**
Write a Python program to print a hollow right-angled triangle aligned to the Left-Hand Side (LHS) for $n$ rows.

**Sample Input:** $n = 5$
**Expected Output:**
```text
* 
* * 
*   * 
*     * 
* * * * * 
```


````python
# Hollow LHS Right-Angled Triangle
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    for j in range(1, i + 1):
        if i == n or j == 1 or i == j:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
````

### Question 2: Hollow RHS Right-Angled Triangle

### Question 6: Hollow Hourglass Pattern (LHS Combination)
**Problem Statement:**
Write a Python program to print a hollow inverted half-diamond (hourglass shape on LHS) for $n$ rows ($2n-1$ total lines).

**Sample Input:** $n = 5$
**Expected Output:**
```text
* * * * * 
*     * 
*   * 
* * 
* 
* * 
*   * 
*     * 
* * * * * 
```


````python
# Hollow Hourglass / Combination Pattern LHS
n = int(input("Enter the number: "))
noc = n
for i in range(1, n * 2):
    for j in range(1, noc + 1):
        if j == 1 or i == 1 or i == (n * 2) - 1 or j == noc:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
    if i < n:
        noc -= 1
    else:
        noc += 1
````

### Question 11: Hollow Butterfly Pattern
**Problem Statement:**
Write a Python program to print a hollow butterfly pattern for a given size $n$.

**Sample Input:** $n = 5$
**Expected Output:**
```text
*               * 
* *           * * 
*   *       *   * 
*     *   *     * 
*       *       * 
*     *   *     * 
*   *       *   * 
* *           * *
*               * 
```


````python
# Hollow Butterfly Pattern
n = int(input("Enter the number: "))
noc = 1
nor = (n * 2) - 1
for i in range(1, n * 2):
    for j in range(1, n * 2):
        if j == 1 or j == (n * 2) - 1 or noc == j or nor == j:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
    if i < n:
        noc += 1
        nor -= 1
    else:
        noc -= 1
        nor += 1
````

### Question 12: Hollow Parallelogram (LHS Spacing)
**Problem Statement:**
Write a Python program to print a hollow parallelogram slanting to the right for $n$ rows.

**Sample Input:** $n = 5$
**Expected Output:**
```text
 * * * * * 
  *       * 
   *       * 
    *       * 
     * * * * * 
```


````python
# Hollow Parallelogram (LHS Spacing)
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    for k in range(1, i + 1):
        print(" ", end="")
    for j in range(1, n + 1):
        if i == n or i == 1 or j == 1 or j == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
````
