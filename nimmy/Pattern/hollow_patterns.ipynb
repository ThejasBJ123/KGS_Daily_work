# Hollow Patterns

A **Hollow Pattern** displays only the outer border/edges of geometric shapes (lines, squares, triangles, pyramids, parallelograms, arrows, diamonds, hourglasses, butterfly) while leaving internal spaces empty.

### Key Logic for Hollow Patterns:
- Use boundary conditions in an `if` statement to print `*`:
  - First / Last Row (`i == 1` or `i == n` or `i == (n*2)-1`)
  - First / Last Column (`j == 1` or `j == n` or `j == noc`)
  - Diagonals (`j == noc` or `j == nor` or `j == i * 2 - 1`)
- Print space `" "` inside empty areas.


### Question 1: Hollow Vertical Line Star Pattern
**Problem Statement:**
Write a Python program to print $n$ stars vertically.

**Sample Input:** $n = 5$
**Expected Output:**
```text

*

```


````python
# Hollow Vertical Line Pattern (Stars)
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    print("*")
````

### Question 4: Hollow Square Star Pattern
**Problem Statement:**
Write a Python program to print an $n \times n$ hollow square grid of stars.

**Sample Input:** $n = 5$
**Expected Output:**
```text
* * * * * 
*       * 
*       * 
*       * 
* * * * * 
```


````python
# Hollow Square Star Pattern
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    for j in range(1, n + 1):
        if i == 1 or i == n or j == 1 or j == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
````

### Question 16: Hollow Half-Diamond LHS (Right-Pointing Arrow)
**Problem Statement:**
Write a Python program to print a hollow right-pointing arrow / half-diamond pattern on LHS for $n$ rows.

**Sample Input:** $n = 5$
**Expected Output:**
```text
*
**
* *
*  *
*   *
*  *
* *
**
*
```


````python
# Hollow Half-Diamond LHS (Right-Pointing Arrow)
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for j in range(1, noc + 1):
        if j == 1 or j == noc:
            print("*", end="")
        else:
            print(" ", end="")
    print()
    if i < n:
        noc += 1
    else:
        noc -= 1
````

### Question 24: Hollow Butterfly Pattern
**Problem Statement:**
Write a Python program to print a hollow butterfly pattern for size $n$.

**Sample Input:** $n = 5$
**Expected Output:**
```text
*               * 
* *           * * 
*   *       *   * 
*     *   *     * 
*       *       * 
*     *   *     * 
*   *       *   * 
* *           * * 
*               * 
```


````python
# Hollow Butterfly Pattern
n = int(input("Enter the number: "))
noc = 1
nor = (n * 2) - 1
for i in range(1, n * 2):
    for j in range(1, n * 2):
        if j == 1 or j == (n * 2) - 1 or j == noc or j == nor:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
    if i < n:
        noc += 1
        nor -= 1
    else:
        noc -= 1
        nor += 1
````
