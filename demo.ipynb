```python
def reverse(n):

    temp=n   
    if n<0:
        n=n*-1


    rev=0
    while n>0:
        base=n%10
        rev=(rev*10)+base
        n=n//10

    if temp<0:
        rev=rev*-1

    return rev

num=int(input("Enter the num: "))
res = reverse(num)
print(" reverse the num", num,"is : ", res)
```

```python
def palindrome(n):

    temp=n
    if n<0:
        n*=-1
    
    pal=0
    while n>0:
        base=n%10
        pal=(pal*10)+base
        n=n//10

    if temp<0:
        pal*= -pal

    return pal==temp

num=int(input("enter the number : "))
if palindrome(num):
    print("this the palindrome: ", num)

else:
    print("this is not palindrome: ", num)
```
