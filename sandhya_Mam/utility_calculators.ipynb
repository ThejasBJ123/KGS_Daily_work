GST Calculator:

Different products require GST calculation.
Create a function that accepts price and GST percentage.
Return the final amount.

### Question 3
Write a Python program for: **Python Program Implementation 3**

```python
def cal_gst(price, gst_p):
    gst_a=(price*gst_p)/100
    final=gst_a+price

    return final


p=int(input(" enter the price of the product : "))
g=int(input(" enter the gst percentage: "))

amount=cal_gst(p,g)
print("final amount: ", amount)
```

### Question 4
Write a Python program for: **Python Program Implementation 4**

```python
def cal_bmi(weight, height):
    Bmi=weight/(height*height)
    return Bmi

w=float(input("enter the weight of the body (Kg): "))
h=float(input("enter the height [meter] : "))

result=cal_bmi(w,h)

print("output :", result)


```
