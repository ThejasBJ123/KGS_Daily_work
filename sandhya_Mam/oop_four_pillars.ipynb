### Four Pillars of Python OOP
1. **Encapsulation**: Wrapping data and methods inside a class and restricting direct access.
2. **Inheritance**: Acquiring properties and behaviors from a parent class.
3. **Polymorphism**: Ability to take multiple forms (same method name, different behaviors).
4. **Abstraction**: Hiding internal implementation details and showing only essential features.

### Question 1: Encapsulation - Direct Public Attribute Access
Write a Python program to demonstrate direct attribute modification without encapsulation restrictions.

```python
# Encapsulation - Public Attribute Access
class Book:
    def __init__(self, value):
        self.pages = value


b1 = Book(100)
print("Initial Pages:", b1.pages)

b1.pages = 90
print("Modified Pages:", b1.pages)

b1.pages = -99  # Invalid value accepted due to direct access
print("Invalid Pages:", b1.pages)
```

### Question 2: Encapsulation - Private Attributes with Getter & Setter Methods
Write a Python program implementing private attributes (`__pages`) with getter and setter validation logic.

```python
class Book:
    def __init__(self, value):
        self.__pages = value

    def setter(self, value):
        if value > 0:
            self.__pages = value
        else:
            print("Invalid page count! Must be positive.")

    def getter(self):
        return self.__pages


b1 = Book(100)
b1.setter(200)
print("Updated Pages:", b1.getter())

b1.setter(-99)
print("Current Pages after invalid set:", b1.getter())
```

### Question 3: Private Attributes - Person Class Implementation
Write a Python program using private variable `__name` with getter and setter methods.

```python
class Person:
    def __init__(self):
        self.__name = ""

    def setter(self, name):
        self.__name = name

    def getter(self):
        return self.__name


p1 = Person()
p1.setter("rama")
print("Person Name:", p1.getter())
```

### Question 4: Encapsulation - Using `@property` and `@setter` Decorators
Write a Python program to access and modify private attributes pythonically using property decorators.

```python
# Property Decorator Implementation
class Person:
    def __init__(self):
        self.__name = ""

    @property
    def dataaccess(self):
        return self.__name

    @dataaccess.setter
    def dataaccess(self, name):
        self.__name = name


p1 = Person()
p1.dataaccess = "rama"   # Calls setter decorator
print("Accessed via Property:", p1.dataaccess)   # Calls getter decorator
```

### Question 5: Single-Level Inheritance - Parent and Child Classes
Write a Python program demonstrating single-level inheritance where child inherits parent attributes.

```python
class Parent:
    def __init__(self):
        self.a = 10


class Child(Parent):
    def __init__(self):
        super().__init__()
        self.b = 20


c1 = Child()
print("Child Variable b:", c1.b)
print("Inherited Parent Variable a:", c1.a)
```

### Question 6: Multilevel Inheritance - Multi-tier Class Hierarchy
Write a Python program demonstrating multilevel inheritance across classes A, B, and C.

```python
class A:
    def __init__(self):
        self.a = 10


class B(A):
    def __init__(self):
        super().__init__()
        self.b = 20


class C(B):
    def __init__(self):
        super().__init__()
        self.c = 30


c1 = C()
print("Level C Attribute c:", c1.c)
print("Level B Attribute b:", c1.b)
print("Level A Attribute a:", c1.a)
```

### Types of Inhertance

* single inheritance
* multilevel inheritance
* hierarchial inheritance
* multiple inheritance
* hybrid inheritance


```python
# single inheritance

class A:
    def dispA(self):
        print("A")


class B(A):
    def dispB(self):
        print("B")
        A.dispA(self)


b1 = B()
b1.dispB()
b1.dispA()
```

```python
#  multilevel inheritance

class A:
    def dispA(self):
        print("A")
    
class B(A):
    def dispB(self):
        print("B")

class C(B):
    def dispC(self):
        print("c")

c1=C()
c1.dispC()
c1.dispB()
c1.dispA()

    
```

```python
# hierarchial inheritance
class A:
    def dispA(self):
        print("A")

class B(A):
    def dispB(self):
        print("B")

class C(A):
    def dispC(self):
        print("C")

b1 = B()
b1.dispB()
b1.dispA()

c1 = C()
c1.dispC()
c1.dispA()
```

```python
# multiple inheritance
class A:
    def dispA(self):
        print("A")

class B:
    def dispB(self):
        print("B")

class C(A, B):
    def dispC(self):
        print("C")

c1 = C()
c1.dispC()
c1.dispA()
c1.dispB()
```

```python
# hybrid inheritance
class A:
    def dispA(self):
        print("A")

class B(A):
    def dispB(self):
        print("B")

class C(A):
    def dispC(self):
        print("C")

class D(B, C):
    def dispD(self):
        print("D")

d1 = D()
d1.dispD()
d1.dispB()
d1.dispC()
d1.dispA()
```

### Super Key

```python
# invoking a parent class construction manually
# super().__init__()

class A:
    def __init__(self):
        self.a=10

class B(A):
    def __init__(self): 
        super().__init__()
        self.b=20

class C(B):
    def __init__(self):
        super().__init__()
        self.c1=30

cf=C()
print(cf.c1)
print(cf.b)
print(cf.a)
```

# **Polymorphism**: 

```python
class plane:
    def takeoff(self):
        print("plane is takeoff")
    def land(self):
        print("plane is landing")
    def fly(self):
        print("Plane is flying ")

class Cargo(plane):
    def takeoff(self):
        print("cargo is takeoff")
    def land(self):
        print("cargo is landing")
    def fly(self):
        print("Cargo is flying")

class Passenger(plane):
    def takeoff(self):
        print("Passenger is takeoff")
    def land(self):
        print("Passenger is landing ")
    def fly(self):
        print("Passenger is flying ")

c=Cargo()
p=Passenger()
def allowplane(ref):
    ref.takeoff()
    ref.land()
    ref.fly()

allowplane(c)
allowplane(p)
    
```
