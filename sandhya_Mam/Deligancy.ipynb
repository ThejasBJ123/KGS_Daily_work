```python
class Radio:
    def turnon(self,x):
        if x==1:
            print("Radio is on")
        else:
            print("Radio is off")

class car:
     def __init__(self, min, max):
        self.cmin=min
        self.cmax=max
        self.r=Radio()

c=car(60,180)
print(c.cmin)
print(c.cmax)
c.r.turnon(1)
c.r.turnon(2)
```
