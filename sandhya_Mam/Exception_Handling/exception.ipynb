# exception handling 

```python
# exception handling 
a=int(input("Enter a n1"))
b=int(input("Enter a n2"))
try:
    c=a/b
    print(c)
except Exception as e:
    print("Error is occured")
else:
    print("leaving a program")
```

```python
def fun1():
    print("entering func1")
    try:
        fun2()
    except Exception as e:
        print("Error in fun1")
    print("leaving  fun1")

def fun2():
    print("entering to fun 2")
    a=a/b
    print(a)
    print("leaving the fun2")

print("program started")
fun1()
print("program end")
```
