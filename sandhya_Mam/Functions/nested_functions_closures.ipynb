```python
def Outer():
    print("Entering outer")
    print("Processing")

    def inner():
        print("Entering inner")
        print("processing")
        print("leaving inner")

    print("calling inner")
    inner()


print("program started")
print("calling outer")
Outer()
print("program end")
```
