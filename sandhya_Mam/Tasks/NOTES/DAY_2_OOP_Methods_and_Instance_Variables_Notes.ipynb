## 1. Standard Way of Writing a Class

* **Class:** Blueprint / Template used to create objects.
* **Object:** Physical instance of a class occupying memory in the Heap.
* **Class Anatomy:**
  * **Properties / Variables (`has`):** Characteristics of the object (e.g., `breed`, `color`).
  * **Behaviours / Methods (`does`):** Actions performed by the object (e.g., `bark()`, `sleep()`).

### Standard Rules:
1. Class name must start with a **Capital letter** (`class Dog:`).
2. Class body must have **1 tab space (4 spaces)** indentation.
3. `self` parameter holds the **memory address of the current object** in the Heap.

### Steps in Object Creation:
1. PVM allocates memory in Heap.
2. PVM searches for `__init__()` constructor.
3. Initializes variables using `self`.
4. Returns the object address to the reference variable.


```python
# Standard Way of Writing a Class
class Dog:
    # Constructor (State / Properties)
    def __init__(self, breed, color):
        self.breed = breed
        self.color = color

    # Methods (Behaviours)
    def bark(self):
        print(f"{self.breed} is barking: Woof!")

    def sleep(self):
        print(f"{self.breed} is sleeping: Zzz...")

# Creating an Object
d1 = Dog("Labrador", "Black")
print(f"Dog: {d1.breed}, Color: {d1.color}")
d1.bark()
d1.sleep()
```

## 2. Ways of Declaring Instance Variable

An **Instance Variable** belongs to a specific object. There are **3 ways** to declare it:

| # | Location | Syntax | When Memory is Allocated |
| :---: | :--- | :--- | :--- |
| **1** | **Inside Constructor (`__init__`)** | `self.var = val` | At the time of object creation |
| **2** | **Inside Instance Method** | `self.var = val` | Only when that method is invoked |
| **3** | **Outside the Class** | `obj_ref.var = val` | Dynamically when assigned outside |


```python
# 3 Ways of Declaring Instance Variables
class Demo:
    def __init__(self):
        # 1. Inside Constructor (__init__)
        self.x = 10
        self.y = 20

    def create_z(self):
        # 2. Inside Instance Method
        self.z = 30

# Creating Object
d1 = Demo()

# Invoking method to declare z
d1.create_z()

# 3. Outside Class using object reference
d1.a = 40

print(f"Declared in __init__: x={d1.x}, y={d1.y}")
print(f"Declared in Method:   z={d1.z}")
print(f"Declared Outside:     a={d1.a}")
```

## 3. Accessing Instance Variable

Instance variables can be accessed in **two locations**:

1. **Inside the Class:** Using `self.<variable_name>`
2. **Outside the Class:** Using `<object_reference>.<variable_name>`

> [!NOTE]
> Trying to access an instance variable before it is initialized will raise an `AttributeError`.


```python
# Accessing Instance Variables Inside and Outside Class
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        # 1. Accessing INSIDE the class using self
        print(f"Inside Class  -> Name: {self.name}, Age: {self.age}")

s1 = Student("Raju", 23)

# Accessing inside class
s1.display()

# 2. Accessing OUTSIDE the class using object reference
print(f"Outside Class -> Name: {s1.name}, Age: {s1.age}")

# Modifying outside class
s1.age = 24
print(f"Modified Age  -> Age: {s1.age}")
```

## 4. Methods and its Types

Python has **3 types of methods**:

1. **Instance Method:**
   * Uses `self` as the first parameter.
   * Accesses and modifies object instance variables.
   * Called using: `object_ref.method_name()`
2. **Static Method (`@staticmethod`):**
   * General utility method; does not take `self` or `cls`.
   * Called using: `ClassName.method_name()` or `obj.method_name()`
3. **Class Method (`@classmethod`):**
   * Takes `cls` as the first parameter.
   * Accesses and modifies class-level / static variables.
   * Called using: `ClassName.method_name()`

---

### 4 Variations of Instance Methods:
1. **No Parameter, No Return Value**
2. **No Parameter, With Return Value**
3. **With Parameter, No Return Value**
4. **With Parameter, With Return Value**


```python
# Methods and its Types Demo
class Demo:
    class_var = 10  # Class / Static Variable

    def __init__(self):
        self.inst_var = 100  # Instance Variable

    # 1. Instance Method
    def instance_method(self):
        print(f"Instance Method -> Instance Var: {self.inst_var}")

    # 2. Static Method
    @staticmethod
    def static_method():
        print("Static Method   -> General utility (No self/cls)")

    # 3. Class Method
    @classmethod
    def class_method(cls):
        print(f"Class Method    -> Class Var: {cls.class_var}")

# 4 Variations of Instance Methods
class Calculator:
    # 1. No Param, No Return
    def add_v1(self):
        a, b = 10, 20
        print(f"V1 (No Param, No Return): {a + b}")

    # 2. No Param, With Return
    def add_v2(self):
        a, b = 10, 20
        return a + b

    # 3. With Param, No Return
    def add_v3(self, a, b):
        print(f"V3 (With Param, No Return): {a + b}")

    # 4. With Param, With Return
    def add_v4(self, a, b):
        return a + b

d1 = Demo()
d1.instance_method()
Demo.static_method()
Demo.class_method()

print("\n--- Calculator 4 Variations ---")
calc = Calculator()
calc.add_v1()
print("V2 Result:", calc.add_v2())
calc.add_v3(100, 200)
print("V4 Result:", calc.add_v4(100, 200))
```
