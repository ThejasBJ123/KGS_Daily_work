## 4. Compilation vs Interpretation & PVM

### 🔄 Difference:
* **Compiler:** Executes / translates the **whole code at once** into machine code.
* **Interpreter:** Executes code **line by line**.

### ⚙️ PVM (Python Virtual Machine) - "Heart of Python":
* Python source code (`.py`) is compiled into intermediate Bytecode.
* **PVM** executes the bytecode, making Python **platform independent**.

```
[ .py Source Code ] ──► [ Bytecode ] ──► [ PVM (Heart of Python) ] ──► [ Microprocessor / Output ]
```   


## 5. Organization of RAM & Hard Disk

* **Hard Disk (Permanent Storage):** Stores files (Audio, Video, Images, `.py` files).
* **RAM (Temporary Storage):** When we open/run a script (`open()` / Run), it loads into RAM for the microprocessor to execute. Saved back on `close()`.

```
┌──────────────────────────────────────┐
│       Hard Disk (Permanent)          │
│     (Audio, Video, Image, .py)       │
└──────────────────┬───────────────────┘
      loading      │  ▲ closing
      (open())     ▼  │ (save)
┌──────────────────────────────────────┐
│           RAM (Temporary)            │ ◄───► [ Microprocessor ] ──► [ Output ]
│            (ex.py -> Run)            │
└──────────────────────────────────────┘
```


## 6. Class and Object (States & Behaviours)

* **Class:** Blueprint / template to create objects.
* **Object:** An actual instance of a class.
* **State (Attributes / Data):** Properties or variables (e.g., `name`, `course`).
* **Behaviour (Methods / Functions):** Actions performed by the object (e.g., `display()`).


````python
# Simple Class and Object Example

class Student:
    # State (Attributes)
    def __init__(self, name, course):
        self.name = name
        self.course = course

    # Behaviour (Method)
    def display(self):
        print(f"Student: {self.name} | Course: {self.course}")

# Creating Objects
s1 = Student("Pooja", "Python Full Stack")
s2 = Student("Rahul", "Data Science")

s1.display()
s2.display()

````

## 8. Variables & Rules of Variables

A **Variable** is a named location used to store data in memory.

### 📌 Rules for Naming Variables:
1. Must start with a letter (`a-z`, `A-Z`) or underscore (`_`).
2. Cannot start with a digit (`1a` ❌, `a1` ✅).
3. Can only contain letters, digits, and underscores.
4. No special symbols (`@`, `$`, `-`, `%`).
5. Case-sensitive (`name` and `Name` are different).
6. Cannot use reserved keywords (`if`, `for`, `while`, `pass`, `break`).


````python
# Variable Examples
student_name = "Pooja"
age = 21
_score = 95
num1 = 50

print(student_name, age, _score, num1)

````

## 11. Looping Statements (`for`, `while`)

Used to repeat a block of code.

---

### A. `for` Loop

#### 1) Without `range`:
* `for i in "1234": print(i)` ➔ `1`, `2`, `3`, `4`
* `for i in "pooja": print(i)` ➔ `p`, `o`, `o`, `j`, `a`


````python
# for loop without range
print("Loop over '1234':")
for i in "1234":
    print(i)

print("\nLoop over 'pooja':")
Str = "pooja"
for i in Str:
    print(i)

````

#### 2) `range()` function in `for` loop:
1. **`range(stop)`** ➔ `range(5)` gives `0, 1, 2, 3, 4`
2. **`range(start, stop)`** ➔ `range(11, 16)` gives `11, 12, 13, 14, 15`
3. **`range(start, stop, step)`** ➔ `range(2, 11, 2)` gives `2, 4, 6, 8, 10`


````python
# range(stop)
print("range(5) with 'kgs':")
for i in range(5):
    print("kgs")

# range(start, stop)
print("\nrange(11, 16):")
for i in range(11, 16):
    print(i)

# range(start, stop, step)
print("\nrange(2, 11, 2):")
for i in range(2, 11, 2):
    print(i)

````

### B. `while` Loop

Repeats as long as a condition is **True**.

#### Syntax / Flow:
```python
# 1. Initialize variable
i = 0
# 2. Condition check
while (i <= 4):
    # 3. Statements
    print("python")
    # 4. Increment
    i = i + 1
print(i)  # prints 5
```


````python
# while loop example
i = 0
while i <= 4:
    print("python")
    i = i + 1

print("Final value of i:", i)

````

## 12. Jumping Statements (`break`, `continue`, `pass`)

* **`break`:** Stops and exits the loop immediately.
* **`continue`:** Skips the current iteration.
* **`pass`:** Does nothing / placeholder (ignored by interpreter).


````python
# 1. break (stops when i == 3)
print("break at i == 3:")
for i in range(5):
    if i == 3:
        break
    print(i)

# 2. continue (skips i == 3)
print("\ncontinue at i == 3:")
for i in range(5):
    if i == 3:
        continue
    print(i)

# 3. pass (ignored at i == 3)
print("\npass at i == 3:")
for i in range(5):
    if i == 3:
        pass
    print(i)

````
