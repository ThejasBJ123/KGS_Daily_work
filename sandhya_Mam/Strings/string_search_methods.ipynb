### Question 4: Core11 Exercise 4
Write a Python program for: **Core11 Exercise 4**

```python
class Demo:
    def swap(self , a, b):
        print("befor swapping")
        print(a)
        print(b)

        temp=a
        a=b
        b=temp

        print("after the swapping ")
        print(a)
        print(b)

d1=Demo()
n1=100
n2=200

d1.swap(n1,n2)
```

### Question 6: Core11 Exercise 6
Write a Python program for: **Core11 Exercise 6**

```python
class Demo:
    def add(self, a, b):
        c=a+b
        return c

    def sub(self, a, b):
        c=a-b
        return c

    def mul(self, a, b):
        c=a*b
        return c
    
d1=Demo()

x=10
y=20

r1=d1.add(x, y)
print(r1)

r2=d1.sub(x, y)
print(r2)

r3=d1.mul(x, y)
print(r3)

```

### Question 7: Core11 Exercise 7
Write a Python program for: **Core11 Exercise 7**

```python
class Demo:
     def arith_op(self, a, b):
        r1=a+b
        r2=a-b
        r3=a*b

        return r1, r2, r3

d1=Demo()

x=10
y=20


res1, res2, res3= d1.arith_op(x,y)

print(res1)
print(res2)
print(res3)
```
