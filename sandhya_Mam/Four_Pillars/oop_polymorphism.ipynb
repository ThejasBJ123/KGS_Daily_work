# Polymorphism in Python (OOP)
Polymorphism means 'many forms'. It allows different classes to implement methods with the same name, providing different implementations (Method Overriding, Loose Coupling, Polymorphic function callers).

# **Polymorphism**: 

```python
class plane:
    def takeoff(self):
        print("plane is takeoff")
    def land(self):
        print("plane is landing")
    def fly(self):
        print("Plane is flying ")

class Cargo(plane):
    def takeoff(self):
        print("cargo is takeoff")
    def land(self):
        print("cargo is landing")
    def fly(self):
        print("Cargo is flying")

class Passenger(plane):
    def takeoff(self):
        print("Passenger is takeoff")
    def land(self):
        print("Passenger is landing ")
    def fly(self):
        print("Passenger is flying ")

c=Cargo()
p=Passenger()
def allowplane(ref):
    ref.takeoff()
    ref.land()
    ref.fly()

allowplane(c)
allowplane(p)
    
```
