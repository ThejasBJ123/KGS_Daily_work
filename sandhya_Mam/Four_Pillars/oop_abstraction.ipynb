# Abstraction in Python (OOP)
Abstraction is the concept of hiding internal implementation details and showing only essential features using Abstract Base Classes (`ABC`) and `@abstractmethod`.

# **Abstraction**

```python
from abc import ABC, abstractmethod

class payment(ABC):
    @abstractmethod
    def transaction(self):
        pass

class credit(payment):
     def transaction(self):
        print("Transactions is done by using cride card")
    
class debit(payment):
     def transaction(self):
        print("Transactions is done by using debit card")


class UPI(payment):
     def transaction(self):
        print("Transactions is done by using UPI")


c=credit()
d=debit()
u=UPI()
c.transaction()
u.transaction()
d.transaction()

```
