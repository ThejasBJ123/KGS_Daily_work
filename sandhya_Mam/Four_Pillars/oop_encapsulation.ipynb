# Encapsulation in Python (OOP)
Encapsulation is the mechanism of wrapping data (variables) and code acting on data (methods) together as a single unit, and restricting direct access to internal details.

### Question 1: Encapsulation - Direct Public Attribute Access
Write a Python program to demonstrate direct attribute modification without encapsulation restrictions.

```python
# Encapsulation - Public Attribute Access
class Book:
    def __init__(self, value):
        self.pages = value


b1 = Book(100)
print("Initial Pages:", b1.pages)

b1.pages = 90
print("Modified Pages:", b1.pages)

b1.pages = -99  # Invalid value accepted due to direct access
print("Invalid Pages:", b1.pages)
```

### Question 2: Encapsulation - Private Attributes with Getter & Setter Methods
Write a Python program implementing private attributes (`__pages`) with getter and setter validation logic.

```python
class Book:
    def __init__(self, value):
        self.__pages = value

    def setter(self, value):
        if value > 0:
            self.__pages = value
        else:
            print("Invalid page count! Must be positive.")

    def getter(self):
        return self.__pages


b1 = Book(100)
b1.setter(200)
print("Updated Pages:", b1.getter())

b1.setter(-99)
print("Current Pages after invalid set:", b1.getter())
```

### Question 3: Private Attributes - Person Class Implementation
Write a Python program using private variable `__name` with getter and setter methods.

```python
class Person:
    def __init__(self):
        self.__name = ""

    def setter(self, name):
        self.__name = name

    def getter(self):
        return self.__name


p1 = Person()
p1.setter("rama")
print("Person Name:", p1.getter())
```

### Question 4: Encapsulation - Using `@property` and `@setter` Decorators
Write a Python program to access and modify private attributes pythonically using property decorators.

```python
# Property Decorator Implementation
class Person:
    def __init__(self):
        self.__name = ""

    @property
    def dataaccess(self):
        return self.__name

    @dataaccess.setter
    def dataaccess(self, name):
        self.__name = name


p1 = Person()
p1.dataaccess = "rama"   # Calls setter decorator
print("Accessed via Property:", p1.dataaccess)   # Calls getter decorator
```
