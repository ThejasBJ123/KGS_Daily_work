### Question 3: Class Method - No Parameter & No Return Value
Write a Python program for: **Class Method - No Parameter & No Return Value**

```python
# no parameter and no return value
class calculator:
    def __init__(self):
        self.brand="casio"
        self.cost=200

    def add(self):
        a=10
        b=20
        c=a+b
        print(c)
c1=calculator()
print(c1.brand)
print(c1.cost)
c1.add()
```

### Question 4: Class Method - No Parameter & With Return Value
Write a Python program for: **Class Method - No Parameter & With Return Value**

```python
# no parameter and  wiht return value

class calculator:
    def __init__(self):
        self.brand="casio"
        self.cost=200

    def add(self):
        a=10
        b=25
        c=a+b
        return c
c1=calculator()
print(c1.brand)
print(c1.cost)
res=c1.add()
print(res)
```

### Question 5: Class Method - With Parameter & No Return Value
Write a Python program for: **Class Method - With Parameter & No Return Value**

```python
# with parameter and no return value
class calculator:
    def __init__(self):
        self.brand="casio"
        self.cost=200

    def add(self , a, b):
        c=a+b
        print(c)

c1=calculator()
n1=100
n2=300
c1.add(n1,n2)
```

### Question 6: Class Method - With Parameter & With Return Value
Write a Python program for: **Class Method - With Parameter & With Return Value**

```python
# with parameter and with return value
class calculator:
    def __init__(self):
        self.brand="casio"
        self.cost=200

    def add(self , a, b):
        c=a+b
        return c

c1=calculator()
n1=100
n2=400
res = c1.add(n1,n2)
print(res)
```
