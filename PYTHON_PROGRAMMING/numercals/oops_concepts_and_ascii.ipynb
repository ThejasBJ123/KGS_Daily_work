```python
# convert lower case to upper case using while loop
str=input("Enter the string : ")
newstr=" "
i=0
while(i<=len(str)-1):
    data =str[i]
    ASCII=ord(data)
    if (ASCII >= 97 and ASCII<=122 ):
        Newascii=ASCII-32

        convchar=chr(Newascii)
        newstr=newstr+convchar

    else:
        newstr=newstr+data

    i=i+1
print(newstr)
```

### Question 7
Write a Python program for: **Python Program Implementation 7**

```python
str = input("Enter the string: ")
newstr = ""
i = 0

while i <= len(str) - 1:
    data = str[i]
    ASCII = ord(data)

    # lowercase → uppercase
    if ASCII >= 97 and ASCII <= 122:
        Newascii = ASCII - 32
        convchar = chr(Newascii)
        newstr = newstr + convchar

    # uppercase → lowercase
    elif ASCII >= 65 and ASCII <= 90:
        Newascii = ASCII + 32
        convchar = chr(Newascii)
        newstr = newstr + convchar

    # spaces, numbers, special characters
    else:
        newstr = newstr + data

    i = i + 1

print(newstr)
```

## Theory & Key Concepts Used in the Codes

1. **ASCII Conversion**: 
   - `ord(char)` returns ASCII integer value of a character.
   - `chr(ascii_val)` returns character corresponding to ASCII value.
2. **Case Conversion Algorithm**:
   - Lowercase to Uppercase: `chr(ord(c) - 32)`
   - Uppercase to Lowercase: `chr(ord(c) + 32)`
3. **Loops for String Processing**: Iterating over strings using `while` and `for` loops.
