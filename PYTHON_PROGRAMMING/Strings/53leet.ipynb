```python
def longestPalindrome(self, s):
    lps = ""
    n = len(s)

    for i in range(1, len(s) - 1):
        left, right = i, i + 1

        while left >= 0 and right < n and s[left] == s[right]:

            palindrome = s[left:right + 1]

            maxlen = max(len(palindrome), len(lps))

            if maxlen == len(palindrome):
                lps = palindrome

            left -= 1
            right += 1

    return lps
```
