### String Palindrome:
    A string is said to be a palindrome if the charcter when traversed from forward direction and backward direction is same, irrespective of letter casing, special or spaces


* Constraints:

    - s[i]==> alphabets or spaces 
    - len(s) ==> 1 to 10**5


### WAP to check whether the given string is a palindrome or not 

```python
def filterations(s):
    nstr=""
    for i in s:
        if "A"<= i<="Z":
            nstr += chr(ord(i)+32)
            
        elif "0" <= i <="9" or "a" <= i <="z":
            nstr +=i
        
    return nstr

def palindromestr(s):
    s=filterations(s)
    i,j=0,len(s)-1
    while i<j:
        if s[i]!=s[j]:
            return False

        i += 1
        j -= 1
    
    return True

s=input("Enter the string: ")
flag=palindromestr(s)
if flag:
    print("Palindrome")
else:
    print("NOt a Palindrome")    
```

## Pangrams:
     A string which have all the 26 alphabetic letters for atleast once ignoring the spaces, casing , char digits and special symbols.
