### String Palindrome:
    A string is said to be a palindrome if the charcter when traversed from forward direction and backward direction is same, irrespective of letter casing, special or spaces


* Constraints:

    - s[i]==> alphabets or spaces 
    - len(s) ==> 1 to 10**5


### WAP to check whether the given string is a palindrome or not 

```python
def filterations(s):
    nstr=""
    for i in s:
        if "A"<= i<="Z":
            nstr += chr(ord(i)+32)
            
        elif "0" <= i <="9" or "a" <= i <="z":
            nstr +=i
        
    return nstr

def palindromestr(s):
    s=filterations(s)
    i,j=0,len(s)-1
    while i<j:
        if s[i]!=s[j]:
            return False

        i += 1
        j -= 1
    
    return True

s=input("Enter the string: ")
flag=palindromestr(s)
if flag:
    print("Palindrome")
else:
    print("NOt a Palindrome")    
```

### Anagrams: 
    Given 2 Str's are said to be anagrams if ann only if all the letters along wiht their occurrences are same ignoring casing, spacing char digits and special symbols

* s1 ="seat"      s2 = "E ,1A t s"

    - s --> 1          e --> 1
    - e --> 1          a --> 1
    - a --> 1          t --> 1
    - t --> 1          s --> 1

==> ANGARAMS

```python
def filterations(s):
    nstr = ""

    for i in s:
        if "A" <= i <= "Z":
            nstr += chr(ord(i) + 32)

        elif "a" <= i <= "z":
            nstr += i

    return nstr


def isAnagram(s1, s2):
    s1 = filterations(s1)
    s2 = filterations(s2)

    if len(s1) != len(s2):
        return False

    else:
        dict = {}

        for i in range(0, len(s1)):

            # Addition
            if s1[i] in dict:
                dict[s1[i]] = dict[s1[i]] + 1
            else:
                dict[s1[i]] = 1

            # Subtraction
            if s2[i] in dict:
                dict[s2[i]] = dict[s2[i]] - 1
            else:
                dict[s2[i]] = -1

        print(dict)

        for key, val in dict.items():
            if val != 0:
                return False

        return True


s1 = input("Enter the first string: ")
s2 = input("Enter the second string: ")

flag = isAnagram(s1, s2)

if flag:
    print("Anagram")
else:
    print("Not an Anagram")
```
