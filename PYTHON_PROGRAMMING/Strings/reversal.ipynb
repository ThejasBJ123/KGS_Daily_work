# reversal of strings

```python
# Decrement Logic
def reversal(s):
    nstr=" "
    for i in range(len(s)-1,0-1,-1):
        nstr=nstr +s[i]
    return nstr
    
s=input("Enter the String: ")
print("Original String: ")
res=reversal(s)
print("Reversal Str using Decremnt loop: ",res)

```

```python
# def reversal2(s)
# logic: nstr=nstr+s[i]
# para: nstr, s,i 
# basecon: if i>0:

def reversal2(nstr, s, i):
    if i < 0:
        return nstr

    nstr = nstr + s[i]
    return reversal2(nstr, s, i - 1)

s = input("Enter the String: ")
res = reversal2(" ", s, len(s) - 1)
print(res)
    
```

```python
# increment loop 
def increemntLoop(s):
    nstr=""
    for i in range(0,len(s)):
        nstr=s[i]+nstr
    return nstr

s=input("Enter the String: ")
print("Original String: ",s)
res=increemntLoop(s)
print("Reversal Str using Decremnt loop: ",res)
```

* INCREMENT REC CYCLE:

```python
# incremeting recusive cycle
def icrementrec(s,nstr,i):
    if i==len(s):
        return nstr
    
    nstr=s[i]+nstr
    return icrementrec(s, nstr, i+1)

s=input("Enter the number: ")
rec=icrementrec(s," ", 0)
print(rec)
```

```python
def reversalswap(s):
    arr = list(s)
    i, j = 0, len(arr) - 1

    while i < j:
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1

    return "".join(arr)


s = input("Enter the string: ")
res = reversalswap(s)
print(res)
```
