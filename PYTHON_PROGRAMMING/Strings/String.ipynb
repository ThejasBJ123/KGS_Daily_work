# write a program to convert all the lower case charcters present in the given string into uppercase
eg: 
        s="AP  76ple"     --> "AP76PLE"

```python
# WAP to convert all the lowercase characters
# present in the given string into uppercase

# eg: s = "AP 76ple"  --->  "AP76PLE"

def toUpperCase(s):
    l1 = list(s)

    for i in range(0, len(l1)):
        if 'a' <= l1[i] <= 'z':
            l1[i] = chr(ord(l1[i]) - 32)

    # Conversion of list to string
    nstr = ""
    for i in l1:
        if i != " ":
            nstr = nstr + i

    return nstr


s = input("Enter the string: ")

print("Original String:", s)
print("Uppercase String:", toUpperCase(s))
```

# write a program to switch the character cases present in the given string.
eg:  
        s="ApPLE"         ==> "aPple"


```python
# WAP to switch the character cases
# present in the given string.

# eg: s = "ApPLE"  ==>  "aPple"

def switchCase(s):
    l1 = list(s)

    for i in range(0, len(l1)):
        if 'A' <= l1[i] <= 'Z':
            l1[i] = chr(ord(l1[i]) + 32)

        elif 'a' <= l1[i] <= 'z':
            l1[i] = chr(ord(l1[i]) - 32)

    # Conversion of list to string
    nstr = ""
    for i in l1:
        nstr = nstr + i

    return nstr


s = input("Enter the string: ")

print("Original String:", s)
print("Switched String:", switchCase(s))
```

# WAP to return the sum of all the individual charcter digits in a given string.
eg: s="1ba8a99l1l2"   output is ball30

    

```python
def sumDigits(s):
    nstr = ""
    num = 0

    for i in range(0, len(s)):
        if '0' <= s[i] <= '9':
            num = num + (ord(s[i]) - 48)
        else:
            nstr = nstr + s[i]

    return nstr + str(num)


s = input("Enter the String: ")
res = sumDigits(s)
print(res)
```

```python
def stringsInt(s):
    num = 0

    for i in range(len(s)):
        if '0' <= s[i] <= '9':
            num = (num * 10) + (ord(s[i]) - 48)

    return num


s = input("Enter the String: ")
res = stringsInt(s)
print(res)
```
