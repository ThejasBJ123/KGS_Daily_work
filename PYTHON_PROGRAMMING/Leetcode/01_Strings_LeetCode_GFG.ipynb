## 📌 Concepts & Methods Used

| Method | How it Works |
|--------|--------------|
| **Two-Pointer Swap** | `i=0, j=len-1` → swap → `i++, j--` until `i>=j` |
| **Increment-Prepend Loop** | `nwrd = s[i] + nwrd` (builds reverse as we go forward) |
| **Decrement Loop** | `range(len-1, -1, -1)` traverse from end |
| **Word boundary trick** | Add trailing space `s = s+' '`, collect word until space |
| **filteration()** | ASCII: `'A'<='Z'` → `+32` (uppercase to lowercase) |
| **Dictionary frequency** | `if key in dict: dict[key]+=1  else: dict[key]=1` |

## Q1. LeetCode 344 — Reverse String
🔗 https://leetcode.com/problems/reverse-string/

### Problem:
Write a function that reverses an array of characters **in-place**.

```
Input  : s = ['h','e','l','l','o']
Output : ['o','l','l','e','h']
```

### Method: **Two-Pointer Swap**
```
i → starts from LEFT  (index 0)
j → starts from RIGHT (index len-1)
swap s[i] and s[j], then i++, j--
stop when i >= j
```

### 📝 Simple Tracing:
```
s = ['h', 'e', 'l', 'l', 'o']
       0    1    2    3    4

Step 1: i=0, j=4  → swap s[0] & s[4]  → ['o','e','l','l','h']
Step 2: i=1, j=3  → swap s[1] & s[3]  → ['o','l','l','e','h']
Step 3: i=2, j=2  → i >= j → STOP

Result: ['o','l','l','e','h'] ✅
```

````python
# Q1. LeetCode 344 - Reverse String
# Method: Two-Pointer Swap

def reverseString(s):
    i, j = 0, len(s) - 1        # i = left pointer, j = right pointer
    
    while i < j:                 # loop until pointers meet
        s[i], s[j] = s[j], s[i] # swap characters
        i += 1                   # move left pointer forward
        j -= 1                   # move right pointer backward
    
    return s

# Test
s = list("hello")
print("Input  :", list("hello"))
print("Output :", reverseString(s))
````

## Q2. LeetCode 151 — Reverse Words in a String
🔗 https://leetcode.com/problems/reverse-words-in-a-string/

### Problem:
Reverse the **ORDER** of words. Extra spaces should be removed.
```
Input  : "  the sky is  blue  "
Output : "blue is sky the"
```

### Method: **Word boundary loop + Prepend trick**
```
- Add trailing space so every word ends with a space
- Collect word char by char
- When space is hit: PREPEND collected word to sentence
  (prepending = reversing the word order)
```

### 📝 Simple Tracing:
```
s = "the sky is"
After adding space: s = "the sky is "

i=0  → 't' ≠ ' '  → nwrd = '' + 't' = 't'
i=1  → 'h' ≠ ' '  → nwrd = 't' + 'h' = 'th'
i=2  → 'e' ≠ ' '  → nwrd = 'th' + 'e' = 'the'
i=3  → ' '        → nsen='' → nsen = 'the'     nwrd=''
i=4  → 's'        → nwrd = 's'
i=5  → 'k'        → nwrd = 'sk'
i=6  → 'y'        → nwrd = 'sky'
i=7  → ' '        → nsen = 'sky' + ' ' + 'the' = 'sky the'   nwrd=''
i=8  → 'i'        → nwrd = 'i'
i=9  → 's'        → nwrd = 'is'
i=10 → ' '        → nsen = 'is' + ' ' + 'sky the' = 'is sky the'  nwrd=''

Result: "is sky the"  ✅
```

````python
# Q2. LeetCode 151 - Reverse Words in a String
# Method: Word boundary loop + Prepend to reverse word order

def reverseWords(s):
    s = s + " "        # add trailing space so last word gets processed
    nsen = ""           # new sentence (word order will be reversed)
    nwrd = ""           # current word being built

    for i in range(0, len(s)):
        if s[i] != " ":               # character found
            nwrd = nwrd + s[i]        # collect word
        elif nwrd != "":              # space found and word exists
            if nsen == "":            # first word
                nsen = nwrd
            else:
                nsen = nwrd + " " + nsen  # PREPEND → reverses order
            nwrd = ""                 # reset word
    return nsen

# Test
s = "  the sky is  blue  "
print("Input  :", s)
print("Output :", reverseWords(s))
````

## Q3. GFG — Reverse Individual Words
🔗 https://www.geeksforgeeks.org/reverse-individual-words/

### Problem:
Reverse each individual word. **Word positions stay the same.**
```
Input  : "Sky is blue"
Output : "ykS si eulb"
```

### Method: **Increment-Prepend Loop** (`nwrd = s[i] + nwrd`)
```
- Each character is PREPENDED to nwrd
- This automatically builds each word in REVERSE
- Append reversed word to sentence (maintain order)
```

### 📝 Simple Tracing:
```
s = "Sky is blue"

i=0  → 'S' → nwrd = 'S' + '' = 'S'
i=1  → 'k' → nwrd = 'k' + 'S' = 'kS'
i=2  → 'y' → nwrd = 'y' + 'kS' = 'ykS'
i=3  → ' ' → nsen = 'ykS'   nwrd = ''
i=4  → 'i' → nwrd = 'i' + '' = 'i'
i=5  → 's' → nwrd = 's' + 'i' = 'si'
i=6  → ' ' → nsen = 'ykS' + ' ' + 'si' = 'ykS si'  nwrd = ''
i=7  → 'b' → nwrd = 'b'
i=8  → 'l' → nwrd = 'l' + 'b' = 'lb'
i=9  → 'u' → nwrd = 'u' + 'lb' = 'ulb'
i=10 → 'e' → nwrd = 'e' + 'ulb' = 'eulb'
i=11 → ' ' → nsen = 'ykS si' + ' ' + 'eulb' = 'ykS si eulb'

Result: "ykS si eulb"  ✅
```

````python
# Q3. GFG - Reverse Individual Words
# Method: Increment-Prepend Loop  (nwrd = s[i] + nwrd)

def reverseIndividualWords(s):
    s = s + " "      # trailing space for last word
    nsen = ""         # final sentence
    nwrd = ""         # current word

    for i in range(0, len(s)):
        if s[i] != " ":
            nwrd = s[i] + nwrd    # PREPEND → builds word in reverse
        elif nwrd != "":
            if nsen == "":
                nsen = nwrd
            else:
                nsen = nsen + " " + nwrd   # APPEND → maintains word order
            nwrd = ""
    return nsen

# Test
s = "Sky is blue"
print("Input  :", s)
print("Output :", reverseIndividualWords(s))
````

## Q4. GFG — Check if a String is Palindrome
🔗 https://www.geeksforgeeks.org/check-if-a-string-is-palindrome-or-not/

### Problem:
Check if a string reads the same forwards and backwards (case-insensitive).
```
Input  : "madam"  →  True
Input  : "hello"  →  False
```

### Method: **filteration() + Two-Pointer**
```
Step 1 - filteration(): convert uppercase → lowercase using ASCII
         'A'...'Z' → ord(ch)+32 → 'a'...'z'
Step 2 - Two-Pointer: compare s[i] and s[j]
         if s[i] != s[j] → NOT palindrome
         i++, j-- until i >= j → IS palindrome
```

### 📝 Simple Tracing:
```
s = "Madam"
After filteration → s = "madam"

i=0, j=4 → s[0]='m', s[4]='m' → match → i=1, j=3
i=1, j=3 → s[1]='a', s[3]='a' → match → i=2, j=2
i=2, j=2 → i >= j → STOP

Result: True (Palindrome) ✅
```

````python
# Q4. GFG - Check if a String is Palindrome
# Method: filteration() + Two-Pointer

def filteration(s):
    """Keep only alphanumeric characters, convert uppercase to lowercase."""
    nstr = ""
    for i in s:
        if "A" <= i <= "Z":                       # uppercase letter
            nstr += chr(ord(i) + 32)              # convert to lowercase using ASCII (+32)
        elif "a" <= i <= "z" or "0" <= i <= "9": # lowercase or digit
            nstr += i
    return nstr

def isPalindrome_str(s):
    s = filteration(s)         # Step 1: clean the string
    i, j = 0, len(s) - 1      # Step 2: set two pointers
    
    while i < j:               # Step 3: compare from both ends
        if s[i] != s[j]:
            return False       # mismatch → not palindrome
        i += 1
        j -= 1
    return True                # all matched → palindrome

# Test
print("madam →", isPalindrome_str("madam"))    # True
print("hello →", isPalindrome_str("hello"))    # False
print("Racecar →", isPalindrome_str("Racecar"))  # True
````

## Q5. LeetCode 125 — Valid Palindrome
🔗 https://leetcode.com/problems/valid-palindrome/

### Problem:
After removing all non-alphanumeric characters and converting to lowercase, check if it's a palindrome.
```
Input  : "A man, a plan, a canal: Panama"
Output : True
```

### Method: **filteration() + Two-Pointer** (same as Q4)
```
filteration removes: spaces, commas, colons, special chars
keeps only: letters + digits, all lowercase
Then Two-Pointer check
```

### 📝 Simple Tracing:
```
s = "A man, a plan, a canal: Panama"
After filteration → "amanaplanacanalpanama"

i=0 'a', j=19 'a' → match
i=1 'm', j=18 'm' → match
i=2 'a', j=17 'a' → match
... all match ...

Result: True ✅
```

````python
# Q5. LeetCode 125 - Valid Palindrome
# Method: filteration() + Two-Pointer (reuses filteration from Q4)

def isValidPalindrome(s):
    s = filteration(s)         # remove special chars, make lowercase
    i, j = 0, len(s) - 1
    
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1
        j -= 1
    return True

# Test
s1 = "A man, a plan, a canal: Panama"
s2 = "race a car"
print("Input  :", s1)
print("Output :", isValidPalindrome(s1))   # True
print()
print("Input  :", s2)
print("Output :", isValidPalindrome(s2))   # False
````

## Q6. GFG — Count Vowels in a String
🔗 https://www.geeksforgeeks.org/program-count-vowels-string/

### Problem:
Count the number of vowels (a, e, i, o, u) in a string.
```
Input  : "Hello World"   Output : 3
```

### Method: **ASCII case conversion + vowel check loop**

### 📝 Simple Tracing:
```
s = "Hello"

i=0 → 'H' → uppercase → ch='h' → 'h' is vowel? NO
i=1 → 'e' → lowercase  → ch='e' → 'e' is vowel? YES → count=1
i=2 → 'l' → not vowel
i=3 → 'l' → not vowel
i=4 → 'o' → 'o' is vowel? YES → count=2

Result: 2
```

````python
# Q6. GFG - Count Vowels in a String
# Method: ASCII case conversion + vowel check

def countVowels(s):
    count = 0
    
    for i in range(0, len(s)):
        ch = s[i]
        if "A" <= ch <= "Z":              # convert uppercase to lowercase
            ch = chr(ord(ch) + 32)
        if ch == "a" or ch == "e" or ch == "i" or ch == "o" or ch == "u":
            count += 1                    # it's a vowel
    
    return count

# Test
s = "Hello World"
print("Input  :", s)
print("Vowels :", countVowels(s))   # 3 (e, o, o)
````

## Q7. GFG — Count Digits and Characters
🔗 https://www.geeksforgeeks.org/count-number-digits-characters-string/

### Problem:
Count digits and alphabetic characters separately.
```
Input  : "Hello123World"   Output : Digits=3, Letters=10
```

### Method: **ASCII range comparison** (`'0'<= ch <='9'`)

### 📝 Simple Tracing:
```
s = "H1e2"

i=0 → 'H' → 'A'<='H'<='Z' → letters=1
i=1 → '1' → '0'<='1'<='9' → digits=1
i=2 → 'e' → 'a'<='e'<='z' → letters=2
i=3 → '2' → '0'<='2'<='9' → digits=2
```

````python
# Q7. GFG - Count Digits and Characters
# Method: ASCII range comparison

def countDigitsAndChars(s):
    digits  = 0
    letters = 0
    
    for i in range(0, len(s)):
        ch = s[i]
        if "0" <= ch <= "9":                         # digit check
            digits += 1
        elif ("a" <= ch <= "z") or ("A" <= ch <= "Z"): # letter check
            letters += 1
    
    return digits, letters

# Test
s = "Hello123World"
d, l = countDigitsAndChars(s)
print("Input   :", s)
print("Digits  :", d)
print("Letters :", l)
````

## Q8. GFG — Count Special Characters
🔗 https://www.geeksforgeeks.org/count-special-characters-in-a-string/

### Problem:
Count characters that are not letters, not digits, and not spaces.
```
Input  : "Hello@World#2025!"   Output : 3
```

### 📝 Simple Tracing:
```
s = "Hi@!"
i=0 → 'H' → letter → skip
i=1 → 'i' → letter → skip
i=2 → '@' → NOT letter/digit/space → count=1
i=3 → '!' → NOT letter/digit/space → count=2
```

````python
# Q8. GFG - Count Special Characters
# Method: ASCII exclusion check

def countSpecialChars(s):
    count = 0
    
    for i in range(0, len(s)):
        ch = s[i]
        # If NOT a letter, NOT a digit, NOT a space → it's special
        if not (("a" <= ch <= "z") or ("A" <= ch <= "Z") or
                ("0" <= ch <= "9") or ch == " "):
            count += 1
    
    return count

# Test
s = "Hello@World#2025!"
print("Input          :", s)
print("Special Chars  :", countSpecialChars(s))  # 3 (@, #, !)
````

## Q9. GFG — Count Words in a String
🔗 https://www.geeksforgeeks.org/count-words-in-a-given-string/

### Problem:
Count words separated by spaces (ignore extra spaces).
```
Input  : "  Hello World Python  "   Output : 3
```

### 📝 Simple Tracing:
```
s = "hi bye "
i=0 'h' → nwrd='h'
i=1 'i' → nwrd='hi'
i=2 ' ' → nwrd != '' → count=1, nwrd=''
i=3 'b' → nwrd='b'
i=4 'y' → nwrd='by'
i=5 'e' → nwrd='bye'
i=6 ' ' → count=2, nwrd=''

```

````python
# Q9. GFG - Count Words in a String
# Method: Word-boundary detection with trailing-space trick

def countWords(s):
    s = s + " "   # trailing space ensures last word is counted
    count = 0
    nwrd  = ""
    
    for i in range(0, len(s)):
        if s[i] != " ":
            nwrd = nwrd + s[i]   # collect word
        elif nwrd != "":         # space found + word exists
            count += 1           # count the word
            nwrd = ""            # reset for next word
    
    return count

# Test
s = "  Hello World Python  "
print("Input :", s)
print("Words :", countWords(s))   # 3
````

## Q10. GFG — Toggle Case
🔗 https://www.geeksforgeeks.org/toggle-case-string-using-bitwise-operators/

### Problem:
Toggle: uppercase → lowercase, lowercase → uppercase.
```
Input  : "ApPLE"   Output : "aPple"
```

### Method: **ASCII ±32** (same as switchCase() in class)
```
Uppercase 'A'(65) + 32 = 97 = 'a'
Lowercase 'a'(97) - 32 = 65 = 'A'
```

### 📝 Simple Tracing:
```
s = "ApPLE"
l1 = ['A','p','P','L','E']

i=0 'A' → uppercase → 'A'+32 = 'a'  → l1=['a','p','P','L','E']
i=1 'p' → lowercase → 'p'-32 = 'P'  → l1=['a','P','P','L','E']
i=2 'P' → uppercase → 'P'+32 = 'p'  → l1=['a','P','p','L','E']
i=3 'L' → uppercase → 'L'+32 = 'l'  → l1=['a','P','p','l','E']
i=4 'E' → uppercase → 'E'+32 = 'e'  → l1=['a','P','p','l','e']

Result: "aPple" ✅
```

````python
# Q10. GFG - Toggle Case
# Method: ASCII ord()/chr() ±32  (same as switchCase() from String.ipynb)

def toggleCase(s):
    l1 = list(s)                      # convert string to list (mutable)
    
    for i in range(0, len(l1)):
        if "A" <= l1[i] <= "Z":       # uppercase letter
            l1[i] = chr(ord(l1[i]) + 32)  # → lowercase (+32)
        elif "a" <= l1[i] <= "z":     # lowercase letter
            l1[i] = chr(ord(l1[i]) - 32)  # → uppercase (-32)
    
    nstr = ""                         # convert list back to string
    for i in l1:
        nstr = nstr + i
    return nstr

# Test
s = "ApPLE"
print("Input  :", s)
print("Output :", toggleCase(s))   # aPple
````

## Q11. GFG — Convert Uppercase ↔ Lowercase
🔗 https://www.geeksforgeeks.org/convert-uppercase-to-lowercase-and-vice-versa/

### Problem:
```
Input  : "AP 76ple"   Output : "ap 76PLE"
```
*(Same logic as Q10 — toggle case)*

````python
# Q11. GFG - Convert Uppercase to Lowercase and Vice Versa
# Method: ASCII ord()/chr() ±32  (toUpperCase + tolowerCase from String.ipynb)

def convertCase(s):
    l1 = list(s)
    for i in range(0, len(l1)):
        if "A" <= l1[i] <= "Z":
            l1[i] = chr(ord(l1[i]) + 32)   # uppercase → lowercase
        elif "a" <= l1[i] <= "z":
            l1[i] = chr(ord(l1[i]) - 32)   # lowercase → uppercase
    nstr = ""
    for i in l1:
        nstr = nstr + i
    return nstr

# Test
s = "AP 76ple"
print("Input  :", s)
print("Output :", convertCase(s))   # ap 76PLE
````

## Q12. GFG — Frequency of Characters
🔗 https://www.geeksforgeeks.org/frequency-of-characters-in-a-string/

### Problem:
Count frequency of each character.
```
Input  : "hello"   Output : {'h':1, 'e':1, 'l':2, 'o':1}
```

### Method: **Dictionary frequency counting**
```
if char already in dict  →  dict[char] = dict[char] + 1
else                     →  dict[char] = 1
```

### 📝 Simple Tracing:
```
s = "hello"
freq = {}

i=0 'h' → not in freq → freq={'h':1}
i=1 'e' → not in freq → freq={'h':1, 'e':1}
i=2 'l' → not in freq → freq={'h':1, 'e':1, 'l':1}
i=3 'l' → IN freq     → freq={'h':1, 'e':1, 'l':2}
i=4 'o' → not in freq → freq={'h':1, 'e':1, 'l':2, 'o':1}

Result: {'h':1, 'e':1, 'l':2, 'o':1} ✅
```

````python
# Q12. GFG - Frequency of Characters
# Method: Dictionary frequency counting (same as isPangram() in class)

def charFrequency(s):
    freq = {}                          # empty dictionary
    
    for i in range(0, len(s)):
        if s[i] in freq:               # character already seen
            freq[s[i]] = freq[s[i]] + 1  # increment count
        else:                          # character seen first time
            freq[s[i]] = 1             # set count to 1
    
    return freq

# Test
s = "hello"
print("Input     :", s)
print("Frequency :", charFrequency(s))
````

## Q13. GFG — Remove Duplicates from a String
🔗 https://www.geeksforgeeks.org/remove-duplicates-from-a-given-string/

### Problem:
Remove duplicate characters, keeping first occurrence only.
```
Input  : "geeksforgeeks"   Output : "geksfor"
```

### 📝 Simple Tracing:
```
s = "geeks"
seen={}, nstr=''

i=0 'g' → not in seen → nstr='g',  seen={'g':1}
i=1 'e' → not in seen → nstr='ge', seen={'g':1,'e':1}
i=2 'e' → IN seen     → skip
i=3 'k' → not in seen → nstr='gek'
i=4 's' → not in seen → nstr='geks'

Result: "geks"
```

````python
# Q13. GFG - Remove Duplicates from String
# Method: Dictionary seen-tracker

def removeDuplicates_str(s):
    seen = {}    # tracks characters already added
    nstr = ""    # result string
    
    for i in range(0, len(s)):
        if s[i] not in seen:           # first occurrence
            nstr = nstr + s[i]         # add to result
            seen[s[i]] = 1             # mark as seen
        # else: already seen, skip
    
    return nstr

# Test
s = "geeksforgeeks"
print("Input  :", s)
print("Output :", removeDuplicates_str(s))   # geksfor
````

## Q14. LeetCode 387 — First Unique Character
🔗 https://leetcode.com/problems/first-unique-character-in-a-string/

### Problem:
Find the index of the first non-repeating character. Return -1 if none.
```
Input  : "leetcode"   Output : 0  (l appears only once)
Input  : "aabb"       Output : -1
```

### 📝 Simple Tracing:
```
s = "leetcode"
freq = {'l':1, 'e':3, 't':1, 'c':1, 'o':1, 'd':1}  ← Step 1

Step 2: scan left to right
i=0 's[0]'='l' → freq['l']=1 → UNIQUE! → return 0
```

````python
# Q14. LeetCode 387 - First Unique Character
# Method: Dictionary frequency (2 passes)

def firstUniqueChar(s):
    freq = {}                          # Step 1: count all characters
    for i in range(0, len(s)):
        if s[i] in freq:
            freq[s[i]] = freq[s[i]] + 1
        else:
            freq[s[i]] = 1
    
    for i in range(0, len(s)):         # Step 2: find first with freq=1
        if freq[s[i]] == 1:
            return i                   # return its index
    
    return -1                          # no unique char

# Test
print("'leetcode' →", firstUniqueChar("leetcode"))   # 0
print("'aabb'     →", firstUniqueChar("aabb"))       # -1
print("'loveleet' →", firstUniqueChar("loveleet"))   # 2
````

## Q15. GFG — First Repeated Character
🔗 https://www.geeksforgeeks.org/find-the-first-repeated-character-in-a-string/

### Problem:
Find the first character that appears more than once.
```
Input  : "geeksforgeeks"   Output : 'e'
```

### 📝 Simple Tracing:
```
s = "geeks"
seen={}

i=0 'g' → not in seen → seen={'g':1}
i=1 'e' → not in seen → seen={'g':1,'e':1}
i=2 'e' → IN seen → return 'e'  ✅
```

````python
# Q15. GFG - First Repeated Character
# Method: Dictionary seen-tracker (single pass)

def firstRepeatedChar(s):
    seen = {}                     # tracks already seen characters
    
    for i in range(0, len(s)):
        if s[i] in seen:          # character seen before
            return s[i]           # this is the first repeated char
        else:
            seen[s[i]] = 1        # mark as seen
    
    return None                   # no repeated character

# Test
print("'geeksforgeeks' →", firstRepeatedChar("geeksforgeeks"))  # e
print("'abcde'         →", firstRepeatedChar("abcde"))          # None
````
