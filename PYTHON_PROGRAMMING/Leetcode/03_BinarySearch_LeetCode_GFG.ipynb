## 📌 Core Method: Binary Search

```
start = 0
end   = len(arr) - 1

while start <= end:
    mid = (start + end) // 2
    
    if target == arr[mid]:
        return mid
    elif target < arr[mid]:
        end = mid - 1       ← search LEFT half
    else:
        start = mid + 1     ← search RIGHT half

return -1   (not found)
```

**Time Complexity:** O(log n) — much faster than linear search O(n)

## Q35. LeetCode 35 — Search Insert Position
🔗 https://leetcode.com/problems/search-insert-position/description/

### Problem:
Return index of target in sorted array; if not found, return where it **should be inserted**.
```
Input  : nums=[1,3,5,6], target=5   Output : 2
Input  : nums=[1,3,5,6], target=2   Output : 1
Input  : nums=[1,3,5,6], target=7   Output : 4
```

### Method: **Binary Search** (start pointer = insert position at end)

### 📝 Simple Tracing  (nums=[1,3,5,6], target=2):
```
start=0, end=3

Cycle 1: mid=(0+3)//2=1  → arr[1]=3
         2 < 3 → end = 1-1 = 0

Cycle 2: mid=(0+0)//2=0  → arr[0]=1
         2 > 1 → start = 0+1 = 1

start=1 > end=0 → loop ends
return start = 1  ✅  (insert at index 1)
```

````python
# Q35. LeetCode 35 - Search Insert Position
# Method: Binary Search (from 09_binary_search.ipynb)

def searchInsert(nums, target):
    start, end = 0, len(nums) - 1
    
    while start <= end:
        mid = (start + end) // 2        # find middle index
        
        if target == nums[mid]:         # found!
            return mid
        elif target < nums[mid]:        # target in LEFT half
            end = mid - 1
        else:                           # target in RIGHT half
            start = mid + 1
    
    return start                        # start = correct insert position

# Test
print("target=5 →", searchInsert([1, 3, 5, 6], 5))  # 2 (found)
print("target=2 →", searchInsert([1, 3, 5, 6], 2))  # 1 (insert)
print("target=7 →", searchInsert([1, 3, 5, 6], 7))  # 4 (insert at end)
````

## Q36. LeetCode 34 — First and Last Position of Element
🔗 https://leetcode.com/problems/find-first-and-last-position-of-element-in-sorted-array/

### Problem:
Find starting and ending index of target.
```
Input  : nums=[5,7,7,8,8,10], target=8   Output : [3,4]
Input  : target=6                         Output : [-1,-1]
```

### Method: **Binary Search twice**
```
findFirst:  when found → result=mid, then end=mid-1  (keep searching LEFT)
findLast:   when found → result=mid, then start=mid+1 (keep searching RIGHT)
```

### 📝 Simple Tracing  (nums=[5,7,7,8,8,10], target=8):
```
findFirst: start=0, end=5
  mid=2 → arr[2]=7 < 8  → start=3
  mid=4 → arr[4]=8 = 8  → result=4, end=3
  mid=3 → arr[3]=8 = 8  → result=3, end=2
  start > end → return result=3

findLast: start=0, end=5
  mid=2 → arr[2]=7 < 8  → start=3
  mid=4 → arr[4]=8 = 8  → result=4, start=5
  mid=5 → arr[5]=10 > 8 → end=4
  start > end → return result=4

Result: [3, 4] ✅
```

````python
# Q36. LeetCode 34 - First and Last Position
# Method: Binary Search twice (bias left for first, bias right for last)

def findFirst(nums, target):
    start, end, result = 0, len(nums) - 1, -1
    while start <= end:
        mid = (start + end) // 2
        if target == nums[mid]:
            result = mid
            end = mid - 1       # keep searching LEFT for earlier occurrence
        elif target < nums[mid]:
            end = mid - 1
        else:
            start = mid + 1
    return result

def findLast(nums, target):
    start, end, result = 0, len(nums) - 1, -1
    while start <= end:
        mid = (start + end) // 2
        if target == nums[mid]:
            result = mid
            start = mid + 1     # keep searching RIGHT for later occurrence
        elif target < nums[mid]:
            end = mid - 1
        else:
            start = mid + 1
    return result

def searchRange(nums, target):
    return [findFirst(nums, target), findLast(nums, target)]

# Test
print("target=8 →", searchRange([5, 7, 7, 8, 8, 10], 8))  # [3,4]
print("target=6 →", searchRange([5, 7, 7, 8, 8, 10], 6))  # [-1,-1]
````

## Q37. GFG — Floor in a Sorted Array
🔗 https://www.geeksforgeeks.org/problems/floor-in-a-sorted-array-1587115620/1

### Problem:
Find the **largest element ≤ x** (floor value).
```
Input  : arr=[1,2,8,10,12,19], x=5   Output : 2  (largest ≤ 5)
```

### 📝 Simple Tracing  (arr=[1,2,8,10,12,19], x=5):
```
start=0, end=5, floor_idx=-1

mid=2 → arr[2]=8 > 5  → end=1
mid=0 → arr[0]=1 < 5  → floor_idx=0, start=1
mid=1 → arr[1]=2 < 5  → floor_idx=1, start=2
start > end → return floor_idx=1

arr[1] = 2  ✅
```

````python
# Q37. GFG - Floor in a Sorted Array
# Method: Binary Search with floor tracking

def floorInSortedArray(arr, x):
    start, end = 0, len(arr) - 1
    floor_idx  = -1
    
    while start <= end:
        mid = (start + end) // 2
        
        if arr[mid] == x:
            return mid                  # exact match IS the floor
        elif arr[mid] < x:
            floor_idx = mid             # potential floor (largest so far ≤ x)
            start = mid + 1            # search RIGHT for bigger floor
        else:
            end = mid - 1              # arr[mid] > x → search LEFT
    
    return floor_idx

# Test
arr = [1, 2, 8, 10, 12, 19]
fi  = floorInSortedArray(arr, 5)
print("Floor of 5  :", arr[fi] if fi != -1 else -1)   # 2
fi2 = floorInSortedArray(arr, 20)
print("Floor of 20 :", arr[fi2] if fi2 != -1 else -1) # 19
````

## Q38. GFG — Ceil in a Sorted Array
🔗 https://www.geeksforgeeks.org/problems/ceil-the-floor2802/1

### Problem:
Find the **smallest element ≥ x** (ceil value).
```
Input  : arr=[1,2,8,10,12,19], x=5   Output : 8  (smallest ≥ 5)
```

### 📝 Simple Tracing:
```
start=0, end=5, ceil_idx=-1

mid=2 → arr[2]=8 > 5  → ceil_idx=2, end=1
mid=0 → arr[0]=1 < 5  → start=1
mid=1 → arr[1]=2 < 5  → start=2
start > end → return ceil_idx=2

arr[2] = 8  ✅
```

````python
# Q38. GFG - Ceil in a Sorted Array
# Method: Binary Search with ceil tracking

def ceilInSortedArray(arr, x):
    start, end = 0, len(arr) - 1
    ceil_idx   = -1
    
    while start <= end:
        mid = (start + end) // 2
        
        if arr[mid] == x:
            return mid                  # exact match IS the ceil
        elif arr[mid] > x:
            ceil_idx = mid             # potential ceil (smallest so far ≥ x)
            end = mid - 1             # search LEFT for smaller ceil
        else:
            start = mid + 1           # arr[mid] < x → search RIGHT
    
    return ceil_idx

# Test
arr = [1, 2, 8, 10, 12, 19]
ci  = ceilInSortedArray(arr, 5)
print("Ceil of 5  :", arr[ci] if ci != -1 else -1)   # 8
ci2 = ceilInSortedArray(arr, 20)
print("Ceil of 20 :", arr[ci2] if ci2 != -1 else -1) # -1 (no ceil)
````

## Q39. LeetCode 33 — Search in Rotated Sorted Array
🔗 https://leetcode.com/problems/search-in-rotated-sorted-array/

### Problem:
Search target in a **rotated** sorted array (no duplicates).
```
Input  : nums=[4,5,6,7,0,1,2], target=0   Output : 4
Input  : target=3                          Output : -1
```

### Method: **Modified Binary Search — find which half is sorted**
```
At each step:
  if nums[start] <= nums[mid]  →  LEFT half is sorted
     if target in [nums[start], nums[mid])  →  search LEFT
     else  →  search RIGHT
  else  →  RIGHT half is sorted
     if target in (nums[mid], nums[end]]  →  search RIGHT
     else  →  search LEFT
```

### 📝 Simple Tracing  (nums=[4,5,6,7,0,1,2], target=0):
```
start=0, end=6

mid=3  arr[3]=7
nums[0]=4 <= nums[3]=7 → LEFT half [4,5,6,7] is sorted
Is 0 in [4,7)? NO → search RIGHT: start=4

mid=5  arr[5]=1
nums[4]=0 <= nums[5]=1 → LEFT half [0,1] is sorted
Is 0 in [0,1)? YES → search LEFT: end=4

mid=4  arr[4]=0 = 0 → return 4 ✅
```

````python
# Q39. LeetCode 33 - Search in Rotated Sorted Array
# Method: Modified Binary Search (identify sorted half)

def searchRotatedArray(nums, target):
    start, end = 0, len(nums) - 1
    
    while start <= end:
        mid = (start + end) // 2
        
        if nums[mid] == target:                 # found
            return mid
        
        # Check which half is sorted
        if nums[start] <= nums[mid]:            # LEFT half is sorted
            if nums[start] <= target < nums[mid]:
                end = mid - 1                   # target in left → go left
            else:
                start = mid + 1                 # target in right → go right
        else:                                   # RIGHT half is sorted
            if nums[mid] < target <= nums[end]:
                start = mid + 1                 # target in right → go right
            else:
                end = mid - 1                   # target in left → go left
    
    return -1

# Test
print("target=0 →", searchRotatedArray([4, 5, 6, 7, 0, 1, 2], 0))  # 4
print("target=3 →", searchRotatedArray([4, 5, 6, 7, 0, 1, 2], 3))  # -1
````

## Q40. LeetCode 153 — Find Minimum in Rotated Sorted Array
🔗 https://leetcode.com/problems/find-minimum-in-rotated-sorted-array/

### Problem:
Find the minimum element in a rotated sorted array.
```
Input  : [3,4,5,1,2]   Output : 1
Input  : [4,5,6,7,0,1,2]   Output : 0
```

### Method: **Binary Search — minimum is at rotation point**
```
if nums[mid] > nums[end]  → minimum is in RIGHT half
else                      → minimum is in LEFT half (including mid)
```

### 📝 Simple Tracing  (nums=[3,4,5,1,2]):
```
start=0, end=4

mid=2  nums[2]=5 > nums[4]=2 → min in right → start=3
mid=3  nums[3]=1 <= nums[4]=2 → min in left (incl mid) → end=3
start=3 = end=3 → return nums[3]=1 ✅
```

````python
# Q40. LeetCode 153 - Find Minimum in Rotated Sorted Array
# Method: Binary Search - move towards smaller half

def findMin(nums):
    start, end = 0, len(nums) - 1
    
    while start < end:
        mid = (start + end) // 2
        
        if nums[mid] > nums[end]:      # mid is in the LEFT (bigger) portion
            start = mid + 1            # minimum must be in RIGHT half
        else:                          # mid is in the RIGHT (smaller) portion
            end = mid                  # minimum is in LEFT half including mid
    
    return nums[start]                 # start == end == minimum

# Test
print("[3,4,5,1,2]     →", findMin([3, 4, 5, 1, 2]))      # 1
print("[4,5,6,7,0,1,2] →", findMin([4, 5, 6, 7, 0, 1, 2])) # 0
print("[1]             →", findMin([1]))                    # 1
````

## Q41. LeetCode 162 — Find Peak Element
🔗 https://leetcode.com/problems/find-peak-element/

### Problem:
A peak element is **greater than its neighbors**. Return any peak **index**.
```
Input  : [1,2,3,1]   Output : 2  (nums[2]=3 > nums[1]=2 and nums[3]=1)
Input  : [1,2,1,3,5,6,4]   Output : 5
```

### Method: **Binary Search — move towards higher neighbor**
```
if nums[mid] > nums[mid+1]  → peak is on LEFT side (including mid)
else                         → peak is on RIGHT side
```

### 📝 Simple Tracing  (nums=[1,2,3,1]):
```
start=0, end=3

mid=1  nums[1]=2, nums[2]=3
2 > 3? NO → peak on right → start=2

mid=2  nums[2]=3, nums[3]=1
3 > 1? YES → peak on left including mid → end=2

start=2 = end=2 → return 2 ✅
```

````python
# Q41. LeetCode 162 - Find Peak Element
# Method: Binary Search - move towards higher neighbor

def findPeakElement(nums):
    start, end = 0, len(nums) - 1
    
    while start < end:
        mid = (start + end) // 2
        
        if nums[mid] > nums[mid + 1]:  # descending → peak on LEFT (including mid)
            end = mid
        else:                          # ascending → peak on RIGHT
            start = mid + 1
    
    return start                       # start == end == peak index

# Test
print("[1,2,3,1]       →", findPeakElement([1, 2, 3, 1]))         # 2
print("[1,2,1,3,5,6,4] →", findPeakElement([1, 2, 1, 3, 5, 6, 4])) # 5
````

## Q42. LeetCode 81 — Search in Rotated Sorted Array II (with duplicates)
🔗 https://leetcode.com/problems/search-in-rotated-sorted-array-ii/description/

### Problem:
Same as Q39 but array **may contain duplicates**.
```
Input  : nums=[2,5,6,0,0,1,2], target=0   Output : True
Input  : target=3                          Output : False
```

### Method: **Modified Binary Search + Duplicate handling**
```
Extra case: when nums[start] == nums[mid] == nums[end]
→ Cannot determine which half is sorted
→ Shrink both ends: start++, end--
```

### 📝 Simple Tracing  (nums=[1,1,1,0,1], target=0):
```
start=0, end=4
mid=2  nums[0]=1=nums[2]=1=nums[4]=1 → duplicates! start=1, end=3

mid=2  nums[1]=1 <= nums[2]=1 → LEFT sorted [1,1]
Is 0 in [1,1)? NO → start=3

mid=3  nums[3]=0 == target=0 → True ✅
```

````python
# Q42. LeetCode 81 - Search in Rotated Sorted Array II
# Method: Modified Binary Search with duplicate handling

def searchRotatedArrayII(nums, target):
    start, end = 0, len(nums) - 1
    
    while start <= end:
        mid = (start + end) // 2
        
        if nums[mid] == target:                       # found
            return True
        
        # Handle duplicates: cannot determine sorted half
        if nums[start] == nums[mid] == nums[end]:
            start += 1
            end   -= 1
        
        elif nums[start] <= nums[mid]:                # LEFT half is sorted
            if nums[start] <= target < nums[mid]:
                end = mid - 1
            else:
                start = mid + 1
        
        else:                                         # RIGHT half is sorted
            if nums[mid] < target <= nums[end]:
                start = mid + 1
            else:
                end = mid - 1
    
    return False

# Test
print("[2,5,6,0,0,1,2] target=0 →", searchRotatedArrayII([2, 5, 6, 0, 0, 1, 2], 0))  # True
print("[2,5,6,0,0,1,2] target=3 →", searchRotatedArrayII([2, 5, 6, 0, 0, 1, 2], 3))  # False
````
