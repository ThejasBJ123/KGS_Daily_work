## 1. STAR PATTERNS
The star-pattern section is on pages 1–3 of the assignment.

### Star Pattern 1 – Star Square

```python
n = int(input("Enter the number: "))

for i in range(n):
    for j in range(n):
        print("*", end=" ")
    print()
```

### Star Pattern 2 – Hollow Square

```python
n = int(input("Enter the number: "))

for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

### Star Pattern 3 – Right Shifted Square

```python
n = int(input("Enter the number: "))

for i in range(n):
    for k in range(i):
        print("  ", end="")
    for j in range(n):
        print("*", end=" ")
    print()
```

### Star Pattern 4 – Hollow Slant Shape

```python
n = int(input("Enter the number: "))

for i in range(n):
    for k in range(i):
        print("  ", end="")
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

### Star Pattern 5 – Left Shifted Square

```python
n = int(input("Enter the number: "))

for i in range(n):
    for k in range(n-i-1):
        print("  ", end="")
    for j in range(n):
        print("*", end=" ")
    print()
```

### Star Pattern 6 – Inverted Hollow Square

```python
n = int(input("Enter the number: "))

for i in range(n):
    for k in range(n-i-1):
        print("  ", end="")
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

### Star Pattern 7 – Star Pyramid

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for k in range(n-i):
        print("  ", end="")
    for j in range(2*i-1):
        print("*", end=" ")
    print()
```

### Star Pattern 8 – Inverted Star Pyramid

```python
n = int(input("Enter the number: "))

for i in range(n):
    for k in range(i):
        print("  ", end="")
    for j in range(2*(n-i)-1):
        print("*", end=" ")
    print()
```

### Star Pattern 9 – Hollow Star Pyramid

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for k in range(n-i):
        print("  ", end="")
    for j in range(2*i-1):
        if j == 0 or j == 2*i-2 or i == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

## 2. NUMBER PATTERNS
The number-pattern section begins on page 4. fileciteturn0file0L65-L82

### Number Pattern 1 – Increasing and Decreasing

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(1, i+1):
        print(j, end=" ")
    print()

for i in range(n-1, 0, -1):
    for j in range(1, i+1):
        print(j, end=" ")
    print()
```

### Number Pattern 2 – 5 4 3 2 1 Up and Down

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(n, n-i, -1):
        print(j, end=" ")
    print()

for i in range(n-1, 0, -1):
    for j in range(n, n-i, -1):
        print(j, end=" ")
    print()
```

### Number Pattern 3 – Continuous Numbers

```python
n = int(input("Enter the number: "))

num = 1

for i in range(1, n+1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

### Number Pattern 4 – Pascal Triangle

```python
n = int(input("Enter the number: "))

for i in range(n):
    value = 1
    for j in range(i+1):
        print(value, end=" ")
        value = value * (i-j) // (j+1)
    print()
```

### Number Pattern 5 – Number Palindrome

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(1, i+1):
        print(j, end=" ")
    for j in range(i-1, 0, -1):
        print(j, end=" ")
    print()
```

### Number Pattern 6 – Number Pyramid Up and Down

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for k in range(n-i):
        print("  ", end="")
    for j in range(1, i+1):
        print(j, end=" ")
    print()

for i in range(n-1, 0, -1):
    for k in range(n-i):
        print("  ", end="")
    for j in range(1, i+1):
        print(j, end=" ")
    print()
```

### Number Pattern 7 – 1 0 Alternating

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(1, i+1):
        if j % 2 == 1:
            print(1, end=" ")
        else:
            print(0, end=" ")
    print()
```

### Number Pattern 8 – Repeated Row Number

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(n):
        if i == 1:
            print(1, end=" ")
        elif j < i-2:
            print(1, end=" ")
        else:
            print(i, end=" ")
    print()
```

### Number Pattern 9 – 5 4 3 2 1 Decreasing Length

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(i, n+1):
        print(j, end=" ")
    for j in range(n-1, i-1, -1):
        print(j, end=" ")
    print()
```

### Number Pattern 10 – Multiplication Triangle

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(1, i+1):
        print(i*j, end=" ")
    print()
```

## 3. ALPHABET PATTERNS
The alphabet-pattern section is on pages 7–10. fileciteturn0file0L135-L147

### Alphabet Pattern 1 – A, AB, ABC...

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(i):
        print(chr(65+j), end=" ")
    print()
```

### Alphabet Pattern 2 – E, ED, EDC...

```python
n = int(input("Enter the number: "))

for i in range(1, n+1):
    for j in range(i):
        print(chr(69-j), end=" ")
    print()
```

### Alphabet Pattern 3 – A B C D E F Down and Up

```python
n = int(input("Enter the number: "))

for i in range(n, 0, -1):
    for j in range(i):
        print(chr(65+j), end=" ")
    print()

for i in range(2, n+1):
    for j in range(i):
        print(chr(65+j), end=" ")
    print()
```

### Alphabet Pattern 4 – Shifted Alphabet Rows

```python
n = int(input("Enter the number: "))

for i in range(n):
    for k in range(i):
        print("  ", end="")
    for j in range(n):
        print(chr(65+j), end=" ")
    print()
```

### Alphabet Pattern 5 – Alphabet Number-Position Pattern

```python
n = int(input("Enter the number: "))

for i in range(n):
    for j in range(i+1):
        # Position-based alphabet pattern
        pos = i + j
        print(chr(65 + pos), end=" ")
    print()
```

### Alphabet Pattern 6 – Alphabet Reverse Triangle

```python
n = int(input("Enter the number: "))

for i in range(n, 0, -1):
    for j in range(i):
        print(chr(65+j), end=" ")
    print()
```

### Alphabet Pattern 7 – A J K T U Pattern

```python
# This reproduces the 5 x 5 alphabet arrangement shown in the PDF.
letters = [
    "A J K T U",
    "B I L S V",
    "C H M R W",
    "D G N Q X",
    "E F O P Y"
]

for row in letters:
    print(row)
```

### Alphabet Pattern 8 – Alphabet Palindrome

```python
n = int(input("Enter the number: "))

for i in range(n, 0, -1):
    for j in range(i):
        print(chr(65+j), end="")
    for j in range(i-1, -1, -1):
        print(chr(65+j), end="")
    print()
```

### Alphabet Pattern 9 – ABCDE / BCDEF / CDEFG

```python
n = int(input("Enter the number: "))

for i in range(n):
    for j in range(n):
        print(chr(65+i+j), end="")
    print()
```

### Alphabet Pattern 10 – A / CB / DEF / JIHG / KLMNO

```python
# Reproduces the final alphabet pattern shown in the PDF.
rows = [
    "A",
    "C B",
    "D E F",
    "J I H G",
    "K L M N O"
]

for row in rows:
    print(row)
```
