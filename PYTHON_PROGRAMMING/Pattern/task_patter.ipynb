# SECTION 1: STAR PATTERNS

### Question 1: Solid Square Star Pattern
```
* * * * *
* * * * *
* * * * *
* * * * *
* * * * *
```

````python
# Solid Square Pattern
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    for j in range(1, n + 1):
        print("*", end=" ")
    print()
````

### Question 2: Square with Diagonals (X inside Hollow Box)
```
* * * * *
* *   * *
*   *   *
* *   * *
* * * * *
```

````python
# Square with Diagonals
n = int(input("Enter the number: "))
if n % 2 == 0:
    n += 1
for i in range(1, n + 1):
    for j in range(1, n + 1):
        if i == 1 or j == 1 or i == n or j == n or i == j or (i + j) == (n + 1):
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
````

### Question 3: Right-Shifted Parallelogram Star Pattern
```

      * * * * *
        * * * * *
```

### Question 7: Solid Diamond Star Pattern
```
    *
   * *
  * * *
 * * * *
* * * * *
 * * * *
  * * *
   * *
    *
```

````python
# Solid Diamond Star Pattern
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for k in range(n, noc, -1):
        print(" ", end="")
    for j in range(1, noc + 1):
        print("* ", end="")
    print()
    if i < n:
        noc += 1
    else:
        noc -= 1
````

### Question 8: Hollow Diamond Star Pattern
```

  *   *
   * *
    *
```

````python
# Hollow Diamond Star Pattern
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for k in range(n, noc, -1):
        print(" ", end="")
    for j in range(1, noc + 1):
        if j == 1 or j == noc:
            print("* ", end="")
        else:
            print("  ", end="")
    print()
    if i < n:
        noc += 1
    else:
        noc -= 1
````

### Question 9: Butterfly Star Pattern 1 (Hollow Center)
```
*       *
**     **
***   ***
* * * * *
***   ***
**     **
*       *
```

````python
# Butterfly Star Pattern 1
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for j in range(1, noc + 1):
        print("*", end="")
    for k in range(1, 2 * (n - noc) + 1):
        print(" ", end="")
    for j in range(1, noc + 1):
        print("*", end="")
    print()
    if i < n:
        noc += 1
    else:
        noc -= 1
````

### Question 10: Solid Butterfly Star Pattern
```
*      *
**    **
***  ***
********
***  ***
**    **
*      *
```

````python
# Solid Butterfly Star Pattern
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print("*", end="")
    for k in range(1, 2 * (n - i) + 1):
        print(" ", end="")
    for j in range(1, i + 1):
        print("*", end="")
    print()
for i in range(n - 1, 0, -1):
    for j in range(1, i + 1):
        print("*", end="")
    for k in range(1, 2 * (n - i) + 1):
        print(" ", end="")
    for j in range(1, i + 1):
        print("*", end="")
    print()
````

### Question 11: Right Triangle Half-Diamond Number Pattern (1 to 5 to 1)
```
1

1 2 3 4
1 2 3 4 5
1 2 3 4
1 2 3
1 2

```

````python
# Half-Diamond Number Pattern
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for j in range(1, noc + 1):
        print(j, end=" ")
    print()
    if i < n:
        noc += 1
    else:
        noc -= 1
````

### Question 12: Right-Aligned Reverse Half-Diamond Number Pattern (5 to 1 to 5)
```
        5
      5 4
    5 4 3
  5 4 3 2
5 4 3 2 1
  5 4 3 2
    5 4 3
      5 4
        5
```

````python
# Right-Aligned Reverse Half-Diamond
n = int(input("Enter the number: "))
noc = 1
for i in range(1, n * 2):
    for k in range(n, noc, -1):
        print(" ", end=" ")
    for j in range(n, n - noc, -1):
        print(j, end=" ")
    print()
    if i < n:
        noc += 1
    else:
        noc -= 1
````

### Question 13: Floyd's Triangle Number Pattern
```

11 12 13 14 15
```

````python
# Floyd's Triangle
n = int(input("Enter the number: "))
num = 1
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(num, end=" ")
        num += 1
    print()
````

### Question 14: Pascal's Triangle Number Pattern
```
    1
   1 1
  1 2 1
 1 3 3 1
1 4 6 4 1
```

````python
# Pascal's Triangle
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    for k in range(n, i, -1):
        print(" ", end="")
    val = 1
    for j in range(1, i + 1):
        print(val, end=" ")
        val = val * (i - j) // j+1
    print()
````

### Question 18: Grid Boundary Repeating Number Pattern
```
1 1 1 1 1
1 1 2 2 2
1 1 3 3 3
1 4 4 4 4
5 5 5 5 5
```

````python
# Grid Boundary Repeating Number Pattern
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    num_ones = 5 if i == 1 else (2 if i in (2, 3) else (1 if i == 4 else 0))
    for j in range(1, n + 1):
        if j <= num_ones:
            print(1, end=" ")
        else:
            print(i, end=" ")
    print()
````

### Question 25: Inverted Bottom-to-Top Alphabet Triangle Pattern
```
J
H I
E F G
A B C D
```

````python
# Inverted Bottom-to-Top Alphabet Triangle
n = int(input("Enter the number: "))
for i in range(1, n + 1):
    total = n * (n + 1) // 2
    sum_prev = (i - 1) * i // 2
    start_val = total - sum_prev - (i - 1)
    for j in range(i):
        print(chr(64 + start_val + j), end=" ")
    print()
````

### Question 26: Full Vertical Snake Grid Alphabet Pattern (5x5)
```
A J K T U
B I L S V
C H M R W
D G N Q X
E F O P Y
```

````python
# Full Vertical Snake Grid Alphabet Pattern
n = int(input("Enter the number: "))
grid = [[""] * n for _ in range(n)]
val = 1
for col in range(n):
    if col % 2 == 0:
        for row in range(n):
            grid[row][col] = chr(64 + val)
            val += 1
    else:
        for row in range(n - 1, -1, -1):
            grid[row][col] = chr(64 + val)
            val += 1

for row in range(n):
    for col in range(n):
        print(grid[row][col], end=" ")
    print()
````

### Question 27: Palindromic Hollow Alphabet Diamond Pattern
```
ABCDEFEDCBA
ABCDE EDCBA
ABCD   DCBA
ABC     CBA
AB       BA
A         A
AB       BA
ABC     CBA
ABCD   DCBA
ABCDE EDCBA
ABCDEFEDCBA
```

````python
# Palindromic Hollow Alphabet Diamond
n = int(input("Enter the number: "))
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(chr(64 + j), end="")
    if i < n:
        print(" " * (2 * (n - i) - 1), end="")
        for j in range(i, 0, -1):
            print(chr(64 + j), end="")
    else:
        for j in range(i - 1, 0, -1):
            print(chr(64 + j), end="")
    print()

for i in range(2, n + 1):
    for j in range(1, i + 1):
        print(chr(64 + j), end="")
    if i < n:
        print(" " * (2 * (n - i) - 1), end="")
        for j in range(i, 0, -1):
            print(chr(64 + j), end="")
    else:
        for j in range(i - 1, 0, -1):
            print(chr(64 + j), end="")
    print()
````

### Question 29: Alternating Boustrophedon Alphabet Triangle Pattern
```
A
C B
D E F
J I H G
K L M N O
```

````python
# Alternating Boustrophedon Alphabet Triangle
n = int(input("Enter the number: "))
val = 1
for i in range(1, n + 1):
    if i % 2 != 0:
        for j in range(val, val + i):
            print(chr(64 + j), end=" ")
        val += i
    else:
        for j in range(val + i - 1, val - 1, -1):
            print(chr(64 + j), end=" ")
        val += i
    print()
````
