# **Dictionary**:

- It is a data structure that stores the data in key - value format
- It is an iterable object
- Each pair of key - val is called as item
- It is also called as collection of items 
- len() --> It return the count of num of element's present in an iterable object 

### Key:
    - It is a unique element that represents each item individually
    - If we try to duplicate an existing key then the existing key's val will get update 
    - Mutable objects are not allowed to be taken as a key
    - The predefined method to display all the key are ==> dict.keys(), 
            it returns all the keys of the calling dictinory of type class 'dict_keys'
    
### Values:
    - It can be duplicated
    - It can be of any datatype
    - The predefined method to display all the keys are ==> dict.values(),
    it returns all the keys of the calling dictionary of type class 'dict_values'
- to get all the items of the calling dictinary ==> dict.items()

### Note: 
* In programming dict is used:
    * when it is required to maintain the count of number of occurrences of any element
    * when it is required to store the data in key - value format



```python
dict1= {101:"abc", 102: "bat", 103:"abc", 104:"adsf", 105:"adfa"}
print(dict1)
print(len(dict1.keys()))
```

### 1. WAP to segregate all the ele's present ain a given array based on duplicates, non- duplicates and unique ele's 
    * duplicates : the repeated ele's 
    * non - duplicates : the ele's that occurs atleast once
    * unique: The ele's which are occuring exactly once 

    eg= arr= [ 3,45, 67, 3,45,1,3]

    duplicates = [3,45]
    unique = [1,67]
    non-duplicates = [3,1,45,67]

    

```python
def createIntarray():
    l1=[]
    while True:
        try:
            n=int(input())
            l1.append(n)
        except Exception as e:
            return l1
        
def seggregation(arr):
    dict={}
    for i in range(0,len(arr)):
        if arr[i] in dict:
            dict[arr[i]]=dict[arr[i]] + 1
        else:
            dict[arr[i]]=1

    # print(dict)
    dup,nodup,Unique=[],[],[]
    
    # Traversal through dictionary
    for key, val in dict.items():
        if val>1:
            dup.append(key)
        if val==1:
            Unique.append(key)
        nodup.append(key)
    return dup, nodup, Unique


print("Enter the elements into the array: ")
arr=createIntarray()
print("Created array: ", arr)

resDup, resNonDup, resunique =seggregation(arr)
print("Duplicates:", resDup)  

print("Non-Duplicates: ", resNonDup)
print("Unique:", resunique)
        
```
