# **Find Largest Element in an Array**

## **Question:**
Write a Python program to create a user-defined integer array and find the **largest element** present in the given array along with its **index**.


```python
def createIntArray():
    l1 = []
    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)
        except Exception as e:
            return l1

def findMaxElement(arr):
    if not arr:
        return None, -1
    maxEle, maxInd = -2**31, -1
    for i in range(0, len(arr)):
        if arr[i] > maxEle:
            maxEle = arr[i]
            maxInd = i
    return maxEle, maxInd

# Main program execution
print("Enter the elements into the array to be created:")
arr = createIntArray()
print("The created array:", arr)

resEle, resInd = findMaxElement(arr)
print("The largest element is:", resEle, "found at index:", resInd)
```
