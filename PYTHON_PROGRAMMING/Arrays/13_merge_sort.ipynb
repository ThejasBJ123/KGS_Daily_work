# **MergeSort**

- It majorly includes 3 steps
    * Divide the array to it's least from
        * LHS  --> start = remains same, end = prev mid
        * RHS  --> start = (prev min+1),  end = remains same
        
    * Merge the array in sorted order into a 3rd Memory
    * update the sorted val's into the original memory

```python
def createIntArray():
    arr = []
    while True:
        try:
            n = int(input("Enter the val: "))
            arr.append(n)
        except Exception:
            return arr


# Merge two sorted parts
def mergeSortMerge(arr, start, mid, end):

    i = start
    j = mid + 1

    res = []

    # Compare both halves
    for k in range(end - start + 1):
        
        if i <= mid and j <= end:
            if arr[i] < arr[j]:
                res.append(arr[i])
                i += 1

            else:
                res.append(arr[j])
                j += 1
        else:

            if i <= mid:
                res.append(arr[i])
                i += 1

            elif j <= end:
                res.append(arr[j])
                j += 1
    # Update original array
    for k in range(len(res)):
        arr[start] = res[k]
        start += 1


# Divide
def mergeSortDivide(arr, start, end):

    if start >= end:
        return

    mid = (start + end) // 2

    # Left half
    mergeSortDivide(arr, start, mid)

    # Right half
    mergeSortDivide(arr, mid + 1, end)

    # Merge
    mergeSortMerge(arr, start, mid, end)


print("Enter ele's into the array:")
arr = createIntArray()

print("Created Array:", arr)
mergeSortDivide(arr, 0, len(arr) - 1)

print("Sorted Array:", arr)
```
