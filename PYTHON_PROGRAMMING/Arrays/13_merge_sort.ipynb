# **MergeSort**

- It majorly includes 3 steps
    * Divide the array to it's least from
        * LHS  --> start = remains same, end = prev mid
        * RHS  --> start = (prev min+1),  end = remains same
        
    * Merge the array in sorted order into a 3rd Memory
    * update the sorted val's into the original memory

```python
def createIntArray():
    arr = []
    while True:
        try:
            n = int(input("Enter the val: "))
            arr.append(n)
        except Exception:
            return arr


# Merge two sorted parts
def mergeSortMerge(arr, start, mid, end):

    i = start
    j = mid + 1

    res = []

    # Compare both halves
    for k in range(end - start + 1):
        
        if i <= mid and j <= end:
            if arr[i] < arr[j]:
                res.append(arr[i])
                i += 1

            else:
                res.append(arr[j])
                j += 1
        else:

            if i <= mid: 
                res.append(arr[i])
                i += 1

            elif j <= end:
                res.append(arr[j])
                j += 1
    # Update original array
    for k in range(len(res)):
        arr[start] = res[k]
        start += 1


# Divide
def mergeSortDivide(arr, start, end):

    if start >= end:
        return

    mid = (start + end) // 2

    # Left half
    mergeSortDivide(arr, start, mid)

    # Right half
    mergeSortDivide(arr, mid + 1, end)

    # Merge
    mergeSortMerge(arr, start, mid, end)


print("Enter ele's into the array:")
arr = createIntArray()

print("Created Array:", arr)
mergeSortDivide(arr, 0, len(arr) - 1)

print("Sorted Array:", arr)
```

# write a program to merge to given to 2 ASC Sorted array into DESC array

arr1 = [1,3,5,5,7,8,10,11]
arr2 = [-87, -87, ,10, -7, -1, 0, 1, 1, 4]

res = [ 11, 10,8,7, 5,5,4,3,1,1,1,....-87]



```python
def creatarray():
    l1 = []
    while True:
        try:
            n = int(input())
            l1.append(n)
        except Exception as e:
            return l1


def merge(arr1, arr2):

    i = len(arr1) - 1
    j = len(arr2) - 1

    res = []

    for k in range(0, len(arr1) + len(arr2)):

        if i >= 0 and j >= 0:

            if arr1[i] > arr2[j]:
                res.append(arr1[i])
                i -= 1

            else:
                res.append(arr2[j])
                j -= 1

        else:

            if i >= 0:
                res.append(arr1[i])
                i -= 1

            elif j >= 0:
                res.append(arr2[j])
                j -= 1

    return res


arr1 = creatarray()
print("created array 1", arr1)

arr2 = creatarray()
print("created array 2", arr2)

res = merge(arr1, arr2)

print("DESC array", res)
```

### Write a program to merge the given 2 DESC sorted array into ASC
eg: 
    created array 1 [98, 97, 65, 64, 11]
    created array 2 [88, 84, 83, 21, 9, 1]
    ASC array: [1, 9, 11, 21, 64, 65, 83, 84, 88, 97, 98]

```python
def creatarray():
    l1 = []

    while True:
        try:
            n = int(input())
            l1.append(n)
        except Exception:
            return l1


def merge(arr1, arr2):

    i = len(arr1) - 1
    j = len(arr2) - 1

    res = []

    for k in range(0, len(arr1) + len(arr2)):

        if i >= 0 and j >= 0:

            if arr1[i] < arr2[j]:
                res.append(arr1[i])
                i -= 1

            else:
                res.append(arr2[j])
                j -= 1

        else:

            if i >= 0:
                res.append(arr1[i])
                i -= 1

            elif j >= 0:
                res.append(arr2[j])
                j -= 1

    return res


arr1 = creatarray()
print("created array 1", arr1)

arr2 = creatarray()
print("created array 2", arr2)

res = merge(arr1, arr2)

print("ASC array:", res)
```

### Write a program to merge the given 2 DESC sorted array into DESC arry

eg: 
    arr1 = [ 98,97,65,64,32,11]
    arr2 = [ 88, 84, 83, 21, 1]

    res= [98, 97, 88,84,83,65,64,32,21,11,1]
    

```python
def creatarray():
    l1 = []
    while True:
        try:
            n = int(input())
            l1.append(n)
        except Exception as e:
            return l1


def merge(arr1, arr2):

    i = 0
    j = 0

    res = []

    for k in range(0, len(arr1) + len(arr2)):

        if i < len(arr1) and j < len(arr2):

            if arr1[i] > arr2[j]:
                res.append(arr1[i])
                i += 1

            else:
                res.append(arr2[j])
                j += 1

        else:

            if i < len(arr1):
                res.append(arr1[i])
                i += 1

            elif j < len(arr2):
                res.append(arr2[j])
                j += 1

    return res

arr1 = creatarray()
print("created array 1", arr1)

arr2 = creatarray()
print("created array 2", arr2)

res = merge(arr1, arr2)

print("DESC array", res)
```
