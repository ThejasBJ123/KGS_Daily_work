```python
def createInArray():
    l1 = []

    while True:
        try:
            num = int(input("Enter the number: "))
            l1.append(num)

        except Exception as e:
            return l1


def linearSearch(arr, target):
    for i in range(0, len(arr)):
        if target == arr[i]:
            return i, True

    return -1, False


print("Enter the elements in the array:")
arr = createInArray()

print("The created array is:", arr)

target = int(input("Enter the element to search: "))

resInd, flag = linearSearch(arr, target)

if flag:
    print(target, "element is found at index", resInd)
else:
    print(target, "element is not found")
```

### WAP  to sort a given array using Selection sort
* ascending order
* decending order 

```python
# WAP to sort a given array using Selection Sort in ascending order

def createInArray():
    l1 = []

    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)

        except Exception as e:
            return l1

def selectionsortasc(arr):
    n = len(arr)
    for i in range(0, n - 1):
        actualInd = n - 1 - i
        currMaxEle, currMaxInd = -2**31, -1
        for j in range(0, n - i):
            if arr[j] > currMaxEle:
                currMaxEle = arr[j]
                currMaxInd = j
        arr[actualInd], arr[currMaxInd] = arr[currMaxInd], arr[actualInd]

    return arr

# Main program
arr = createInArray()
print("Original Array:", arr)

selectionsortasc(arr)
print("Sorted Array:", arr)
```

```python
# WAP to sort a given array using Selection Sort in descending order



def selectionsortdesc(arr):
    n = len(arr)

    for i in range(0, n - 1):
        actualInd = n - 1 - i

        currMinEle, currMinInd = 2**31, -1

        for j in range(0, n - i):
            if arr[j] < currMinEle:
                currMinEle = arr[j]
                currMinInd = j

        arr[actualInd], arr[currMinInd] = arr[currMinInd], arr[actualInd]

    return arr


# Main program
arr = createInArray()

print("Original Array:", arr)

selectionsortdesc(arr)

print("Sorted Array:", arr)
```

### wap to sort the given array using **BUBBLE SORT**
- ascending order 
- Decending order

```python
# bubble sort wiht ascending order 
def createIntArray():
    l1 = []

    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)

        except Exception as e:
            return l1


def Bubblesort(arr):
    n = len(arr)

    for i in range(0, n - 1):
        for j in range(0, n - 1 - i):

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr


print("Enter the ele's into the created array:")

arr = createIntArray()

print("The created array:", arr)

res = Bubblesort(arr)

print("Bubble sort:", res)
```

### wap to sort the given array using **Insertion SORT**
- ascending order 
- Decending order

```python
# insertion sort in the ascending order 
def createIntArray():
    l1 = []

    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)

        except Exception as e:
            return l1

def insertionSort(arr):
    n = len(arr)

    for i in range(0, n - 1):
        for j in range(i+1, 0,-1):
            if arr[j-1]>arr[j]:
                arr[j-1],arr[j] =arr[j],arr[j-1]
    return arr


print("Enter the ele's into the created array:")
arr = createIntArray()
print("The created array:", arr)

res = insertionSort(arr)
print("insertion sort : ",res)
```

```python
# Descending Order

def createIntArray():
    l = []
    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except Exception as e:
            return l


def binarySearchDESC(arr, target):
    start, end = 0, len(arr) - 1

    while start <= end:
        mid = (start + end) // 2

        if target == arr[mid]:
            return mid

        # For descending order
        if target < arr[mid]:
            start = mid + 1
        else:
            end = mid - 1

    return -1


print("Enter the elements into the array in descending order:")

arr = createIntArray()

target = int(input("Enter the element to be searched: "))

print("The created array:", arr)

resInd = binarySearchDESC(arr, target)

if resInd != -1:
    print(target, "is found at index:", resInd)
else:
    print(target, "is not found")
```

### there are 2 major problem in this current code

1. as we are restricting the user to enter only a sorted arry, the user can enter , either an ASC or DESC sorted array 
2. multiple function that are performing the same task

solu for 1st problem:




```python
# Order Agnostic Binary Search

def createIntArray():
    l = []

    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except Exception as e:
            return l


def orderAgnosticBinarySearch(arr, target):

    start, end = 0, len(arr) - 1

    if arr[start] < arr[end]:
        flag = "asc"
    else:
        flag = "desc"

    while start <= end:

        mid = (start + end) // 2

        if target == arr[mid]:
            return mid

        if flag == "asc":

            if target < arr[mid]:
                end = mid - 1
            else:
                start = mid + 1

        else:  # descending

            if target < arr[mid]:
                start = mid + 1
            else:
                end = mid - 1

    return -1


print("Enter the elements into the array:")
arr = createIntArray()

target = int(input("Enter the element to be searched: "))

print("The created array:", arr)

resInd = orderAgnosticBinarySearch(arr, target)

if resInd != -1:
    print(target, "is found at index:", resInd)
else:
    print(target, "is not found")
```

### Write a program to revers a array 

```python
# logic 1
def createIntArray():
    l = []

    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except Exception as e:
            return l


def reverseArray(arr):
    res = []

    for i in range(len(arr) - 1, -1, -1):
        res.append(arr[i])

    return res


print("Enter the elements into the array:")
arr = createIntArray()

print("The original array:", arr)

res = reverseArray(arr)

print("The reversed array:", res)
```
