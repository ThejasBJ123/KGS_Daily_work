# **Bubble Sort**


## **Question 1: Bubble Sort in Ascending Order**
Write a Python program to sort a given user-defined integer array using **Bubble Sort** in **ascending order**.


```python
def createIntArray():
    l1 = []
    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)
        except Exception as e:
            return l1

def bubbleSortAsc(arr):
    n = len(arr)
    for i in range(0, n - 1):
        for j in range(0, n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# Main program execution
print("Enter the elements into the created array:")
arr = createIntArray()
print("The created array:", arr)

res = bubbleSortAsc(arr)
print("Bubble Sort (Ascending):", res)
```

## **Question 2: Bubble Sort in Descending Order**
Write a Python program to sort a given user-defined integer array using **Bubble Sort** in **descending order**.


```python
def bubbleSortDesc(arr):
    n = len(arr)
    for i in range(0, n - 1):
        for j in range(0, n - 1 - i):
            if arr[j] < arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

# Main program execution
print("Enter the elements into the created array:")
arr = createIntArray()
print("The created array:", arr)

res = bubbleSortDesc(arr)
print("Bubble Sort (Descending):", res)
```
