# **Selection Sort**


## **Question 1: Selection Sort in Ascending Order**
Write a Python program to sort a given user-defined integer array using **Selection Sort** in **ascending order**.


```python
def createIntArray():
    l1 = []
    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)
        except Exception as e:
            return l1

def selectionSortAsc(arr):
    n = len(arr)
    for i in range(0, n - 1):
        actualInd = n - 1 - i
        currMaxEle, currMaxInd = -2**31, -1
        for j in range(0, n - i):
            if arr[j] > currMaxEle:
                currMaxEle = arr[j]
                currMaxInd = j
        arr[actualInd], arr[currMaxInd] = arr[currMaxInd], arr[actualInd]
    return arr

# Main program execution
arr = createIntArray()
print("Original Array:", arr)

selectionSortAsc(arr)
print("Sorted Array (Ascending):", arr)
```

## **Question 2: Selection Sort in Descending Order**
Write a Python program to sort a given user-defined integer array using **Selection Sort** in **descending order**.


```python
def selectionSortDesc(arr):
    n = len(arr)
    for i in range(0, n - 1):
        actualInd = n - 1 - i
        currMinEle, currMinInd = 2**31, -1
        for j in range(0, n - i):
            if arr[j] < currMinEle:
                currMinEle = arr[j]
                currMinInd = j
        arr[actualInd], arr[currMinInd] = arr[currMinInd], arr[actualInd]
    return arr

# Main program execution
arr = createIntArray()
print("Original Array:", arr)

selectionSortDesc(arr)
print("Sorted Array (Descending):", arr)
```
