# **Find Second Largest Element in an Array**

## **Question:**
Write a Python program to find the **second largest element** along with its index in a given integer array without sorting.


```python
def createIntArray():
    l1 = []
    while True:
        try:
            n = int(input("Enter the num: "))
            l1.append(n)
        except Exception as e:
            return l1

def findSecondLargestEle(arr):
    maxEle, maxEleInd = -2**31, -1
    secMaxEle, secMaxEleInd = -2**31, -1

    for i in range(0, len(arr)):
        if arr[i] > maxEle:
            secMaxEle, secMaxEleInd = maxEle, maxEleInd
            maxEle, maxEleInd = arr[i], i
        elif arr[i] != maxEle and arr[i] > secMaxEle:
            secMaxEle, secMaxEleInd = arr[i], i

    return [maxEle, maxEleInd, secMaxEle, secMaxEleInd]

# Main program execution
print("Enter the elements into the created array:")
arr = createIntArray()
print("The created array:", arr)

res = findSecondLargestEle(arr)
print("Larger value:", res[0], "at index", res[1])
print("Second largest value:", res[2], "at index:", res[3])
```
