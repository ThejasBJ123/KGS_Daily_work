```python
class Solution:
    def searchInsert(self, n: list[int], tar: int) -> int:
        l,r = 0, len(n) -1

        while(l<=r):

            mid=l+(r-1)//2
            if tar==n[mid]:
                return mid

            elif tar<n[mid]:
            
                r = mid-1
            else:
                l=mid+1
        return l
```

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        n=len(nums)
        for i in range(0,(n-1)):
            for j in range((i+1),n):
                if nums[i] +nums[j]  ==target:
                     return [i,j]
```

```python
# kth element
class Solution:
    def findKthPositive(self, arr: list[int], k: int) -> int:
        for i in range(0, len(arr)):
            if arr[i] <= k:
                k+=1
            
            else:
                break
            
        return k
```
