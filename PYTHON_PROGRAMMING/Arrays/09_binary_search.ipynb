# **Binary Search**

### **Why Binary Search over Linear Search?**
In Linear Search, as the number of elements increases in an array, the time consumed increases linearly $O(n)$.
**Binary Search** is significantly more efficient $O(\log n)$ because it repeatedly divides the search space in half. However, it requires the array to be sorted beforehand.


## **Question 1: Binary Search on Ascending Sorted Array**
Write a Python program to search for a target element in an ascending sorted array using **Binary Search**.


```python
def createIntArray():
    l = []
    while True:
        try:
            n = int(input("Enter num: "))
            l.append(n)
        except Exception as e:
            return l

def binarySearchASC(arr, target):
    start, end = 0, len(arr) - 1
    cycle = 0

    while start <= end:
        cycle += 1
        mid = (start + end) // 2

        if target == arr[mid]:
            print("Cycles required:", cycle)
            return mid

        if target < arr[mid]:
            end = mid - 1
        else:
            start = mid + 1

    print("Cycles required:", cycle)
    return -1

# Main program execution
print("Enter the elements into the array in ascending order:")
arr = createIntArray()

target = int(input("Enter the element to be searched: "))
print("The created array:", arr)

resInd = binarySearchASC(arr, target)
if resInd != -1:
    print(target, "is found at index:", resInd)
else:
    print(target, "is not found")
```

## **Question 2: Binary Search on Descending Sorted Array**
Write a Python program to search for a target element in a descending sorted array using **Binary Search**.


```python
def binarySearchDESC(arr, target):
    start, end = 0, len(arr) - 1

    while start <= end:
        mid = (start + end) // 2

        if target == arr[mid]:
            return mid

        # For descending order
        if target < arr[mid]:
            start = mid + 1
        else:
            end = mid - 1

    return -1

# Main program execution
print("Enter the elements into the array in descending order:")
arr = createIntArray()

target = int(input("Enter the element to be searched: "))
print("The created array:", arr)

resInd = binarySearchDESC(arr, target)
if resInd != -1:
    print(target, "is found at index:", resInd)
else:
    print(target, "is not found")
```

## **Question 3: Order Agnostic Binary Search**
Write a Python program to implement **Order Agnostic Binary Search**, which automatically detects whether the array is sorted in ascending or descending order and searches for the target element accordingly.


```python
def orderAgnosticBinarySearch(arr, target):
    if not arr:
        return -1
        
    start, end = 0, len(arr) - 1

    if arr[start] < arr[end]:
        flag = "asc"
    else:
        flag = "desc"

    while start <= end:
        mid = (start + end) // 2

        if target == arr[mid]:
            return mid

        if flag == "asc":
            if target < arr[mid]:
                end = mid - 1
            else:
                start = mid + 1
        else:  # descending
            if target < arr[mid]:
                start = mid + 1
            else:
                end = mid - 1

    return -1

# Main program execution
print("Enter the elements into the sorted array:")
arr = createIntArray()

target = int(input("Enter the element to be searched: "))
print("The created array:", arr)

resInd = orderAgnosticBinarySearch(arr, target)

if resInd != -1:
    print(target, "is found at index:", resInd)
else:
    print(target, "is not found")
```
