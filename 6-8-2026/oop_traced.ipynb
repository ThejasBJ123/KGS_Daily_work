## Section 1: class Demo:

### ❓ Question / Problem Statement
**Problem Statement (class Demo:): Implement and trace the Python program for class demo:.**

--- 
### 💻 My Code

```python
class Demo:
    x=10
    def __init__(self):
        self.a=100
        
    def instancedisp(self):
        print(self.a)
    
    @staticmethod
    def staticdisp():
        print(Demo.x)
        print("welcome to KGS")
    
    @classmethod
    def classdisp(cls):
        print(Demo.x)

    
  
d1=Demo()
d1.instancedisp()
Demo.staticdisp()
Demo.classdisp() # 
```

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `__init__, instancedisp, staticdisp, classdisp` to encapsulate reusable task logic.
2. **Object-Oriented Programming:** Defines class structure with attributes and methods (`__init__`, static, or instance methods).
