```python
class Demo:
    x=10
    def __init__(self):
        self.a=100
        
    def instancedisp(self):
        print(self.a)
    
    @staticmethod
    def staticdisp():
        print(Demo.x)
        print("welcome to KGS")
    
    @classmethod
    def classdisp(cls):
        print(Demo.x)

    
  
d1=Demo()
d1.instancedisp()
Demo.staticdisp()
Demo.classdisp() # 
```
