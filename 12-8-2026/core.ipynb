```python
class Demo:
    def swap(self , a, b):
        print("befor swapping")
        print(a)
        print(b)

        temp=a
        a=b
        b=temp

        print("after the swapping ")
        print(a)
        print(b)

d1=Demo()
n1=100
n2=200

d1.swap(n1,n2)
```

```python
class Demo:
     def disp(self, x=11, y=22, z=33):
        print(x)
        print(y)
        print(z)

d1=Demo()
a=10
b=20
c=30
d1.disp()

print("\n")
d1.disp(a,b,c)

print("\n")
d1.disp(a,b,c)

print("\n")
d1.disp(a,b)

print("\n")
d1.disp(b,c)

print("\n")
d1.disp(y=b,z=c)

print("\n")
d1.disp(x=a,y=b,z=c)

```
