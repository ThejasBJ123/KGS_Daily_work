## Section 3: write a program to reverse a given integer using recursion function

### ❓ Question / Problem Statement
**Problem Statement (write a program to reverse a given integer using recursion function): Implement and trace the Python program for write a program to reverse a given integer using recursion function.**

--- 
### 💻 My Code

````python
# write a program to reverse a given integer using recursion function

# n=7142    op : 2417

def reverseNum(n,rev):
    if n<=0:
        return rev
    rem=n%10
    rev=(rev*10)+rem
    n=n//10
    return reverseNum(n,rev)

num=int(input("Enter a num : "))
res=reverseNum(num,0)
print("The reversal of " ,num ,"is", res)
````

## Section 4: write a program to palindrome a given integer using recursion function

### ❓ Question / Problem Statement
**Problem Statement (write a program to palindrome a given integer using recursion function): Implement and trace the Python program for write a program to palindrome a given integer using recursion function.**

--- 
### 💻 My Code

````python
# write a program to palindrome a given integer using recursion function

# n=7142    op : 2417

def isIntPalindrome(n, rev, temp):
    if n<=0:
        return rev == temp

    rem=n%10
    rev=(rev*10)+rem
    n=n//10
    return isIntPalindrome(n, rev, temp)

num=int(input("Enter a number : "))
flag= isIntPalindrome(num , 0, num)
if flag:
    print(num," is the palindrome ")
else:
    print(num," is not a palindrome")
````

### 🧠 Logic & Explanation
