### Question 2
Write a Python program for: **count digt using customized function**

```python
# count digt using customized function

def isCountDigit(n):
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    return count

n = int(input("Enter the number: "))
c = isCountDigit(n)
print("Number of digits:",n,"is: ", c)
```

write a program to display the count of digits present each indivial number present in the user defined number 

not : when ever you need ot carry forward the current cycle updates value, to the next cycle use the same operand and RHS for operation and LHS for updation.

```python
def isCountDigit(n):
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    return count

sr = int(input("Enter the starting number: "))
er = int(input("Enter the ending number: "))

if sr > er:
    print("Invalid number")
else:
    for i in range(sr, er + 1):
        c = isCountDigit(i)
        print("The count  fo digit ",i, "is", c, "digits")
```

## Theory & Key Concepts Used in the Codes

1. **Floor Division (`//`)**: `n // 10` removes the rightmost digit of a number.
2. **Modulus (`%`)**: `n % 10` extracts the last digit of a number.
3. **`while` Loop**: Repeats processing as long as `n > 0`.
4. **Count Accumulator**: Increments `count += 1` on each loop iteration to tally total digits.
5. **Functions (`def`)**: Wraps digit counting logic inside custom reusable functions.
