## Section 6: count digt using customized function

### ❓ Question / Problem Statement
**Problem Statement (count digt using customized function): Implement and trace the Python program for count digt using customized function.**

--- 
### 💻 My Code

````python
# count digt using customized function

def isCountDigit(n):
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    return count

n = int(input("Enter the number: "))
c = isCountDigit(n)
print("Number of digits:",n,"is: ", c)
````

### 🧠 Logic & Explanation
1. **Modular Function Design:** Defines custom function(s) `isCountDigit` to encapsulate reusable task logic.
2. **User Input Handling:** Prompts user input and casts values (`int()`, `float()`) to appropriate numeric types.
3. **While Loop Control:** Continues execution while state control condition evaluates to `True`.
4. **Return Statement:** Passes calculated result back to caller function.

### 🔍 Step-by-Step Code Tracing (Dry Run)

#### Sample Execution: `n = 153`

**Step 1: Count Digits (`CountDigit(153)`):**
- `153 // 10 = 15` (1)
- `15 // 10 = 1` (2)
- `1 // 10 = 0` (3)
- `count = 3`

**Step 2: Sum of Power of Digits (`isArmstrong(153)`):**
- Digit 3: `3**3 = 27`
- Digit 5: `5**3 = 125`
- Digit 1: `1**3 = 1`
- `sum = 27 + 125 + 1 = 153`

**Step 3: Comparison (`sum == original`):**
- `153 == 153` is **True**.

**Output:**
