write a program as leap year from the start year and end year of the program

```python
def isLeapYear(year):
    return(year % 100!=0 and year %4==0) or (year % 400==0)


star_year=int(input("ente the start year : "))
end_year=int(input("enter the end year : "))

if star_year>end_year:
    print("INvalid number" )

else:
    print("leap year: ")
    for i in range (star_year,(end_year+1)):
        if isLeapYear(i):
            print(i, end=" ")

    print("\n")

    print("not a leap year: ")
    for i in range (star_year, (end_year+1)):
        if not isLeapYear(i):
            print(i,end=" ")
```

```python
# count digt using customized function

def isCountDigit(n):
    count = 0
    while n > 0:
        n = n // 10
        count += 1
    return count

n = int(input("Enter the number: "))
c = isCountDigit(n)
print("Number of digits:",n,"is: ", c)
```
