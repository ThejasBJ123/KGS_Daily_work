write a code for the leap year

```python
year=int(input("enter the year : "))

if(year % 100!=0 and year %4==0) or (year % 400==0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python
def isLeapYear(year):
    return(year % 100!=0 and year %4==0) or (year % 400==0)


star_year=int(input("ente the start year : "))
end_year=int(input("enter the end year : "))

if star_year>end_year:
    print("INvalid number" )

else:
    print("leap year: ")
    for i in range (star_year,(end_year+1)):
        if isLeapYear(i):
            print(i, end=" ")

    print("\n")

    print("not a leap year: ")
    for i in range (star_year, (end_year+1)):
        if not isLeapYear(i):
            print(i,end=" ")
```

## Theory & Key Concepts Used in the Codes

1. **Leap Year Rules**:
   - Year must be divisible by 4 AND not divisible by 100 (`year % 4 == 0 and year % 100 != 0`), OR
   - Year must be divisible by 400 (`year % 400 == 0`).
2. **Logical Operators (`and`, `or`)**: Combines multiple boolean conditions.
3. **Custom Function (`isLeapYear`)**: Encapsulates year validation logic returning boolean.
4. **Range Loop**: Iterates through `range(start_year, end_year + 1)` to filter leap years.
